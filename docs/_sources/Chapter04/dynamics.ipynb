The study of motion is kinematics, but kinematics only *describes* the way objects move through their velocity and acceleration.  **Dynamics** considers why the objects moves through forces that affect moving objects and systems.  Newton's laws of motion are the foundation of dynamics.

Isaac Newton's laws of motion were only one part of the monumental work that made him legendary.  His insights marks the transition from the Renaissance to the modern era and revolutionizes the way people thought about the physical universe.  For centuries, natural philosophers debated the nature of the universe through rules of logic, where a lot of credit was given to classical philosophers (e.g., Aristotle). Newton and Galileo both contributed a great deal to this change in thinking.

Galileo demonstrated how observation was the absolute determinant of truth, rather than "logical" argument.  Galileo's telescope showed the importance of observation, where he discovered moons orbiting Jupiter and other observations that falsified geocentrism.  Galileo developed a foundational principle, which is now called Newton's first law of motion.  Newton made expanded upon the work of Galileo and Kepler (i.e., his predecessors), which enabled him to develop laws of motion, a law of gravity, the method of calculus, and contributions to theories of light and color.

By the early 20th century, experiments showed that Newton's laws of motion produce a good approximation when the objects are moving at speeds much less than the speed of light.  This is when scientists became capable of probing very small objects (particle sizes about $10^{-9}\ {\rm m}$ or $1\ {\rm nm}$) that could attain such high speeds.  This epoch marks the boundary between classical mechanics and the development of quantum mechanics which applies probability to describing the motion of objects.

## Development of Force Conceptually

Dynamics is the study of the forces that *cause* objects to move.  However, we need a working definition of force.  Our intuitive definition describes *contact* forces that involve a direct push or pull.  A push or pull has both a magnitude and direction, which makes a force a *vector* quantity.  For example, a cannon exerts a strong force on a cannonball.  Our everyday experience also gives us a good idea of how multiple forces add.  If two ice skaters push on a third skater, we expect that the total force to be a *combination* of the individual forces.  

The two forces act as components of a total force since force is a vector.  Additionally, we find that it adds just like other vectors and can be represented by arrow using either the head-to-tail graphical method or the analytical method developed in the previous chapter.



Figure {numref}`{number}<fig-force-skater-diagram>` shows our first example of a **free-body diagram** (FBD), which is a technique used to simplify problems by illustrating all the **external** forces acting on a body using only arrows.  The body is represented by a single isolated point (i.e., free body), where only external (outside) forces acting *on* the body are shown.  We can ignore any internal forces within the body (object) that is acted on.  FBDs are very useful in analyzing forces because they help us externalize the problem on paper instead of trying the solve the whole problem in our head.  Drawing FBDs is not about your artistic ability, where it is a technique that helps make problems easier to solve.

## Newton's First Law of Motion

Experience shows us that an object at rest will remain at rest if left alone.  Additionally, an object in motion tends to slow down and stop unless some effort is made to keep it moving.  This second observation is more complicated by friction and air resistance.  As in the example of an ice skater (i.e., a frictionless surface), the skater keeps moving once in motion until acted upon.  **Newton's first law of motion** describes how objects behave in the absence of external forces, under idealized conditions.

```{admonition} Newton's First Law of Motion
> A body at rest remains, or, if in motion, remains in motion at a constant velocity unless acted on by a net external force.
```

Newton's first law states that there must be a *cause* (which is a net external force) *for there to be any change in velocity (either a change in magnitude or direction).*  A net external force will be defined with Newton's second law.  An object sliding across a table or floor slows down due the the net force of friction acting on the object.  If friction disappeared, *would the object slow down?*

The idea of cause and effect is crucial in accurately describing what happens.  Consider an object sliding along a rough horizontal surface, where it quickly grinds to a halt.  If we spray the surface with talcum powder, then the surface is smoother and the object slides further (e.g., shuffleboard).  If we make the surface even smoother by rubbing oil on it (e.g., a bowling lane), then the object slides farther still.  Extrapolating to a frictionless surface, we can imagine the object sliding in a straight line indefinitely.  Friction is thus *causing* the slowing of the object (consistent with Newton's first law) and is an external force.

Consider an air hockey table.  When the air is turned on, it creates a nearly frictionless surface, and the puck glides long distances without slowing down.  When the air is turned off, the puck slides only a short distance before friction slows it to a stop.  If we know enough about the friction, we can develop a model that accurately predicts how quickly the object will slow down.

Newton's first law is completely general and can be applied to objects sliding on tables, as well as blood pumped from the heart. Experiments have thoroughly verified that any change in velocity (in speed or direction) must be caused by an external force. The idea of universal laws that generally apply to everything is a basic feature of all laws of physics. Identifying these laws came from recognizing patterns in nature. Another important idea was parsimony: a good model should not introduce additional assumptions or causes when they are unnecessary. This principle is often associated with the medieval philosopher William of Ockham and is commonly called *Ockham's razor*. 

Galileo set out to show that Aristotle's proposition that the surrounding air continued to push a projectile after it left the thrower's hand. Galileo could describe the motion of arrows without requiring the air to keep pushing them through their flight.  Newton carried this idea further by treating rest and uniform motion as equivalent states: what requires a cause is not motion itself, but a change in motion.

### Inertia & Mass
The property of a body to remain at rest (or to remain in motion with constant velocity) is called **inertia**.  Newton's first law is often called the **law of inertia**.  Some objects have more inertia than others, which makes them harder to move.  For example, it is more difficult to move a boulder than a basketball.  But we cannot measure inertia directly.

Instead, we measure the **mass** as a proxy for inertia.  Roughly speaking, mass is a measure of the amount of "stuff" (or matter) in something.  The quantity of matter is determined by the number of atoms/molecules that makes up the object.  Mass does not vary with location.  The mass of an object is the same on Earth as it is on the Moon.  It is not practical to count *all* the atoms and molecules in an object, so masses are determined by comparison with a standard kilogram.

In May 2019, the new definition of the kilogram became based on physical constants of the universe (e.g., Planck constant).  The kilogram is now measured using a [Kibble balance](https://en.wikipedia.org/wiki/Kibble_balance), which measures an electrical current produced that is proportional to Planck's constant.  Using the exact definition of the Planck constant, this means that the exact measurement of the current can define the kilogram.

````{admonition} [**Veritasium**](https://www.youtube.com/@Veritasium)
:class: tip
```{youtube} c_e1wITe_ig
:width: 560
:height: 315
:align: center
```
````

````{admonition} Check Your Understanding
:class: etamu-exercise

**The Problem**

> Which has more mass: a kilogram of cotton balls or a kilogram of gold?

```{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Conclusion**

Neither has more mass. Both samples have a mass of exactly $1\ {\rm kg}$. The cotton balls occupy a much larger volume because cotton has a much lower density than gold, but the amount of mass specified in the question is the same.

```

*Adapted from [OpenStax, *College Physics 2e*, Section 4.2](https://openstax.org/books/college-physics-2e/pages/4-2-newtons-first-law-of-motion-inertia).*
````

## Newton's Second Law of Motion

**Newton's second law of motion** mathematically states the cause and effect relationship between force and changes in motion.  Newton's first law indicated that an external force is required to cause motion, where Newton's second law gives us a general mathematical form to describe the external force.  Newton's second law relates the mass and acceleration of an object to the applied force.

Intuitively, we can see that more force needs to be applied to accelerate (move) and object.  Recall the definition of acceleration as the change in *velocity* divided by a time interval.  This means that a change in motion is equivalent to a change in velocity.  Newton's first law says that a net external force causes a change in motion and thus, a *net external force causes acceleration.*

An **external force** acts from outside the **system** (i.e., object, or collection of objects) of interest.  For example, a system can consist of a wagon and a child in it that is being pushed by two other children.  The two children supply the external forces because without them, the wagon and child would not move.  

```{figure-md} fig-system-to-fbd
<img src="FBD_diagram.png" alt="Three-panel sequence showing how to construct a free-body diagram for a wagon carrying a child. Panel a identifies the wagon and child as the system while two people push it. Panel b replaces the system with a simplified wagon and retains the external forces. Panel c replaces the system with a point and shows only the applied forces, friction, normal force, and weight." style="width:95%; height:auto;">

Constructing a free-body diagram begins by choosing the **system**. In (a), the wagon and child are enclosed by the dashed boundary, while the two people pushing the wagon remain outside the system. The applied forces $\vec{F}_1$ and $\vec{F}_2$, friction $\vec{f}$, the normal force $\vec{N}$, and weight $\vec{w}$ are external forces acting on the system. In (b), the physical details are removed while the same external forces are retained. In (c), the system is reduced to a point and only the external-force vectors remain, producing the free-body diagram. Adapted from: [OpenStax, *College Physics 2e*, Figure 4.5](https://openstax.org/books/college-physics-2e/pages/4-3-newtons-second-law-of-motion-concept-of-a-system).
```

Internal forces act between elements of the system.  The force that the child in the wagon applies to hang on is an internal force between elements of the system of interest.  The internally forces cancel through Newton's third law and that gives us reason to ignore them.  Only external forces affect the motion of a system according to Newton's first law.  *Defining the boundaries of the system makes it easier to identify which forces are external.*

Acceleration is directly proportional to (indicated mathematically with $\propto$) and in the same direction as the net (total) external force acting on a system.  A smaller force causes a smaller acceleration and vice versa.  The vertical forces are shown for completeness, where they must cancel.  Otherwise the wagon's motion would change along the vertical direction; hopefully not!  These vertical forces are the weight $\vec{w}$ of the wagon and child, along with the support of the ground $\vec{N}$.  There is a horizontal frictional force $\vec{f}$ that opposes the applied force from the two children on the left.  **Frictional forces** oppose the motion of other objects that are touching.  We can simplify the scenario using a FBD with only the external forces, which add together to produce the net force $\vec{F}_{\rm net}$. 

We established that the acceleration $\vec{a}$ is proportional to the net external force $\vec{F}_{\rm net}$, where this is expressed mathematically as

\begin{align}
\vec{a} \propto \vec{F}_{net},
\end{align}

and the $\propto$ symbols means "proportional to".  When a quantity is proportional to another, it means that as one increases so does the other in proportion.

Going back to our intuition, we know that applying more force can cause an object to move faster.  But a more massive object cause the inverse, where the object is harder to move and thus moves (accelerates) slower.  This is where the mass (i.e., inertia) affects the motion of the object.  The same net external force applied to a car produces a much smaller acceleration than when applied to a basketball.  The relationship between magnitude of the acceleration $a$ and the object's mass $m$ is given by

\begin{align}
a \propto \frac{1}{m}.
\end{align}

The acceleration of an object depends only on the net external force and the mass of the object.  Combining the two proportionalities yields Newton's second law of motion.

```{admonition} Newton's Second Law of Motion
:class: note

> The acceleration of a system is directly proportional to and in the same direction as the net external force acting on the system, and inversely proportional to its mass.

In equation form, Newton's second law of motion is

\begin{align}
\vec{a} = \frac{\vec{F}_{\rm net}}{m}.
\end{align}

This is often written in the more familiar form 

\begin{align}
\vec{F} &= m\vec{a}, \\
F &= ma,
\end{align}

where the vector form retains all information concerning the direction of force and acceleration.  The scalar form is useful when Newton’s second law is applied separately to each coordinate direction.
```
### Summing Forces by Direction

Newton's second law involves the **net force**, which means that we must add together all of the external forces acting on the system. Because forces are vectors, we do this separately in each coordinate direction. For motion in two dimensions, Newton's second law can therefore be written as

\begin{align}
\sum F_x &= ma_x, \\
\sum F_y &= ma_y.
\end{align}

The symbol $\sum$ means "sum of." Thus, $\sum F_x$ means to add all of the force components in the $x$-direction, while $\sum F_y$ means to add all of the force components in the $y$-direction.

The signs of the forces depend on the coordinate system that we choose. If we define right and up as positive, then forces pointing to the right or upward are positive, while forces pointing to the left or downward are negative.

For the wagon example (Fig. {numref}`{number}<fig-system-to-fbd>`), the two children push toward the right with forces $F_1$ and $F_2$, while friction $f$ acts toward the left. The horizontal force equation is therefore

\begin{align}
\sum F_x = F_1 + F_2 - f = ma_x.
\end{align}

The normal force $N$ acts upward and the weight $w$ acts downward. Since the wagon does not accelerate vertically,

\begin{align}
\sum F_y = N - w = ma_y = 0.
\end{align}

Therefore,

\begin{align}
N = w.
\end{align}

Notice that $\vec{F}_{\rm net}$ is **not an additional force** that we draw on the free-body diagram. It is the result of adding all of the external forces already acting on the system.

This gives us a general procedure for applying Newton's second law:

1. Choose the system of interest.
2. Draw a free-body diagram showing all external forces acting on the system.
3. Choose convenient positive $x$- and $y$-directions.
4. If a force is angled, break it into $x$- and $y$-components.
5. Add the force components in each direction, including their signs.
6. Set each sum equal to the mass times the acceleration in that direction:

\begin{align}
\sum F_x &= ma_x, \\
\sum F_y &= ma_y.
\end{align}

A system can accelerate in one direction while having zero acceleration in another. In the wagon example, the horizontal forces do not cancel, so the wagon accelerates horizontally. The vertical forces do cancel, so its vertical acceleration is zero.


### Units of Force
The units of force are determined from the units that makeup $\vec{F}_{\rm net} = m\vec{a}$ in terms of mass, length, and time.  The SI unit of force is called the **newton** (abbreviated $\rm N$) and is the force needed to accelerate 1-kg at a rate of $1\ {\rm m/s^2}$.  We can see this by using the scalar form to get

\begin{align*}
F &= ma, \\
1\ {\rm N} &= 1\ {\rm kg \cdot m/s^2}.
\end{align*}

The unit of force in the United States is the pound $(\rm lb)$, where you can convert between the units using $1\ {\rm lb} = 4.45\ {\rm N}$.

### Weight and the Gravitational Force

As we've seen in prior chapters, an object accelerates toward the center of Earth when dropped from a height.  Newton's second law states that a net force on an object is responsible for its acceleration.  Neglecting air resistance, the net force on a falling object is the gravitational force and is commonly called its **weight** $\vec{w}$.  Weight can be denoted as a vector because it has a direction (downward toward Earth's center).  The magnitude of the weight is given by $w$.  Galileo showed that (quite unintuitively) that objects fall with the same acceleration $g$ when air resistance is neglected.  If you drop a hammer and a feather from the same height in a vacuum, they both hit the ground at the same time.

Consider an object with mass $m$ falling downward toward Earth.  It experiences only the downward force of gravity, which has a magnitude $w$.  We can use Newton's second law to write $F_{\rm net} = ma,$ but the only force acting on the object is gravity so that $F_{\rm net} = w$.  We know that the magnitude of the acceleration due to gravity is given by $a = g$.  Thus we can combine these relations together to determine the magnitude of the object's weight $w$.

```{admonition} Weight
The gravitational force on a mass $m$ is given by

\begin{align}
w = mg.
\end{align}

Since $g = 9.81\ {\rm m/s^2}$ on Earth, the weight of a 1-kg object is $9.81\ {\rm N}$.  We can determine this mathematically by

$$ w = mg = (1.0\ {\rm kg})(9.81\ {\rm m/s^2}) = 9.81\ {\rm N}. $$

The weight given above is just its magnitude.  When solving problems with weight, you need to take the orientation of the coordinate system into account to get the magnitude of the respective components.
```

When the net external force is only due to gravity (i.e., its weight), we say that it is in **free-fall**. In the real world, when objects fall downward toward the Earth, there is always some upward force from air acting on the object.  For sufficiently heavy or dense objects, the upward air resistance is negligible compared to the gravitational force.

Weight varies dramatically when an object is far from Earth's surface.  On the Moon, the acceleration due to gravity is about $1/6$ times that of Earth, or $1.625\ {\rm m/s^2}$.  This means that a 1-kg mass has a weight fo only about $1.7\ {\rm N}$ on the Moon compared to the $9.8\ {\rm N}$ on the Earth.

Our definition of weight is measured relative to the largest massive body (e.g., Earth, Moon, Sun, etc.), but popular media can refer to weight differently.  When they speak of "weightlessness" or "microgravity", they are referring to cases where the person (or object) is in free-fall.  The astronauts in the International Space Station (ISS) are not weightless, where they are actually only experiencing ${\sim}10%$ less gravitational force than you do on Earth's surface.

Weight and mass are very different physical quantities.  Mass (in $\rm kg$) is the quantity of matter, where it is not position dependent (i.e., you still have the same mass if you go to the Moon) and it does not vary in classical physics.  Weight (in $\rm N$) depends on the strength of gravity at that location.  


### Example Problem: Lawn Mower Acceleration

`````{admonition} Example Problem: Lawn Mower Acceleration
:class: etamu-exercise

**The Problem**

> A lawn mower with a mass of $24\ {\rm kg}$ experiences a net external force of $51\ {\rm N}$ parallel to the ground and directed to the right. What is the acceleration of the mower?

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

The lawn mower is the system of interest. We define the horizontal $x$-axis parallel to the ground, with $+x$ pointing to the right. The net external force is already known to be $F_{{\rm net},x}=+51\ {\rm N}$, the mower has a mass of $m=24\ {\rm kg}$, and the unknown is the horizontal acceleration $a_x$. Because the net force points entirely along the $x$-axis, the motion can be treated as a one-dimensional Newton's second-law problem.

```{figure-md} os-example-4-1-lawnmower-fig
:align: center

<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/38266b2b19731efad6e5e251249ea092620720a7" alt="A person pushes a lawn mower to the right while a net-force vector points horizontally to the right." style="width:65%; height:auto; display:block; margin:0 auto;">

The net external force on the lawn mower points to the right. **Source: [OpenStax College Physics 2e, Figure 4.7](https://openstax.org/books/college-physics-2e/pages/4-3-newtons-second-law-of-motion-concept-of-a-system).**
```

---

**The Math**

Newton's second law connects the known net external force to the mower's acceleration. Along the horizontal axis, the relationship is

$$ \sum F_x=ma_x. $$

The net force is already given, so solving for the acceleration and substituting the known values gives

$$ a_x=\frac{F_{{\rm net},x}}{m}=\frac{51\ {\rm N}}{24\ {\rm kg}}=2.125\ {\rm m/s^2}\approx2.1\ {\rm m/s^2}. $$

---

**The Conclusion**

The lawn mower accelerates at $2.1\ {\rm m/s^2}$ to the right. The positive sign is consistent with the chosen coordinate system, and the acceleration points in the same direction as the net external force, as required by Newton's second law.

````
`````


*Adapted from [OpenStax College Physics 2e, Example 4.2](https://openstax.org/books/college-physics-2e/pages/4-3-newtons-second-law-of-motion-concept-of-a-system).*


## Newton's Third Law of Motion
```{admonition} Newton's Third Law of Motion
:class: note

>Whenever one body exerts a force on a second body, the first body experience a force that is equal in magnitude and opposite in direction to the force that it exerts.
```

This law represents a certain symmetry in nature, where forces always occur in pairs.  One body cannot exert a force on another without experiencing a force itself.  This is sometimes loosely referred as "action-reaction", where the force exerted is the action and the force experience as a consequence is a reaction.  Newton's third law has practical uses, but we must be careful in defining our system.

Consider a swimmer pushing off from the side of a pool (see Figure {numref}`{number}<fig-swimmer-third-law-fbd>`).  She pushes against the pull wall with her feet and accelerates in the *opposite direction*.  The wall exerted an equal and opposite force back on the swimmer  You might think that th ewo equal and opposite forces would cancel, but they do not *because they act on different systems.*  Otherwise the horizontal forces would be balanced exactly and the swimmer would not move.

```{figure-md} fig-swimmer-third-law-fbd 
<img src="FBD_3rdLaw_diagram.png" alt="Three-panel sequence showing how a free-body diagram is constructed for a swimmer pushing away from a pool wall. Panel a identifies the swimmer as the system and shows buoyancy upward, weight downward, the wall force on the swimmer to the left, the swimmer force on the wall to the right, and acceleration to the left. Panel b isolates the swimmer and retains only the external forces acting on the swimmer. Panel c replaces the swimmer with a point and shows the buoyant force, weight, and wall force that make up the swimmer's free-body diagram." style="width:95%; height:auto;">

Constructing the swimmer's free-body diagram begins by choosing the **system**. In (a), the swimmer is the system, while the water and pool wall are outside it. The swimmer pushes on the wall with $\vec{F}_\text{feet on wall}$, and the wall pushes back on the swimmer with the equal-and-opposite force $\vec{F}_\text{wall on feet}$; these forces act on different objects and therefore do not appear together on the same free-body diagram. In (b), the surroundings are removed and only the external forces acting on the swimmer are retained: the wall force, buoyant force $\overrightarrow{BF}$, and weight $\vec{w}$. In (c), the swimmer is reduced to a point to produce the final free-body diagram.
```

In this case, there are two systems: (1) the swimmer or (2) the wall.  If we select the swimmer to be the system of interest, then $\vec{F}_\text{wall on feet}$ is an external force on this system and affects its motion.  Intuitively we expect the swimmer to move in the direction of $\vec{F}_\text{wall on feet}$.  In contrast, the force $\vec{F}_\text{feet on wall}$ acts on the wall and is **not** on our system of interest.  Thus it is not included in the FBD for the swimmer.  It could be included as a separate FBD for the wall as its own system.  Note that the reaction to her push moves the swimmer in the desired direction.

Other examples of Newton's third law are easy to find.  When you walk across the floor, you are exerting a force backward on the floor, where the floor exerts a reaction force forward that causes you to accelerate forward.  A similar situation occurs for the wheels on your car.  

Rockets move forward by expelling gas backward at high velocity.  The rocket exerts a large backward force on the gas in the rocket's combustion chamber and the gas exerts a large reaction force forward on the rocket, where this reaction force is called **thrust**.  It is a common misconception that the rockets are pushing on the ground or air behind them.  They actually work better in a vacuum, where it is easier to expel the gas.

```{admonition} Interactive Simulation: Gravity Force Lab
:class: tip

<iframe
  src="https://phet.colorado.edu/sims/html/gravity-force-lab-basics/latest/gravity-force-lab-basics_en.html"
  width="100%"
  height="600px"
  style="border: none;"
  allowfullscreen
  loading="lazy">
</iframe>

Use this PhET simulation to visualize the gravitational force. Change the properties of the objects to see how it changes the force.
```

### Example Problem: Choosing the System — Professor and Cart

`````{admonition} Example Problem: Choosing the System — Professor and Cart
:class: etamu-exercise

**The Problem**

> A physics professor pushes a cart containing demonstration equipment. The professor has a mass of $65\ {\rm kg}$, the cart has a mass of $12\ {\rm kg}$, and the equipment has a mass of $7\ {\rm kg}$. The professor pushes backward on the floor with a force of $150\ {\rm N}$. Friction and other forces opposing the motion total $24\ {\rm N}$. What is the acceleration of the professor, cart, and equipment?

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

The acceleration belongs to the professor, cart, and equipment moving together, so all three objects are included in the system of interest. This corresponds to System 1 in the figure below. We define the horizontal $x$-axis along the motion, with $+x$ pointing to the right. The professor pushes backward on the floor with $150\ {\rm N}$, so Newton's third law tells us that the floor pushes forward on the professor with an equal $150\ {\rm N}$ force. That floor force is external to the chosen system, while the $24\ {\rm N}$ resistive force acts in the negative $x$-direction. Forces between the professor and the cart are internal to this system and therefore do not enter the net external force.

```{figure-md} professor-cart-system-fig
:align: center

<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/08954c65d36d788e4376b264e18313b0724101ee" alt="A professor pushes a cart of demonstration equipment. Two system boundaries and their free-body diagrams show how the external forces depend on the chosen system." style="width:90%; height:auto; display:block; margin:0 auto;">

The free-body diagram depends on the system boundary. For this example, System 1 includes the professor, cart, and equipment. **Source: [OpenStax College Physics 2e, Figure 4.10](https://openstax.org/books/college-physics-2e/pages/4-4-newtons-third-law-of-motion-symmetry-in-forces).**
```

---

**The Math**

Because all three objects belong to the system, their masses add to give the total mass that is accelerated:

$$ m=(65\ {\rm kg})+(12\ {\rm kg})+(7\ {\rm kg})=84\ {\rm kg}. $$

Along the horizontal axis, the forward force from the floor is opposed by the resistive force, so the net external force is

$$ \sum F_x=150\ {\rm N}-24\ {\rm N}=126\ {\rm N}. $$

Newton's second law then connects this net force to the acceleration of the complete system:

$$ a_x=\frac{\sum F_x}{m}=\frac{126\ {\rm N}}{84\ {\rm kg}}=1.5\ {\rm m/s^2}. $$

---

**The Conclusion**

The professor, cart, and equipment accelerate together at $1.5\ {\rm m/s^2}$ to the right. Treating the three objects as one system removes the forces between the professor and cart from the calculation because those interactions occur entirely inside the system boundary.

````
`````


*Adapted from [OpenStax College Physics 2e, Example 4.4](https://openstax.org/books/college-physics-2e/pages/4-4-newtons-third-law-of-motion-symmetry-in-forces).*


## Normal, Tension, and other Forces
Forces have many names including push, pull, thrust, lift, weight, friction, and tension.  Typically, they are grouped depending on their source, how they are transmitted, or their effects.

### Normal Force
Weight is a force that acts at all times, but if it were the only vertical force we would all be sinking towards the center of the Earth.  Since this does not happen, there must be a force that balances the weight that in the opposite direction.  Near Earth's surface we can treat the surface as flat, where the weight of a box is represented by an arrow pointing downward.  The balancing force is normal (i.e., perpendicular to the surface) and is point upward with an equivalent magnitude, which is called the **normal** force.  Forces are vectors, which means we represent the normal force as $\vec{N}$.  Note that the magnitude of the normal force is $N$, where the unit symbol for the newton is $\rm N$.  This difference is subtle, but the context usually provides a clear distinction by which one is meant.

For a flat surface, we can represent the normal force by

\begin{align}
\vec{N} = \vec{w} = m\vec{g}, \\
N = w = mg,
\end{align}

in either the vector or scalar form.

*How does an inanimate object (e.g., a table) support the weight of a mass placed on them?*  If you're holding a bag of dog food, it makes intuitive sense on why you hold it up because you apply a force to counteract the weight.  When the dog food is placed on the table, it actually sags slightly under the load.  This is more noticeable with an especially heavy bag and weak table.  The table exerts a *restoring* force much like a deformed spring.  The greater the deformation, the greater the restoring force.  The table applies only enough restoring force to support the weight of the bag.

On an inclined surface, the normal force can be less than the weight which allows for objects to slide down.  When an object rests on an incline that makes an angle $\theta$ with the horizontal, the force of gravity acting on the object is divided into two components:

1. a force acting perpendicular to the plane, $w_y$, and
2. a force acting parallel to the plane, $w_x$.

The normal force $\vec{N}$ is typically equal in magnitude and opposite in direction to $w_y$, where the component parallel to the plane $w_x$ causes the object to accelerate down the incline (see Fig. {numref}`{number}<incline-force-fig>`).  The $w_x$ is the unbalanced force which allows the acceleration to occur, where $w_y$ is balanced by the normal force $N$.

```{figure-md} incline-force-fig
<img src="https://saturnaxis.github.io/IntroPhysics/_images/Inclined_Plane.png" alt="force on an incline example"  style="max-width:70%; height:auto;">

The coordinate system is chosen so that the $x$-axis is parallel to the inclined surface. The angle $\beta\ (= 90^\circ - \theta)$ is complementary to $\theta$. Modified from OpenStax [Figure 5.23](https://openstax.org/books/university-physics-volume-1/pages/5-6-common-forces).
```

```{admonition} Angles on a Triangle
:class: tip, dropdown

*How do we know that the angle $\theta$ given in Fig. {numref}`{number}<incline-force-fig>` is the same in both places?*

Let's assume that they are different: 

- $\theta_1$ is the angle between between $\vec{w}$ and $\vec{w}_y$,
- $\theta_2$ is the angle between the horizontal surface and incline plane.

The angle between $\vec{w}$ and $\vec{w}_x$ is $\beta = 90^\circ - \theta_1$ because $\vec{w}_x$ and $\vec{w}_y$ are perpendicular.  The inclined surface also makes a triangle, with 3 angles that sum to $180^\circ$, or

$$ \theta_2 + 90^\circ + \beta = 180^\circ.$$

We can solve for $\beta$ in terms of $\theta_2$ and substitute into the equation for the equation with $\theta_1$,

\begin{align*}
\beta &= 90^\circ - \theta_2, \\
90^\circ - \theta_2 &= 90^\circ - \theta_1,\\
\theta_2 &= \theta_1.
\end{align*}

With a little geometry, we showed that $\theta_1 = \theta_2 = \theta$.
```

**Be careful when resolving the weight into components.**  Normally when the $x$-component of a force $\vec{F}$ is resolved, we find that $F_x = F\cos{\theta}$.  However, in this case we are using the complementary angle $\beta = 90^\circ - \theta$ to resolve the $w_x$.  Using the red vectors in Fig. {numref}`{number}<incline-force-fig>`, we have

\begin{align*}
w_x &= w\cos{\beta} = w\sin{\theta}, \\
w_y &= w\cos{\theta}.
\end{align*}


### Tension
 A **Tension** is a force along the length of a medium, which is usually represented by a flexible medium, such as a rope or cable.  The work "tension" comes from the Latin word "to stretch", which is why the flexible coords that carry muscle forces are called *tendons*.  Any flexible connector can exert pulls only parallel to its length, which means the forces carried by the connector are in the same direction.  

 ```{figure-md} fig-normal-force-examples
<img src="Normal_force_diagram.png" alt="Three-panel comparison of vertical forces for objects at rest. Panel a shows a cardboard box resting on the floor with an upward normal force and downward weight, together with its free-body diagram. Panel b shows a bag of dog food supported by a person's hands with an upward hand force and downward weight, together with its free-body diagram. Panel c shows the same bag resting on a table with an upward normal force and downward weight, together with its free-body diagram." style="width:95%; height:auto;">

Three examples illustrate how different interactions can produce the upward force that balances an object's weight. In (a), the floor exerts the normal force $\vec{N}$ on a box. In (b), a person's hands exert an upward force $\vec{F}_{\rm hand}$ on a bag of dog food. In (c), the table exerts the normal force $\vec{N}$ on the bag. The free-body diagram beneath each scene keeps only the external forces acting on the object.
```
 
 Consider a person holding a mass $m$ on a rope, where the mass is held stationary.  The tension $\vec{T}$ must equal the weight of the supported mass.  Otherwise there would be a net force moving the mass up or down as expected from Newton's second law.  A stationary mass has $\vec{F}_{\rm net} = 0$, where the only external forces acting on the mass are its weight $\vec{w}$ and the tension $\vec{T}$ supplied by the rope.  Thus, we have

 \begin{align}
 F_{\rm net} = T - w = 0,
 \end{align}

 where $T$ and $w$ are the magnitudes of the tension and weight, respectively.  We consider only the tension acting on the mass $m$, which is upwards and has a positive sign, where the weight acts downwards and carries a negative sign.  Just as you would expect 

 $$ T = w = mg. $$

 For a $5.0$-kg mass (and neglecting the mass of the rope), we see that

 $$ T = mg = (5.0\ {\rm kg})(9.81\ {\rm m/s^2}) = 49\ {\rm N}. $$

 If we cut the rope and insert a spring, the spring would extend a length corresponding to a force of $49\ {\rm N}$.

 Flexible connectors are often used to transmit forces around corners (e.g., a hospital traction system, a finger joint, or a bicycle brake cable).  If there is no friction, the tension is transmitted undiminished. Only its direction changes, and it is always parallel to the flexible connector.

 ### Large Tension

A surprisingly large tension can develop when a force is applied perpendicular to a taut rope or other flexible connector (see Fig. {numref}`{number}<tension-tightrope-fig>`). A tightrope walker provides a useful example. The walker's weight pulls downward on the rope, while the tension in the rope pulls upward and outward on the walker.

```{figure-md} tension-tightrope-fig
<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/ac4a7a65e7119df31a54d79532d93ff21c45c769" alt="A tightrope walker standing at the center of a sagging wire. The wire slopes upward by 5 degrees on each side of the walker, illustrating the tension forces acting along the wire." style="width:90%; height:auto;">

The weight of a tightrope walker causes a wire to sag by $5.0^\circ$. The system of interest here is the point in the wire at which the tightrope walker is standing. [Source: OpenStax College Physics, Figure 4.16](https://openstax.org/books/college-physics-2e/pages/4-5-normal-tension-and-other-examples-of-forces).
```

Suppose the tightrope walker is standing still. Because the walker is not accelerating, the forces must balance in both the horizontal and vertical directions. We divide the tension into the forces from the right and left sections of the rope, $\vec{T}_R$ and $\vec{T}_L$.

1. The vertical components, $T_{Ry}$ and $T_{Ly}$, together support the walker's weight $w$.
2. The horizontal components, $T_{Rx}$ and $T_{Lx}$, point in opposite directions and cancel.

The force balance is therefore

\begin{align*}
F_{{\rm net},y} &= T_{Ry} + T_{Ly} - w = 0, \
F_{{\rm net},x} &= T_{Rx} - T_{Lx} = 0.
\end{align*}

Assume the rope makes the same angle $\theta$ with the horizontal on each side of the walker. The horizontal components are

\begin{align*}
T_{Rx} &= T_R\cos{\theta}, \
T_{Lx} &= T_L\cos{\theta}.
\end{align*}

Because these horizontal components must be equal,

\begin{align*}
T_R\cos{\theta} &= T_L\cos{\theta},
\end{align*}

which gives

$$
\boxed{T_R=T_L=T}.
$$

For an ideal lightweight rope, the tension has the same magnitude throughout the rope. The two sides therefore each pull on the walker with the same tension $T$.

Now consider the vertical direction. Each side provides an upward component

$$
T_y=T\sin{\theta}.
$$

Both sides together support the walker's weight, so

\begin{align*}
2T\sin{\theta}-w &= 0,\
2T\sin{\theta} &= w,\
T &= \frac{w}{2\sin{\theta}}.
\end{align*}

Since $w=mg$,

\begin{align}
T=\frac{mg}{2\sin{\theta}}.
\end{align}

This result explains why tightropes, clotheslines, cables, and similar connectors can experience very large tensions. As the rope becomes more nearly horizontal, $\theta$ becomes smaller and $\sin{\theta}$ decreases. The tension must then increase so that its small vertical components can still support the downward force.

More generally, if a perpendicular force $F_\perp$ is applied at the midpoint of a flexible connector,

\begin{align}
T=\frac{F_\perp}{2\sin{\theta}}.
\end{align}

As $\theta\rightarrow0$, the required tension becomes extremely large. A perfectly horizontal flexible connector supporting a downward force would require infinite tension in this idealized model. This is why real ropes and cables always sag at least slightly under their own weight or any additional load.


### Example Problem: Weight and Motion on an Incline

`````{admonition} Example Problem: Weight and Motion on an Incline
:class: etamu-exercise

**The Problem**

> A skier and their equipment have a combined mass of $60\ {\rm kg}$ and move down a slope that is inclined at $25^\circ$ above the horizontal.
>
> (a) What is the skier's acceleration if friction is negligible?
>
> (b) What is the skier's acceleration if a frictional force of $45\ {\rm N}$ acts up the slope?
>
> (c) What normal force does the slope exert on the skier?

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

The skier is the system of interest. Because the motion follows the surface of the incline, we choose one axis parallel to the slope and a second axis perpendicular to it. The positive parallel direction points downhill, while the positive perpendicular direction points away from the surface. The skier's weight acts vertically downward, the normal force acts perpendicular to the slope, and any frictional force acts parallel to the slope and opposite the downhill motion. The skier has no acceleration perpendicular to the surface, so $a_{\perp}=0$, while the unknown acceleration lies along the slope.

```{figure-md} os-example-4-5-skier-incline-fig
:align: center

<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/b29f730e665f5f19a0ee85d6d249e94a141a4b9a" alt="A skier moves down a 25-degree incline. Weight is resolved into components parallel and perpendicular to the slope, while the normal force points away from the surface and friction points uphill." style="width:85%; height:auto; display:block; margin:0 auto;">

Choosing axes parallel and perpendicular to the incline makes the force components align with the motion and the surface. **Source: [OpenStax College Physics 2e, Figure 4.12](https://openstax.org/books/college-physics-2e/pages/4-5-normal-tension-and-other-examples-of-forces).**
```

---

**The Math**

The skier's weight must first be resolved relative to the chosen axes. The component of weight parallel to the slope is $w_{\parallel}=mg\sin\theta$, while the component perpendicular to the slope is $w_{\perp}=mg\cos\theta$.

**(a)** The friction is negligible, so the parallel component of weight is the only force along the slope. Newton's second law gives

$$ \sum F_{\parallel} = mg\sin\theta = ma_{\parallel}. $$

The mass cancels, and the downhill acceleration is therefore

\begin{align*}
a_{\parallel}&= g\sin\theta, \\
&=(9.81\ {\rm m/s^2})\sin(25^\circ)=4.15\ {\rm m/s^2}\approx4.1\ {\rm m/s^2}.
\end{align*}

**(b)** The $45\ {\rm N}$ frictional force points uphill and therefore enters with a negative sign. The parallel force equation becomes

$$ \sum F_{\parallel}=mg\sin\theta-f=ma_{\parallel}. $$

Solving for the acceleration and evaluating the expression gives

\begin{align*}
a_{\parallel} = \frac{mg\sin\theta-f}{m}, \\
&=\frac{(60\ {\rm kg})(9.81\ {\rm m/s^2})\sin(25^\circ)-45\ {\rm N}}{60\ {\rm kg}}=3.40\ {\rm m/s^2}\approx3.4\ {\rm m/s^2}.
\end{align*}

**(c)** the skier remains in contact with the slope and has no acceleration perpendicular to the surface. The normal force must therefore balance the perpendicular component of the weight:

$$ \sum F_{\perp} = N-mg\cos\theta = 0. $$

The normal force is then

\begin{align*}
N = mg\cos\theta, \\
&=(60\ {\rm kg})(9.81\ {\rm m/s^2})\cos(25^\circ)=5.33\times10^2\ {\rm N}\approx5.3\times10^2\ {\rm N}.
\end{align*}

---

**The Conclusion**

Without friction, the skier accelerates downhill at approximately $4.1\ {\rm m/s^2}$. Adding $45\ {\rm N}$ of uphill friction reduces the acceleration to about $3.4\ {\rm m/s^2}$. The slope exerts a normal force of approximately $5.3\times10^2\ {\rm N}$, which is smaller than the skier's full weight because the normal force balances only the component of weight perpendicular to the surface.

````
`````

## Applying Newton's Laws: A Problem-Solving Strategy

We have introduced several different forces and all three of Newton's laws, but most dynamics problems use the same basic approach.  The difficulty is usually not choosing a new equation.  Instead, we need to determine **what system we are studying, what forces act on it, and how those forces combine in each direction**.

Newton's second law provides the connection between the forces and the resulting motion,

\begin{align}
\sum F_x &= ma_x, \\
\sum F_y &= ma_y.
\end{align}

The important part is building the equations on the left-hand side correctly.  A free-body diagram gives us a visual inventory of the external forces, while the coordinate system tells us whether each force component is positive or negative.

### A General Strategy

We can organize most force problems using the following steps:

1. **Choose the system.** Decide which object, or collection of objects, we want to analyze.
2. **Identify the external forces.** Determine which interactions cross the boundary of the system.
3. **Draw a free-body diagram.** Replace the system with a point and draw only the external forces acting on it.
4. **Choose the coordinate directions.** Define which directions are positive.  Whenever possible, choose axes that simplify the forces or the motion.
5. **Break angled forces into components.** A force that is not along one of the coordinate axes may contribute to both the $x$- and $y$-directions.
6. **Determine the acceleration in each direction.** Ask whether the velocity is changing along $x$ or $y$.
7. **Sum the forces separately in each direction.**

\begin{align*}
\sum F_x &= ma_x, \\
\sum F_y &= ma_y.
\end{align*}

8. **Solve for the unknown quantity and check the result.** The units, sign, direction, and approximate size of the answer should agree with the physical situation.

```{admonition} The Free-Body Diagram Comes Before the Equations
:class: tip

Do not begin a force problem by searching for numbers to substitute into $F=ma$.  First decide on the system and construct the free-body diagram.  The equations should come directly from the forces shown on that diagram.

The **net force is not another force** that should be added to the free-body diagram.  It is the result of adding together the external forces already shown.
```

### Acceleration Tells Us What the Force Sum Must Do

Before writing an equation for a particular direction, determine whether the system accelerates in that direction.

If the acceleration is zero, then

\begin{align*}
\sum F_x &= 0
\end{align*}

or

\begin{align*}
\sum F_y &= 0,
\end{align*}

depending on the direction being considered.

An acceleration of zero does **not** require the object to be stationary.  *An object moving with constant velocity also has zero acceleration and therefore zero net force.*

If the system accelerates in a direction, then the forces along that direction do not cancel.  For example,

\begin{align*}
\sum F_x = ma_x.
\end{align*}

This distinction becomes especially useful when a system accelerates in one direction but not another.  A wagon may accelerate horizontally while remaining on the ground vertically.  In that case,

\begin{align*}
\sum F_x &= ma_x, \\
\sum F_y &= 0.
\end{align*}

The same reasoning applies to more complicated systems.

### Multiple Forces in Two Dimensions

When several forces act at different angles, Newton's second law is still applied independently to each direction.  The free-body diagram may look complicated, but each force contributes only through its $x$- and $y$-components.

Suppose a system has several external forces.  The horizontal equation has the general form

\begin{align*}
\sum F_x = F_{1x}+F_{2x}+F_{3x}+\cdots = ma_x,
\end{align*}

while the vertical equation is

\begin{align*}
\sum F_y = F_{1y}+F_{2y}+F_{3y}+\cdots = ma_y.
\end{align*}

The signs come from the coordinate system that we choose.  If right and up are positive, then a force pointing left has a negative $x$-component and a force pointing downward has a negative $y$-component.

This is why choosing the coordinate system is an important part of the model rather than just a mathematical detail.


### Example Problem: Different Tensions at Different Angles

`````{admonition} Example Problem: Different Tensions at Different Angles
:class: etamu-exercise

**The Problem**

> A $15\ {\rm kg}$ traffic light is suspended by two wires. The left wire makes an angle of $30^\circ$ above the horizontal and has tension $T_1$, while the right wire makes an angle of $45^\circ$ above the horizontal and has tension $T_2$. Find the tension in each wire, neglecting the masses of the wires.

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

The traffic light is the system of interest. We define the horizontal $x$-axis with $+x$ to the right and the vertical $y$-axis with $+y$ upward. The light is stationary, so both acceleration components are zero and the external forces must balance independently along the two axes. Three forces act on the light: its weight $w=mg$ points downward, the left-wire tension $T_1$ points upward and to the left at $30^\circ$ above the horizontal, and the right-wire tension $T_2$ points upward and to the right at $45^\circ$ above the horizontal. The two unknown tension magnitudes require one independent force equation in each direction.

```{figure-md} os-example-4-8-traffic-light-fig
:align: center

<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/bebf3d9c5cc60722f5f19df3df892fe345411da7" alt="A traffic light is suspended by two wires at different angles. The figure shows the tension forces, weight, free-body diagram, and horizontal and vertical tension components." style="width:90%; height:auto; display:block; margin:0 auto;">

The traffic light is in equilibrium, so the horizontal tension components cancel while the vertical components together support its weight. **Source: [OpenStax College Physics 2e, Figure 4.22](https://openstax.org/books/college-physics-2e/pages/4-7-further-applications-of-newtons-laws-of-motion).**
```

---

**The Math**

Because there is no horizontal acceleration, the horizontal components of the two tensions must cancel. Applying Newton's second law along the $x$-axis gives

$$ \sum F_x=T_2\cos(45^\circ)-T_1\cos(30^\circ)=0. $$

This equation relates the two unknown tensions:

$$ T_2=\frac{\cos(30^\circ)}{\cos(45^\circ)}T_1=1.225T_1. $$

The vertical acceleration is also zero, so the upward components of both tensions must balance the weight of the traffic light. The vertical force equation is

$$ \sum F_y=T_1\sin(30^\circ)+T_2\sin(45^\circ)-mg=0. $$

Substituting the horizontal relationship for $T_2$ leaves one equation for $T_1$:

$$ T_1\sin(30^\circ)+(1.225T_1)\sin(45^\circ)=(15\ {\rm kg})(9.81\ {\rm m/s^2}). $$

Solving gives $T_1=108\ {\rm N}$, and the corresponding right-wire tension is

$$ T_2=(1.225)(108\ {\rm N})=132\ {\rm N}. $$

---

**The Conclusion**

The left wire has a tension of approximately $108\ {\rm N}$ and the right wire has a tension of approximately $132\ {\rm N}$. The tensions are different because the wires make different angles with the horizontal. The more steeply angled right wire provides a larger vertical component for a given tension, and the two tension magnitudes adjust so that both the horizontal and vertical force sums are zero.

`````

*Adapted from [OpenStax College Physics 2e, Example 4.8](https://openstax.org/books/college-physics-2e/pages/4-7-further-applications-of-newtons-laws-of-motion).*


### Example Problem: What Does the Bathroom Scale Read in an Elevator?

`````{admonition} Example Problem: What Does the Bathroom Scale Read in an Elevator?
:class: etamu-exercise

**The Problem**

> A $75\ {\rm kg}$ person stands on a bathroom scale in an elevator. What does the scale read (a) when the elevator accelerates upward at $1.20\ {\rm m/s^2}$ and (b) when the elevator moves upward at a constant speed of $1.0\ {\rm m/s}$?

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

The person is the system of interest because the scale reading is determined by the contact force between the scale and the person. We define the vertical $y$-axis with $+y$ upward. Two external forces act on the person: the scale exerts an upward force $F_s$, and the person's weight $w=mg$ acts downward. By Newton's third law, the magnitude of the force that the person exerts on the scale is equal to $F_s$, so finding $F_s$ gives the scale reading. In part (a) the acceleration is upward and positive, while in part (b) the constant velocity means that the acceleration is zero.

```{figure-md} os-example-4-9-elevator-scale-fig
:align: center

<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/cf2e019aa85e3875c7e4550084f78cd89149fa62" alt="A person stands on a bathroom scale in an elevator. The figure shows the forces among the person, scale, and elevator and a simplified free-body diagram for the person with the scale force upward and weight downward." style="width:85%; height:auto; display:block; margin:0 auto;">

Choosing the person as the system reduces the elevator situation to two external forces: the upward force from the scale and the downward weight. **Source: [OpenStax College Physics 2e, Figure 4.23](https://openstax.org/books/college-physics-2e/pages/4-7-further-applications-of-newtons-laws-of-motion).**
```

---

**The Math**

Applying Newton's second law along the vertical axis gives the same governing equation for both parts of the problem:

$$ \sum F_y=F_s-mg=ma_y. $$

Solving for the scale force gives

$$ F_s=m(g+a_y). $$

**(a)** The elevator accelerates upward at $1.20\ {\rm m/s^2}$, so the scale reading is

$$ F_s=(75\ {\rm kg})(9.81\ {\rm m/s^2}+1.20\ {\rm m/s^2})=8.26\times10^2\ {\rm N}\approx826\ {\rm N}. $$

**(b)** Moving upward at constant speed means that the velocity is not changing and therefore $a_y=0$. The scale force then reduces to the person's weight:

$$ F_s=(75\ {\rm kg})(9.81\ {\rm m/s^2})=7.36\times10^2\ {\rm N}\approx736\ {\rm N}. $$

---

**The Conclusion**

The scale reads approximately $826\ {\rm N}$ while the elevator accelerates upward and approximately $736\ {\rm N}$ while it moves upward at constant speed. The larger reading in part (a) occurs because the scale must push upward with more force than the person's weight to produce an upward acceleration. Once the elevator reaches constant velocity, the acceleration is zero and the scale force again equals the person's weight.

````
`````

### Connecting Dynamics Back to Kinematics

Dynamics tells us what causes acceleration, while kinematics tells us how acceleration changes velocity and position.  Many real problems require both ideas.

The connection is

\begin{align}
\text{forces} \longrightarrow \text{acceleration} \longrightarrow \text{change in motion}.
\end{align}

Sometimes a problem gives the forces and asks for the resulting motion.  In that case, Newton's second law is used first to determine the acceleration, followed by the kinematics equations.

Other problems give information about the motion and ask what force produced it.  In those cases, we work in the opposite direction: use kinematics to determine the acceleration and then use Newton's second law to determine the force.

### Example Problem: What Force Must a Soccer Player Exert to Reach Top Speed?

````{admonition} Example Problem: What Force Must a Soccer Player Exert to Reach Top Speed?
:class: etamu-exercise

**The Problem**

> A soccer player starts from rest and reaches a forward velocity of $8.00\ {\rm m/s}$ in $2.50\ {\rm s}$. (a) What is the player's average acceleration? (b) What average force does the player exert backward on the ground to produce this acceleration? The player's mass is $70\ {\rm kg}$, and air resistance is negligible.

```{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

This problem connects kinematics and dynamics. We define the horizontal $x$-axis along the player's motion, with $+x$ pointing forward. The player starts from rest, so $v_{0x}=0$, and reaches $v_x=8.00\ {\rm m/s}$ after $\Delta t=2.50\ {\rm s}$. These kinematic quantities determine the average acceleration. The player's mass is $70\ {\rm kg}$, so Newton's second law can then be used to determine the net horizontal force on the player. With air resistance neglected, the horizontal force from the ground is the force responsible for the forward acceleration, and Newton's third law gives an equal-magnitude force that the player exerts backward on the ground.

---

**The Math**

The average acceleration follows from the change in velocity divided by the elapsed time:

$$ a_x=\frac{\Delta v_x}{\Delta t}=\frac{8.00\ {\rm m/s}-0}{2.50\ {\rm s}}=3.20\ {\rm m/s^2}. $$

With the acceleration known, Newton's second law gives the magnitude of the horizontal force exerted by the ground on the player:

$$ F_{\rm net}=ma_x=(70\ {\rm kg})(3.20\ {\rm m/s^2})=224\ {\rm N}. $$

Newton's third law then tells us that the player exerts a force of the same magnitude backward on the ground.

---

**The Conclusion**

The player's average forward acceleration is $3.20\ {\rm m/s^2}$, and the player exerts an average force of approximately $224\ {\rm N}$ backward on the ground. The ground exerts the corresponding $224\ {\rm N}$ force forward on the player, producing the acceleration. This example shows how kinematics can first determine the acceleration and dynamics can then identify the force required to produce that change in motion.

```
````

```{admonition} Putting It All Together
:class: note

Most problems involving Newton's laws can be organized using the same sequence:

**Situation $\rightarrow$ System $\rightarrow$ Free-body diagram $\rightarrow$ Coordinate directions $\rightarrow$ Sum the forces $\rightarrow$ Solve $\rightarrow$ Check**

The equations should follow from the physical model rather than the other way around.  If the free-body diagram and coordinate directions are correct, writing

\begin{align}
\sum F_x &= ma_x, \\
\sum F_y &= ma_y
\end{align}

usually becomes the straightforward part of the problem.
```

The final step is always to ask whether the answer makes physical sense.  Check the units, sign, direction, and approximate magnitude.  A force should have units of newtons, an acceleration should have units of $\rm m/s^2$, and the direction of the net force should agree with the direction of the acceleration.  These checks will not prove that a solution is correct, but they can quickly reveal many common mistakes.
