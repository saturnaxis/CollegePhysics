## What is a Fluid?

Matter most commonly exists as a solid, liquid, gas, or plasma, which are the common *phases of matter*.

- Solids have a definite shape, a specific volume, and strong chemical (molecular) bonds holding the structure together.
- Liquids have a definite volume, but their bonds are weaker so that their shape changes depending on the container in which they are held.
- Gases have neither a definite shape nor a specific volume, which means the bonds are either extremely weak or nonexistent.
- Plasmas are similar to gases, but they are ionized which means that electric currents can freely flow within them.

Liquids, gases, and plasmas behave like a fluid because they yield to shearing forces, whereas solids resist them.  The fluids we are accustomed to are characterized by their *viscosity*, which measures the extent to which the fluids yield to shearing (i.e., how easily and quickly they flow).

Atoms in *solids* are in close contact, which allows the atoms to vibrate but not to change positions with neighboring atoms.  As a result, we can think of the molecular bonds like springs that can be stretched or compressed, but not easily broken.  

- A solid resists all types of stress (see [Elasticity: Stress and Strain](https://saturnaxis.github.io/CollegePhysics/Chapter05/friction-drag-elasticity.html#elasticity-stress-and-strain)).  
- A solid cannot be easily deformed because the bonds prevent the atoms from moving freely.  
- Solids resist compression because their atoms form part of a lattice structure in which the atoms remain a relatively fixed distance apart.  Under compression, the atoms would be forced closer together.

Liquids deform easily when stressed and do not spring back into their original shape once the force is removed because the atoms are free to slide about and change neighbors.  They flow with the molecules held together by their mutual attraction.  When a liquid is placed in a container, it takes the shape of the container where the atoms resist compression (like solids).

Atoms in gases and plasmas are separated by distances that are very large compared to the size of the individual particles.  Intermolecular forces in a dilute gas are usually weak except during collisions; charged particles in a plasma can also interact through long-range electric and magnetic fields. Gases and plasmas flow easily and are relatively easy to compress because there is so much space (and little force) between the particles.  In an open container, gases and plasmas will escape because they have so much energy.

Due to their general properties of flow, we treat liquids, gases, and plasmas simply as fluids, although their viscosity, compressibility, and electrical properties differ.

## Density

*Which weighs more: a ton of feathers or a ton of bricks?*

This old riddle plays with the distinction between mass, weight, and density.  A ton of something weighs the same no matter its composition.  Bricks have a greater density than feathers, so we are tempted to think of them as heavier.

**Density** is crucial in determining whether an object sinks or floats in a fluid.

```{admonition} Density
:class: note

Density $\rho$ is mass $m$ per unit volume $V$, or:

\begin{align}
\rho = \frac{m}{V},
\end{align}

where the volume $V$ is usually determined by an assumed shape of the matter (e.g., cube, sphere, cylinder, etc.).
```

In the riddle regarding the feathers and bricks, the weights are the same but the volume occupied by the feathers is much greater resulting in a lower average density.  The SI unit of density is ${\rm kg/m^3}$, which was used to derive the first units in the metric system.  The basic mass unit (the kilogram) was defined by the mass of $1\ {\rm L}$ of water, where $1\ {\rm L}$ is a volume of $1000\ {\rm mL}$, $1000\ {\rm cm^3}$, or $1000\ {\rm cc}$.

The density of an object can help identify its composition.  For example, the density of gold is about $2.5\times$ the density of iron, which itself is about $2.5\times$ the density of aluminum. The densities of gases are much lower because the atoms are separated by large amounts of empty space.

### Example Problem: Mass of a Reservoir

`````{admonition} Example Problem: Mass of a Reservoir
:class: etamu-exercise

**The Problem**

> A reservoir has a surface area of $50\ {\rm km^2}$ and an average depth of $40\ {\rm m}$. Estimate the mass of water held behind the dam. Use $\rho_{\rm w}=1.000\times10^3\ {\rm kg/m^3}$.

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

We treat the water as having uniform density. The reservoir need not have a flat bottom: multiplying its surface area by its average depth gives its volume. Density then connects that volume to the unknown mass. Because the dimensions describe a large, irregular reservoir, the answer is an estimate.

---

**The Math**

The reservoir volume depends on its surface area and average depth. We first express the area in square meters so that both dimensions use compatible units. Because area contains two factors of length, the conversion from kilometers to meters must be squared.

$$
A=(50\ {\rm km^2})\left(\frac{10^3\ {\rm m}}{1\ {\rm km}}\right)^2=5.0\times10^7\ {\rm m^2}.
$$

The average depth accounts for variations in the shape of the reservoir bottom. Multiplying this depth by the surface area gives the total volume of water.

$$
V=A\bar h=(5.0\times10^7\ {\rm m^2})(40\ {\rm m})=2.0\times10^9\ {\rm m^3}.
$$

Each cubic meter contains a mass of water determined by its density. The relationship $m=\rho V$ therefore converts the calculated volume into the total mass held in the reservoir.

$$
m=(1.000\times10^3\ {\rm kg/m^3})(2.0\times10^9\ {\rm m^3})\approx2.0\times10^{12}\ {\rm kg}.
$$

---

**The Conclusion**

The reservoir holds approximately $2.0\times10^{12}\ {\rm kg}$ of water. This large mass is consistent with a lake covering tens of square kilometers. The mass alone does not determine the horizontal force on the dam; that force also depends on how pressure varies with depth.
````
`````

*Adapted from [OpenStax, College Physics 2e, Example 11.1](https://openstax.org/books/college-physics-2e/pages/11-2-density).*


## Pressure

```{admonition} Pressure
:class: note

Pressure is defined as the force divided by the area perpendicular to the force over which the force is applied, or

\begin{align}
P = \frac{F_\perp}{A},
\end{align}

where the force $F_\perp$ is perpendicular to the area $A$.
```

A given force can have a significantly different effect depending on the area over which it is exerted.  The SI unit for pressure is the *pascal*, where $1\ {\rm Pa} = 1\ {\rm N/m^2}$.  

In addition to the pascal, there are many other units for pressure that are in common use.

```{list-table} Common Units of Pressure (Referenced to Pascals)
:header-rows: 1
:name: pressure-units
:widths: 40 60

* - Unit
  - Equivalent in Pascals

* - Pascal (Pa)
  - $1\ {\rm Pa} = 1\ {\rm N/m^2}$

* - pound per square inch (psi)
  - $1\ {\rm psi} = 6895\ {\rm Pa}$

* - atmosphere (atm)
  - $1\ {\rm atm} = 1.013\times10^{5}\ {\rm Pa}$

* - bar
  - $1\ {\rm bar} = 10^{5}\ {\rm Pa}$

* - torr (mm Hg)
  - $1\ {\rm torr} = 133.3\ {\rm Pa}$

* - inches of mercury (in Hg)
  - $1\ {\rm in\ Hg} = 3386\ {\rm Pa}$

* - millibar (mbar)
  - $1\ {\rm mbar} = 100\ {\rm Pa}$
```

The force exerted on the end of a tank is perpendicular to its inside surface, which is a property of a static, or stationary, fluid.  A fluid at rest cannot sustain a *shearing* (or sideways) stress; viscous shear forces arise when layers of fluid move relative to one another.

Fluid pressure has no direction, which means it is a scalar quantity.  The forces due to pressure have well-defined directions (i.e., they are always exerted perpendicular to any surface).  Pressure is exerted on all surfaces.  Swimmers and tires feel pressure on all sides.

```{figure-md} fig-fluid-pressure-tire-swimmer
<img src="fluid-pressure-tire-swimmer.png" alt="Two panels illustrate fluid pressure. A cutaway tire shows gas molecules colliding with its inner walls. A submerged swimmer experiences inward pressure forces, with larger forces on equal areas at greater depth. Magnified views show molecular interactions at the surfaces." style="width: 1000px; max-width: 100%; height: auto;">

**Fluid pressure acts perpendicular to surfaces.** (a) Gas molecules transfer momentum to the tire walls when they collide with them, producing an outward force on the walls. (b) Water exerts inward forces on the swimmer from all directions. Pressure increases with depth, so equal areas on the swimmer’s underside experience larger forces than those above. Water’s molecular spacing remains nearly unchanged; molecular motion and intermolecular forces transmit the pressure. Red arrows represent forces exerted by the fluid, and the molecular views are schematic and not to scale. Adapted from OpenStax, *College Physics 2e*.
```

### Example Problem: Force on an Air Tank

`````{admonition} Example Problem: Force on an Air Tank
:class: etamu-exercise

**The Problem**

> An astronaut works outside the International Space Station, where the external pressure is essentially zero. Her air-tank gauge reads $6.9\times10^6\ {\rm Pa}$. What force does the air exert on the tank's flat circular end, which has a diameter of $0.15\ {\rm m}$?

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

We treat the pressure inside the tank as uniform over its circular end. The gauge reading is also approximately the absolute pressure because the pressure outside is negligible. The unknown is the magnitude of the outward force perpendicular to the end; the tank wall must supply an opposing force to hold the air.

---

**The Math**

The pressure measures the normal force per unit area, so multiplying it by the end area gives the total outward force. The circular end has radius $d/2$ and area $\pi(d/2)^2$. These geometric relationships express the force directly in terms of the given diameter.

$$
F_\perp=PA=P\pi\left(\frac{d}{2}\right)^2.
$$

The pressure acts over the entire circular end. We evaluate the area using half the given diameter and multiply by the measured pressure to determine the outward force.

$$
F_\perp=(6.9\times10^6\ {\rm N/m^2})\pi\left(\frac{0.15\ {\rm m}}{2}\right)^2\approx1.2\times10^5\ {\rm N}.
$$

---

**The Conclusion**

The air exerts $1.2\times10^5\ {\rm N}$ outward on the flat end. A high pressure acting over even this modest area produces a large force, which explains why a compressed-air tank requires strong walls.
````
`````

## Variation of Pressure with Depth in a Fluid

At the Earth's surface, the air pressure exerted on you is a result of the weight of air above you.  This pressure is reduced as you climb in altitude and the weight of the air above you decreases.  Under water, the pressure exerted on you increases as you descend into deeper waters.  In this case, the pressure exerted on you is a result of both the weight of water and that of the atmosphere above you.

You may notice an air pressure change on an elevator ride in a very tall building, but you need only dive a meter (or so) below the surface of a pool to feel a pressure increase.  The difference is that water is much denser than air (about $775\times$ as dense).

Consider a container of water with vertical sides, where its bottom supports the weight of the water in it.  We can calculate the pressure exerted on the bottom by the weight of the water.  The weight of water is determined as the mass of the water $m$ times the acceleration due to gravity $g$, or $W= mg$.  This is the force exerted that is perpendicular to the area $A$ at the base of the container.  Then the pressure is calculated directly as

\begin{align}
P = \frac{F_\perp}{A} = \frac{mg}{A}.
\end{align}

We can find the mass of the water from its volume and density:

\begin{align}
m = \rho V, 
\end{align}

where the volume $V$ is the column of fluid above the surface.  Thus, it is natural to use the volume of a cylinder, or

\begin{align}
V = Ah,
\end{align}

where $A$ is the area of a circle and $h$ is the depth (height below the surface of the liquid).  Combining these expressions, we find that the mass is given by

\begin{align}
m = \rho Ah.
\end{align}

We can then substitute this mass into our expression for the weight of the fluid and thereby into the equation for the pressure, or

\begin{align}
P = \frac{(\rho A h) g}{A} = \rho g h,
\end{align}

where the area $A$ canceled.  This is the pressure due to *the weight of a fluid*.  This equation for pressure is generally applicable to static fluids.  For liquids, this equation holds to great depths because liquids are nearly incompressible.  For gases, one can apply this equation as long as the density changes are small over the depth considered because they are quite compressible.

```{figure-md} fig-pressure-water-column
<img src="pressure-water-column.png" alt="Two panels show pressure produced by fluid weight. Panel a shows a water column of depth h and area A, with surface pressure P_0 and downward weight mg. Air density decreases with altitude." style="width: 1000px; max-width: 100%; height: auto;">

```

*Atmospheric pressure* is an example of pressure due to the weight of a fluid (i.e., air) above a given height.  The atmospheric pressure at the Earth's surface varies a little due to the large-scale flow of the atmosphere, which creates regions of high and low pressure for weather.  The average pressure at sea level is given by the *standard atmospheric pressure* $P_{\rm atm}$, which is given as

\begin{align}
1\ \text{atmosphere (atm)} &= P_{\rm atm}, \\
P_{\rm atm} &= 1.01 \times 10^5\ {\rm N/m^2} = 101\ {\rm kPa}.
\end{align}

On average (at sea level) a column of air above 1-${\rm m^2}$ of Earth's surface has a weight of $101\ {\rm kN}$, or $1.01\times 10^5\ {\rm N}$.

- *What is the total pressure at a depth of $10.3\ {\rm m}$ in a swimming pool?*  
- *Does the atmospheric pressure on the water's surface affect the pressure below?*

Yes.  Both the water's weight and the atmosphere's weight must be supported.  So the *total pressure* at a depth of $10.3\ {\rm m}$ is 

\begin{align*}
P &= \rho_wgh_w + P_{\rm atm}, \\
&= (1000\ {\rm kg/m^3})(9.81\ {\rm m/s^2})(10.3\ {\rm m}) + 1.01 \times 10^5\ {\rm N/m^2} = 2.02 \times 10^5\ {\rm N/m^2} = \boxed{2\ {\rm atm}}.
\end{align*}

### Example Problem: Pressure and Force on a Dam

`````{admonition} Example Problem: Pressure and Force on a Dam
:class: etamu-exercise

**The Problem**

> The reservoir in the [previous density example](#ex-reservoir-mass) contains approximately $2.0\times10^{12}\ {\rm kg}$ of water. Its dam has a vertical face about $500\ {\rm m}$ wide, and the water at the dam is $80\ {\rm m}$ deep.
>
> **(a)** Estimate the average pressure on the submerged face due to the water.
> **(b)** Estimate the horizontal force on the dam and compare it with the weight of all the water in the reservoir. Use $\rho_{\rm w}=1.000\times10^3\ {\rm kg/m^3}$.

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

We model the submerged face as a vertical rectangle and the water as stationary with uniform density. Let depth $h$ increase downward from the water surface. The water's gauge pressure increases linearly from zero at the surface to its largest value at the bottom, so its average is the pressure halfway down. Atmospheric pressure acts on both sides of the dam and cancels in the net horizontal force. The reservoir bed supports most of the water's weight.

```{figure-md} fig-dam-pressure
<img src="dam-pressure.png" alt="Cutaway of a concrete dam holding water 80 m deep against a vertical face 500 m wide. Horizontal pressure arrows grow longer with depth. A companion panel explains how to calculate average pressure, submerged area, and horizontal force." style="width: 1000px; max-width: 100%; height: auto;">

**Pressure and force on a dam.** Water pressure increases with depth, as indicated by the longer arrows near the bottom. For the rectangular submerged face, the average gauge pressure is $\bar P=\rho_{\rm w}gh/2$, and the area is $A=Lh$. The total horizontal force is therefore $F=\bar P A$. Atmospheric pressure acts on both sides and cancels in the net force. The illustration is not to scale. Adapted from [OpenStax, *College Physics 2e*, Figure 11.9](https://openstax.org/books/college-physics-2e/pages/11-4-variation-of-pressure-with-depth-in-a-fluid).
```

---

**The Math**

**(a)** The water pressure increases linearly from zero gauge pressure at the surface to its greatest value at the bottom. Equal increments of depth correspond to equal areas of the rectangular face, so the area-average pressure equals the mean of the top and bottom pressures.

$$
\bar P=\frac{0+\rho_{\rm w}gh}{2}=\rho_{\rm w}g\frac{h}{2}.
$$

This average is also the pressure at half the full depth. We therefore use the water density and a depth of $40\ {\rm m}$ to evaluate the average pressure on the dam.

$$
\bar P=(1.000\times10^3\ {\rm kg/m^3})(9.81\ {\rm m/s^2})\frac{80\ {\rm m}}{2}\approx3.9\times10^5\ {\rm Pa}.
$$

**(b)** The submerged area is the dam's width $L$ times the full depth $h$. Multiplying that area by the average pressure gives the horizontal force. The combined expression contains $h^2$ because a greater depth increases both the submerged area and the average pressure. We evaluate this expression directly to avoid rounding the pressure before calculating the force.

\begin{align*}
F &= \bar P Lh=\frac{\rho_{\rm w}gLh^2}{2}, \\
&= \frac{(1.000\times10^3\ {\rm kg/m^3})(9.81\ {\rm m/s^2})(500\ {\rm m})(80\ {\rm m})^2}{2}, \\
&\approx1.6\times10^{10}\ {\rm N}.
\end{align*}

The reservoir's weight is $W=mg$, whereas the force on the dam acts horizontally. Their ratio compares the magnitudes of these different forces; it does not represent a vertical force balance. The common factor $g$ cancels, allowing the comparison to be calculated from the reservoir mass and dam dimensions.

$$
\frac{F}{W}=\frac{\rho_{\rm w}Lh^2}{2m}=\frac{(1.000\times10^3\ {\rm kg/m^3})(500\ {\rm m})(80\ {\rm m})^2}{2(2.0\times10^{12}\ {\rm kg})}\approx8.0\times10^{-4}.
$$

---

**The Conclusion**

The average water pressure is about $3.9\times10^5\ {\rm Pa}$, and the horizontal force is about $1.6\times10^{10}\ {\rm N}$, or only $0.080\%$ of the reservoir's weight. The force depends on the water depth and the submerged face of the dam, not directly on the reservoir's total horizontal extent.
````
`````

*Adapted from [OpenStax, College Physics 2e, Example 11.3](https://openstax.org/books/college-physics-2e/pages/11-4-variation-of-pressure-with-depth-in-a-fluid).*


### Example Problem: Average Density of the Atmosphere

`````{admonition} Example Problem: Average Density of the Atmosphere
:class: etamu-exercise

**The Problem**

> For a simple estimate, treat the atmosphere as a column extending $120\ {\rm km}$ above Earth's surface. Find its average density using $P_{\rm atm}=1.01\times10^5\ {\rm Pa}$, and compare it with the tabulated near-surface air density of $1.29\ {\rm kg/m^3}$.

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

The atmosphere's density decreases with altitude, so we seek a column-average density rather than a uniform local density. We approximate gravity as constant and neglect pressure at the top of the column. The surface pressure then equals the weight of the column divided by its area. The assumed height is a useful cutoff for this estimate, not a sharp physical boundary of the atmosphere.

---

**The Math**

The surface pressure supports the weight of the atmospheric column above it. With gravity treated as constant, the mass per unit area is the column-average density multiplied by its height. The hydrostatic relation therefore connects surface pressure to that average density.

$$
P_{\rm atm}=\bar\rho gh.
$$

Dividing the surface pressure by $gh$ isolates the column-average density. We express the assumed height as $1.2\times10^5\ {\rm m}$ so that the pressure, gravitational acceleration, and height all use SI units.

$$
\bar\rho=\frac{P_{\rm atm}}{gh}=\frac{1.01\times10^5\ {\rm Pa}}{(9.81\ {\rm m/s^2})(1.2\times10^5\ {\rm m})}\approx0.086\ {\rm kg/m^3}.
$$

The near-surface density is about $1.29/0.086\approx15$ times this average.

---

**The Conclusion**

The average density is approximately $0.086\ {\rm kg/m^3}$, much less than the near-surface value. This is reasonable because gravity compresses the lower atmosphere, while most of the assumed column is occupied by much thinner air.
````
`````

### Example Problem: Water Depth Equivalent to One Atmosphere

`````{admonition} Example Problem: Water Depth Equivalent to One Atmosphere
:class: etamu-exercise

**The Problem**

> At what depth below a freshwater surface does the pressure due to the water alone equal one atmosphere? Use $P_{\rm atm}=1.01\times10^5\ {\rm Pa}$ and $\rho_{\rm w}=1.000\times10^3\ {\rm kg/m^3}$.

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

We treat the water as stationary and incompressible. Depth $h$ is measured downward from the surface. The requested pressure is the increase below the surface, so the atmosphere above the water is not included in the pressure difference used to find $h$.

---

**The Math**

The water produces a pressure increase of $\rho_{\rm w}gh$ relative to the surface. We require this increase to equal the specified atmospheric pressure. Dividing that pressure by $\rho_{\rm w}g$ determines the required depth; the atmospheric pressure already acting on the surface is not an additional term in this calculation.

$$
h=\frac{P_{\rm atm}}{\rho_{\rm w}g}=\frac{1.01\times10^5\ {\rm Pa}}{(1.000\times10^3\ {\rm kg/m^3})(9.81\ {\rm m/s^2})}=10.3\ {\rm m}.
$$

---

**The Conclusion**

A water depth of $10.3\ {\rm m}$ produces a pressure increase of one atmosphere. The much shorter water column compared with the atmospheric column reflects water's much greater density. If the water surface is open to the atmosphere, the absolute pressure at this depth is approximately two atmospheres.
````
`````

*Adapted from [OpenStax, College Physics 2e, Example 11.5](https://openstax.org/books/college-physics-2e/pages/11-4-variation-of-pressure-with-depth-in-a-fluid).*


## Pascal's Principle

*Can pressure be increased in a fluid by pushing directly on the fluid?*

Yes, but it is much easier if the fluid is enclosed.  For example, the heart increases blood pressure by pushing directly on the blood in an enclosed system (e.g., valves closed in a chamber).  If you try to push on a fluid in an open system (e.g., a river), the fluid flows away. An enclosed fluid cannot flow away, so pressure is more easily increased by an applied force.

*What happens to a pressure in an enclosed fluid?*

Because atoms in a fluid are free to move about (and rotate), they transmit a pressure change to all parts of the fluid and to the walls of the container.  The pressure is transmitted *undiminished*, or $\Delta P_{\rm in} = \Delta P_{\rm out}$.  This phenomenon is called **Pascal's principle** because it was first clearly stated by the French philosopher [Blaise Pascal](https://en.wikipedia.org/wiki/Blaise_Pascal).

```{admonition} Pascal's Principle
:class: note

A change in pressure applied to an enclosed fluid is transmitted undiminished to all portions of the fluid and to the walls of the container.  Mathematically, this is represented by 

\begin{align}
\Delta P_{\rm in} = \Delta P_{\rm out}.
\end{align}
```

Pascal's principle is important in fluids since a change in pressure is transmitted undiminished in an enclosed fluid.  It implies that the *total pressure in a fluid is the sum of the pressures from different sources*, meaning that pressures add.

### Application of Pascal's Principle

Pascal's principle is found in a hydraulic system, which is an enclosed system used to exert forces.  The most common hydraulic systems are those that operate car brakes.

```{figure-md} fig-pascal-hydraulic-system
<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/c816f88b0b80fa529fe915607589fc2a4d3e8b1b" alt="Two fluid-filled hydraulic cylinders connected by a tube. A downward force on the smaller piston creates the same pressure in the fluid and a larger upward force on the larger piston." style="width: 760px; max-width: 100%; height: auto;">

**Force multiplication in a hydraulic system.** A pressure increase produced at the smaller piston is transmitted through the enclosed fluid. Because the second piston has a larger area, the same pressure produces a larger output force. Source: [OpenStax, *College Physics 2e*, Figure 11.11](https://openstax.org/books/college-physics-2e/pages/11-5-pascals-principle).
```


Consider two fluid-filled pistons where each piston has an area $A_1$ and $A_2$ at the top.  The two pistons are at the same height so that there will be no difference in pressure due to a difference in depth.  The pressure due to $F_1$ acts on the area $A_1$ of the first piston and is defined as

$$ P_1 = \frac{F_1}{A_1}. $$

According to Pascal's principle, this pressure is transmitted undiminished throughout the fluid and to the walls of the container.  Thus, a pressure $P_2$ is felt at the other piston that is equal to $P_1$, and defined by

$$ P_2 = \frac{F_2}{A_2}. $$

Since $P_1 = P_2$, we can see that

$$ \frac{F_1}{A_1} = \frac{F_2}{A_2}. $$

This equation relates the ratios of force to area in any hydraulic system, providing the pistons are at the same vertical height and that friction in the system is negligible.

Hydraulic systems can increase or decrease the force applied to them.  To make the force larger, the pressure is applied to a larger area.  For example, if a $100\ {\rm N}$ force is applied to the smaller cylinder and the larger one has an area $5\times$ greater, then the force out is $500\ {\rm N}$.  More explicitly, we can show

$$ F_2 = \frac{A_2}{A_1}F_1 = \left(\frac{5A_1}{A_1}\right)(100\ {\rm N}) = 500\ {\rm N}. $$

Hydraulic systems are analogous to simple levers, but they have the advantage that pressure can be sent through complicated, curved lines to several places at once.




A simple hydraulic system (e.g., a simple machine) can increase the force but cannot do more work than is done on it.  Work is force times distance moved, where the output cylinder usually moves very little despite exerting a greater force.  Many hydraulic systems have a motorized pump that actually does most of the work in the system.  The movement of a spider's leg is achieved partly by hydraulics. A jumping spider can use this hydraulic mechanism to generate the force needed for rapid jumps.

### Example Problem: Force from Hydraulic Brakes

`````{admonition} Example Problem: Force from Hydraulic Brakes
:class: etamu-exercise

**The Problem**

> A driver applies $100\ {\rm N}$ to a brake pedal. The pedal's lever increases this to $500\ {\rm N}$ on the pedal cylinder. The pedal-cylinder diameter is $0.50\ {\rm cm}$, and each of four wheel cylinders has a diameter of $2.5\ {\rm cm}$. What force does the fluid exert at each wheel cylinder?

```{figure-md} fig-hydraulic-brakes
<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/07f682ab2c594e0da0bde036df2651f85f19ff53" alt="A brake pedal and lever drive a small pedal cylinder connected by fluid lines to larger wheel cylinders." style="width: 900px; max-width: 100%; height: auto;">

**Hydraulic braking.** The pedal lever increases the input force, and the enclosed fluid transmits the pressure increase to each wheel cylinder. The circular cylinder areas are labeled $A_1$ and $A_2$. Source: [OpenStax, *College Physics 2e*, Figure 11.12](https://openstax.org/books/college-physics-2e/pages/11-5-pascals-principle).
```

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

We model the brake fluid as an enclosed, incompressible fluid and neglect friction and differences in cylinder height. Pascal's principle gives the same pressure increase at the pedal cylinder and at every wheel cylinder. The input force for the hydraulic calculation is the force on the pedal cylinder, not the driver's smaller force on the lever.

---

**The Math**

The enclosed fluid transmits the input pressure increase to each wheel cylinder. The force per unit area is therefore the same at the input and output. The output force equals the input force multiplied by the ratio of cylinder areas.

$$
\frac{F_1}{A_1}=\frac{F_2}{A_2},\qquad F_2=F_1\frac{A_2}{A_1}.
$$

For circular cylinders, the area is $\pi d^2/4$. The common factor $\pi/4$ cancels from the area ratio, leaving the square of the diameter ratio. The wheel-cylinder diameter is five times the pedal-cylinder diameter, so the hydraulic system multiplies the pedal-cylinder force by a factor of 25.

$$
F_2=F_1\left(\frac{d_2}{d_1}\right)^2=(500\ {\rm N})\left(\frac{2.5\ {\rm cm}}{0.50\ {\rm cm}}\right)^2\approx1.3\times10^4\ {\rm N}.
$$

---

**The Conclusion**

Each wheel cylinder exerts approximately $1.3\times10^4\ {\rm N}$. The pressure is transmitted to every cylinder, so the output force is not divided by four. This force increase does not create energy: supplying fluid to all four cylinders requires a correspondingly larger input displacement.
````
`````

## Gauge Pressure, Absolute Pressure, and Pressure Measurement

When you have a flat tire, the tire gauge reads nearly zero when you begin to fill it.  If there were a hole in your tire, the gauge would read zero even though atmospheric pressure exists in the tire.  Tire gauges are designed to read zero at atmospheric pressure and positive when the pressure is greater than atmospheric.

Atmospheric pressure adds to blood pressure in every part of the circulatory system.  But atmospheric pressure has no effect on blood flow since it adds to the pressure coming out of the heart and going back into it too. Blood pressure measurements are made relative to atmospheric pressure like tire pressure.

It is very common for pressure gauges to ignore atmospheric pressure, where they read zero at atmospheric pressure.  We define *gauge pressure* as the pressure relative to atmospheric pressure.

```{admonition} Gauge vs Absolute Pressure
:class: note

**Gauge pressure** $P_{\rm G}$ is the pressure relative to atmospheric pressure.  Gauge pressure is positive for pressures above atmospheric pressure and negative for pressures below it.

**Absolute pressure**, or total pressure, $P_{\rm abs}$ is the sum of gauge and atmospheric pressure. Mathematically, we can relate absolute pressure to gauge pressure by

\begin{align}
P_{\rm abs} = P_{\rm G} + P_{\rm atm}.
\end{align}

For example, if your tire gauge reads $34\ {\rm psi}$ (pounds per square inch), then the absolute pressure is $34\ {\rm psi} + 14.7\ {\rm psi} = 48.7\ {\rm psi}$ and is equivalent to  $336\ {\rm kPa}$.
```

In most cases, the absolute pressure in fluids cannot be negative.  Fluids push rather than pull, so the smallest absolute pressure is zero. The smallest possible gauge pressure is $P_{\rm G} = -P_{\rm atm}$ so that $P_{\rm abs} = 0$.  There is no theoretical limit to how large a gauge pressure can be.

Pressure can be measured using tire gauges as well as blood pressure cuffs.  Pascal's principle describes the undiminished transmission of a pressure change through a fluid, which allows for precise remote sensing of pressures.  Remote sensing is more convenient than putting a measuring device into a person's artery.

Mechanical pressure gauges measure pressure through the force transmitted to a secondary device (e.g., flexible bellows).  A U-shaped tube is called a *manometer*, which can have both sides open to the atmosphere.  Atmospheric pressure pushes down on each side so its effect cancels.  If the fluid is deeper on one side, there is a greater pressure on that side.  The fluid flows away from the deeper side until the depths are equal.

```{figure-md} fig-open-tube-manometers
<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/56ab4a55d47c4bde3b616b70f5b831224a0c3bfa" alt="Three open-tube manometers showing equal fluid levels, positive gauge pressure from a balloon, and negative gauge pressure from a vacuum-packed jar." style="width: 1000px; max-width: 100%; height: auto;">

**Open-tube manometers.** Equal atmospheric pressure on both sides gives equal fluid levels. A source pressure above atmospheric pressure raises the opposite column, while a pressure below atmospheric pressure produces the opposite displacement. Source: [OpenStax, *College Physics 2e*, Figure 11.14](https://openstax.org/books/college-physics-2e/pages/11-6-gauge-pressure-absolute-pressure-and-pressure-measurement).
```


Suppose one side of the manometer is connected to some source of absolute pressure (e.g., a toy balloon or a vacuum-packed peanut jar).  Pressure is transmitted undiminished to the manometer, and the fluid levels are no longer equal.  In both cases, the absolute pressure differs from the atmospheric pressure by an amount $h\rho g$, where $\rho$ is the density of the fluid in the manometer.

- In the case of the toy balloon, the absolute pressure is greater than atmospheric pressure ($P_{\rm abs}>P_{\rm atm}$).  The manometer can support a column of fluid with a height $h$ and so it must exert a pressure $h\rho g$ greater than atmospheric pressure ($P_{\rm G}>0$).
- For the peanut jar, the absolute pressure is less than atmospheric pressure ($P_{\rm abs}<P_{\rm atm}$).  The atmospheric pressure can support a column of fluid with a height $h$ and so $P_{\rm abs}<P_{\rm atm}$ by an amount $h\rho g$, which means that the gauge pressure is negative ($P_{\rm G}<0$).

Mercury manometers are often used to measure arterial blood pressure.  An inflatable cuff is placed on the upper arm.  By squeezing the bulb, the person making the measurement exerts pressure, which is transmitted undiminished to both the main artery in the arm and the manometer.  When this applied pressure exceeds blood pressure, blood flow below the cuff is cut off.  The person making the measurement then slowly lowers the applied pressure and listens for blood flow to resume.  Blood pressure pulsates because of the pumping action of the heart reaching a maximum (**systolic pressure**) and a minimum (**diastolic pressure**) with each heartbeat.

Systolic pressure is measured by noting the value of $h$ when blood flow first begins as the cuff pressure is lowered.  Diastolic pressure is measured by noting $h$ when blood flows without interruption.  The typical blood pressure of a young adult raises the mercury to a height of $120\ {\rm mm}$ at systolic and $80\ {\rm mm}$ at diastolic, where this is commonly quoted as $120/80$.  

The first pressure represents the maximum output of the heart, while the second is due to the elasticity of the arteries in maintaining the pressure between beats.  The density of mercury is $13.6$ times that of water, so the height of the mercury column will be $1/13.6$ of that in a water manometer for the same pressure difference.  




A *barometer* is a device that measures atmospheric pressure.  This device measures atmospheric pressure instead of gauge pressure because there is a nearly pure vacuum above the mercury in the tube.  The height of the mercury is such that $h\rho g = P_{\rm atm}$.  When atmospheric pressure varies, the mercury rises or falls, which can give important clues to weather forecasters.


*Adapted from [OpenStax, College Physics 2e, Example 11.7](https://openstax.org/books/college-physics-2e/pages/11-6-gauge-pressure-absolute-pressure-and-pressure-measurement).*


## Archimedes' Principle

*Why do some things float while others do not?* 

This is partly answered by the fact that pressure increases with depth in a fluid.  This means that the upward force on the bottom of an object (in a fluid) is greater than the downward force on the top of the object.  There is a net upward, or **buoyant force**, on any object in any fluid.  

- If the buoyant force is greater than the object's weight, then the object will rise to the surface and float.
- If the buoyant force is less than the object's weight, the object will sink.
- If the buoyant force equals the object's weight, the object will remain suspended at that depth. 

The buoyant force is always present whether the object floats, sinks, or is suspended in a fluid.

```{figure-md} fig-buoyancy-pressure-displacement
<img src="fig11_18-11_19_buoyancy-combined.png" alt="Side-by-side diagrams showing that fluid pressure produces a larger upward force on the bottom of a submerged object and that the buoyant force equals the weight of the displaced fluid." style="width: 1100px; max-width: 100%; height: auto;">

**Why a buoyant force acts and what sets its magnitude.** **(a)** Because pressure increases with depth, the upward force on the bottom of a submerged object is greater than the downward force on its top. **(b)** Replacing the object with the fluid it displaced shows that the buoyant force equals the weight of that displaced fluid. Adapted from [OpenStax, *College Physics 2e*, Figures 11.18 and 11.19](https://openstax.org/books/college-physics-2e/pages/11-7-archimedes-principle), combined into one side-by-side figure.
```


Consider what happens when a submerged object is removed from a fluid.  The space occupied is filled by fluid having a weight $w_{\rm fl}$.  This weight is supported by the surrounding fluid so that the buoyant force must equal $w_{\rm fl}$, the weight of the fluid displaced by the object.  

The Greek mathematician and inventor [Archimedes](https://en.wikipedia.org/wiki/Archimedes) stated this principle long before concepts of force were established.  The king wanted to know whether a wreath he commissioned was pure gold or simply covered in gold and wanted a process that would not  destroy the wreath in the process.  Archimedes recognized that a fully submerged wreath displaces a volume of fluid equal to its own volume.  From there, he could also measure the volume of the fluid displaced and determine the submerged wreath's density.  Since gold is more dense than iron, this was an ingenious way to determine the authenticity of the golden wreath.

```{admonition} Archimedes' Principle
:class: note

According to this principle, the buoyant force on an object equals the weight of the fluid it displaces.  In equation form, Archimedes' principle is 

\begin{align}
F_{\rm B} = w_{\rm fl} = m_{\rm fl}g,
\end{align}

where $F_{\rm B}$ is the buoyant force and $w_{\rm fl}$ is the weight of the fluid displaced by the object.
```




### Density and Archimedes' Principle
Density plays a critical role in Archimedes' principle.  The average density of an object is what ultimately determines whether it floats.  If its average density is less than that of the surrounding fluid, it will float.

This is because the fluid (having higher density) contains more mass and hence more weight in the same volume.  While the object is fully submerged, the buoyant force is thus greater than its weight. Once it floats at rest, it displaces less fluid and the buoyant force equals its weight.  Likewise, an object denser than the fluid will sink.

The extent to which a floating object is submerged depends on how the object's density is related to that of the fluid.  For example, an unloaded ship has a lower density and less of it is submerged compared to the same ship fully loaded.

```{figure-md} fig-loaded-unloaded-ship
<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/795a6e46af465085ac69389990e323bd998d10f1" alt="An unloaded ship floating higher in the water beside a loaded cargo ship floating lower in the water." style="width: 900px; max-width: 100%; height: auto;">

**Loading changes the submerged fraction of a floating ship.** Adding cargo increases the ship's average density, so the ship must displace a larger volume of water before the buoyant force balances its weight. Source: [OpenStax, *College Physics 2e*, Figure 11.20](https://openstax.org/books/college-physics-2e/pages/11-7-archimedes-principle).
```
  The fraction submerged is the ratio of the volume submerged to the volume of the object, or

\begin{align}
\text{fraction submerged} = \frac{V_{\rm sub}}{V_{\rm obj}} = \frac{V_{\rm fl}}{V_{\rm obj}}.
\end{align}

The volume submerged $V_{\rm sub}$ equals the volume of the fluid displaced $V_{\rm fl}$.  We can obtain the relationship between the densities by substituting $\rho = m/V$ into the equation, which gives

\begin{align}
\frac{V_{\rm fl}}{V_{\rm obj}} = \left(\frac{m_{\rm fl}}{\rho_{\rm fl}}\right) \left(\frac{\bar{\rho}_{\rm obj}}{m_{\rm obj}}\right),
\end{align}

where $\bar{\rho}_{\rm obj}$ is the average density of the object and $\rho_{\rm fl}$ is the density of the fluid.  Since the object floats, its mass and that of the displaced fluid are equal ($m_{\rm fl}=m_{\rm obj}$). This results in

\begin{align}
\text{fraction submerged} = \frac{\bar{\rho}_{\rm obj}}{\rho_{\rm fl}}.
\end{align}




We use this last relationship to measure densities.  We can measure the fraction of a floating object that is submerged (e.g., with a hydrometer).  It is now useful to define the ratio of the average density of an object $\bar{\rho}$ to a fluid (usually water) as **specific gravity**:

\begin{align}
\text{specific gravity} = \frac{\bar{\rho}}{\rho_{\rm w}},
\end{align}

where $\bar{\rho}$ is the average density of the object and $\rho_{\rm w}$ is the density of water at $4.0^\circ{\rm C}$. Specific gravity is dimensionless.

- If an object floats in water, its specific gravity is less than one.
- If it sinks in water, its specific gravity is greater than one.

The fraction of an object floating in water that is submerged equals its specific gravity. If its specific gravity is exactly 1, it can remain suspended at rest in water.

Scuba divers try to obtain this state so they can hover in the water.  We measure the specific gravity of fluids (e.g., battery acid, radiator fluid, and urine) as an indicator of their condition.

### More Density Measurements

An object is weighed in air and then weighed again while submerged in a liquid.  The density of the object (here a coin) can be calculated if the fluid density is known.  This same technique can also be used to determine the density of the fluid if the density of the object is known.  All of these calculations are based on Archimedes' principle.

Archimedes' principle means that the object appears to weigh less when submerged.  We call this measurement: the object's *apparent weight*.  The object suffers an apparent *weight loss* equal to the weight of the fluid displaced.  Under constant gravity, a balance can measure the mass, where the object suffers an *apparent mass loss* equal to the mass of the fluid displaced:

\begin{align}
\text{apparent weight loss} &= \text{weight of fluid displaced}, \\
\text{apparent mass loss} &= \text{mass of fluid displaced}.
\end{align}



### Example Problem: Solid Steel and a Floating Ship

`````{admonition} Example Problem: Solid Steel and a Floating Ship
:class: etamu-exercise

**The Problem**

> A mass of $1.0\times10^7\ {\rm kg}$ of steel is placed in freshwater. Use $\rho_{\rm st}=7.8\times10^3\ {\rm kg/m^3}$ and $\rho_{\rm w}=1.000\times10^3\ {\rm kg/m^3}$.
>
> **(a)** Find the buoyant force when the steel is a solid, fully submerged block, and compare it with the steel's weight.
> **(b)** Find the maximum buoyant force if the same steel forms a boat capable of displacing $1.0\times10^5\ {\rm m^3}$ of water.

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

Archimedes' principle relates the upward buoyant force to the weight of displaced water. Define the vertical $y$-axis with $+y$ upward. For the solid block, the displaced volume equals the steel's volume. For the boat, the hull encloses space as well as steel and can displace a much larger volume. We distinguish the boat's maximum possible buoyant force from the smaller force needed to support an unloaded boat at rest.

---

**The Math**

**(a)** A solid block displaces water only from the volume occupied by the steel itself. The steel density specifies the mass in each cubic meter, so dividing the given mass by that density determines the displaced volume.

$$
V_{\rm st}=\frac{m}{\rho_{\rm st}}=\frac{1.0\times10^7\ {\rm kg}}{7.8\times10^3\ {\rm kg/m^3}}\approx1.28\times10^3\ {\rm m^3}.
$$

Full submersion makes $V_{\rm fl}=V_{\rm st}$. Archimedes' principle identifies the upward buoyant force with the weight of this displaced water. We therefore multiply the steel volume by the water density and gravitational acceleration; the water density, rather than the steel density, determines the weight of the displaced fluid.

$$
F_{\rm B}=\rho_{\rm w}g\frac{m}{\rho_{\rm st}}=(1.000\times10^3\ {\rm kg/m^3})(9.81\ {\rm m/s^2})\frac{1.0\times10^7\ {\rm kg}}{7.8\times10^3\ {\rm kg/m^3}}\approx1.3\times10^7\ {\rm N}.
$$

The competing downward force is the weight of the steel itself. This force depends on the actual steel mass and gravitational acceleration, regardless of the surrounding water.

$$
W=mg=(1.0\times10^7\ {\rm kg})(9.81\ {\rm m/s^2})\approx9.8\times10^7\ {\rm N}.
$$

The buoyant force is only about $13\%$ of the weight, so an unsupported block accelerates downward.

**(b)** The boat's hull can exclude water from a much larger volume than the solid steel occupies. Its maximum buoyant force equals the weight of the greatest volume of water it can displace. We use the stated displacement capacity in Archimedes' relation to determine this limiting force.

\begin{align*}
F_{{\rm B},\max} &= \rho_{\rm w}gV_{{\rm fl},\max}, \\
&= (1.000\times10^3\ {\rm kg/m^3})(9.81\ {\rm m/s^2})(1.0\times10^5\ {\rm m^3}), \\
&=9.8\times10^8\ {\rm N}.
\end{align*}

---

**The Conclusion**

The solid steel receives an upward force of $1.3\times10^7\ {\rm N}$ and sinks because its weight is larger. The boat can receive up to $9.8\times10^8\ {\rm N}$, ten times the steel's weight, by displacing much more water. An unloaded floating boat has $F_{\rm B}=W$; the additional displacement capacity allows it to support cargo.
````
`````


*Adapted from [OpenStax, College Physics 2e, Example 11.9](https://openstax.org/books/college-physics-2e/pages/11-7-archimedes-principle).*


### Example Problem: Density of an Ancient Coin

`````{admonition} Example Problem: Density of an Ancient Coin
:class: etamu-exercise

**The Problem**

> An ancient Greek coin has a measured mass of $8.630\ {\rm g}$ in air. Its apparent mass is $7.800\ {\rm g}$ when fully submerged in water of density $1.000\ {\rm g/cm^3}$. Find its average density and compare it with silver. Neglect air buoyancy and the effects of the suspending wire.

```{figure-md} fig-coin-apparent-weight
<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/53b6bbc3a511ac8693d5dc8e27ca28d8af22376d" alt="Two measurements of a suspended coin: in air and completely submerged in a container of liquid." style="width: 700px; max-width: 100%; height: auto;">

**Weighing a coin in air and water.** The supporting force decreases when the coin is submerged because the liquid supplies an upward buoyant force. The coin is fully immersed without resting on the container. Source: [OpenStax, *College Physics 2e*, Figure 11.23](https://openstax.org/books/college-physics-2e/pages/11-7-archimedes-principle).
```

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

The coin's actual mass does not change when it is submerged. The balance instead registers a smaller supporting force because water supplies an upward buoyant force. The apparent mass loss equals the mass of displaced water. Full submersion makes the volume of that water equal to the coin's volume, allowing the unknown density to be found. The measured balance readings retain their stated precision.

---

**The Math**

The balance reading decreases because the buoyant force supplies part of the support previously supplied by the balance. Expressed as an apparent mass change, this decrease equals the mass of displaced water. Subtracting the submerged reading from the reading in air determines that water mass.

$$
m_{\rm fl}=m-m_{\rm app}=8.630\ {\rm g}-7.800\ {\rm g}=0.830\ {\rm g}.
$$

The coin is fully submerged, so its volume equals the displaced water volume. Dividing the displaced water mass by the known water density gives this common volume.

$$
V_{\rm coin}=V_{\rm fl}=\frac{m_{\rm fl}}{\rho_{\rm w}}=\frac{0.830\ {\rm g}}{1.000\ {\rm g/cm^3}}=0.830\ {\rm cm^3}.
$$

The coin density must use its actual mass, not the smaller apparent mass recorded underwater. Dividing the mass measured in air by the volume inferred from buoyancy determines its average density.

$$
\bar\rho_{\rm coin}=\frac{m}{V_{\rm coin}}=\frac{8.630\ {\rm g}}{0.830\ {\rm cm^3}}=10.4\ {\rm g/cm^3}.
$$

---

**The Conclusion**

The coin's average density is $10.4\ {\rm g/cm^3}$, close to the tabulated silver density of $10.49\ {\rm g/cm^3}$. The result is consistent with silver, but density alone does not establish authenticity or exclude an alloy with a similar average density.
````
`````

## Surface Tension and Capillary Action

Attractive forces between molecules of the same type are called **cohesive forces**.  Liquids can be held in open containers because cohesive forces hold the molecules together.  Attractive forces between molecules of different types are called **adhesive forces**.

### Surface Tension

Cohesive forces between molecules cause the surface of a liquid to contract to the *smallest possible surface area*.  This general effect is called **surface tension**.  Molecules on the surface are pulled inward by cohesive forces, reducing the surface area.  In the bulk liquid, cohesive attractions act in all directions and cancel on average.

We can explain surface tension effects using a model of a liquid surface acting like a stretched elastic sheet.  For example, some insects can walk on water as we would walk on a trampoline.  

A needle can rest on a water surface.  The iron needle cannot float because its density is greater than that of water.  Its weight is supported by forces in the stretched surface that try to make the surfaces smaller or flatter.  If the needle were placed point down on the surface, its weight acting on a smaller area would break the surface, and it would sink.

```{figure-md} fig-surface-tension-support
<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/1bed78401560e01c5bc4873f2b0a512e6c31e269" alt="An insect leg and an iron needle deforming the water surface, with surface-tension force components supporting their weight." style="width: 1000px; max-width: 100%; height: auto;">

**Surface tension can support objects that are not buoyantly floating.** The deformed liquid surface pulls along the interface; the upward components of those surface-tension forces can balance the object's weight. Source: [OpenStax, *College Physics 2e*, Figure 11.25](https://openstax.org/books/college-physics-2e/pages/11-8-cohesion-and-adhesion-in-liquids-surface-tension-and-capillary-action).
```


Surface tension is proportional to the strength of the cohesive force, which varies with the type of liquid.  Surface tension $\gamma$ is defined as the force $F$ per unit length $L$ exerted by a stretched liquid membrane:

\begin{align}
\gamma = \frac{F}{L}.
\end{align}

Surface tension is the reason why liquids form bubbles and droplets.  The inward surface tension force causes bubbles to be approximately spherical and raises the pressure of the gas trapped inside relative to atmospheric pressure outside.  The gauge pressure $P$ inside a spherical soap bubble with two liquid–air surfaces is given as

\begin{align}
P = \frac{4\gamma}{r},
\end{align}

where $r$ is the radius of the bubble.  For fixed surface tension, the pressure difference is greatest when the bubble is smallest. A liquid droplet or an idealized alveolus has one liquid–air interface, for which the pressure difference is $2\gamma/r$, rather than $4\gamma/r$.




Our lungs contain hundreds of millions of mucus-lined sacs called *alveoli*, which are very similar in size and about $0.1\ {\rm mm}$ in diameter. You can exhale without muscle action by allowing surface tension to contract these sacs.  Medical patients whose breathing is aided by a respirator have air blown into the lungs, but are generally allowed to exhale on their own.

```{figure-md} fig-alveoli-surface-tension
**Alveoli and pulmonary capillaries.** The small air sacs at the ends of the bronchioles are lined by fluid and behave in some ways like tiny bubbles. Even if there is paralysis, surface tension in the alveoli will expel air from the lungs.  Since pressure increases as the radii of the alveoli decrease, an occasional deep cleansing breath is needed to fully reinflate the alveoli.
```

The tension in the walls of the alveoli results from the membrane tissue and a liquid on the walls of the alveoli containing a mixture of lipids and proteins that acts as a surfactant (i.e., surface-tension reducing substance).  The need for the surfactant results from the tendency of small alveoli to collapse and the air to fill into the larger alveoli making them even larger.

- During inhalation, the lipoprotein molecules are pulled apart and the wall tension increases as the radius increases (i.e., increased surface tension).  
- During exhalation, the molecules slide back together and the surface tension decreases, helping to prevent a collapse of the alveoli.

The surfactant serves to change the wall tension so that the small alveoli don't collapse and large alveoli are prevented from expanding too much.  The tension change is a unique property of the surfactants, and is not shared by detergents, which simply lower surface tension.

If water gets into the lungs, the surface tension is too great and you cannot inhale.  This is a problem in resuscitating drowning victims and newborn infants who are born without this surfactant.  This surfactant deficiency can cause *neonatal respiratory distress syndrome*, historically called hyaline membrane disease, especially in premature infants. See the [NHLBI explanation of respiratory distress syndrome](https://www.nhlbi.nih.gov/health/respiratory-distress-syndrome).  

Emphysema produces the opposite problem with alveoli.  Alveolar walls of emphysema patients deteriorate, and the sacs combine to form larger sacs.  Because pressure produced by surface tension decreases with increasing radius, these larger sacs produce smaller pressure, reducing their ability to exhale.  A common test for emphysema is to measure the pressure and volume of air that can be exhaled.

### Adhesion and Capillary Action

*Why is it that water beads up on a waxed car but does not on bare paint?*

Adhesive forces between wax and water are much smaller than those between water and paint.  Competition between forces of adhesion and cohesion is important in the macroscopic behavior of liquids.  The roles of these two forces depend on the angle $\theta$ between the tangent to the liquid surface and the solid surface, measured through the liquid.  The **contact angle** $\theta$ is directly related to the relative strength of the cohesive and adhesive forces.

- When the cohesive force is stronger relative to the adhesive force, $\theta$ is larger and the liquid tends to form a droplet.
- When adhesion is stronger relative to cohesion, $\theta$ is smaller and the adhesive force flattens the drop.

The competition between cohesive and adhesive forces produces **capillary action**, which is the tendency of a fluid to be raised or suppressed in a narrow tube, or *capillary tube*.  This action causes blood to be drawn into a small-diameter tube when the tube touches a drop.

If a capillary tube is placed vertically into a liquid, capillary action will raise or suppress the liquid inside the tube depending on the combination of substances.  The actual effect depends on the relative strength of the cohesive and adhesive forces (i.e., the contact angle).

- If $\theta <90^\circ$, then the fluid will be raised.
- If $\theta > 90^\circ$, then it will be suppressed.

For example, mercury has a very large surface tension and a large contact angle with glass.  When it is placed in a tube, the surface of a column of mercury curves downward, somewhat like a drop.  The curved surface of a fluid in a tube is called a **meniscus**.  The tendency of surface tension is always to reduce the surface area.  Surface tension thus flattens the curved liquid surface in a capillary tube.  This results in a downward force in mercury and an upward force in water.

```{figure-md} fig-capillary-rise-suppression

**Capillary suppression and rise.** Mercury has a contact angle greater than $90^\circ$ with glass, so surface tension contributes a downward force and suppresses the column. Water has a contact angle near $0^\circ$, so surface tension contributes an upward force and raises the column. Source: [OpenStax, *College Physics 2e*, Figure 11.31](https://openstax.org/books/college-physics-2e/pages/11-8-cohesion-and-adhesion-in-liquids-surface-tension-and-capillary-action).
```


Capillary action can move liquids horizontally over very large distances, but the height to which it can raise (or suppress) a liquid is limited by its weight.  This is shown mathematically by

\begin{align}
h = \frac{2\gamma}{\rho g r}\cos{\theta}.
\end{align}

There are some insights we can gain from examining this expression.

- The height $h$ is directly proportional to the surface tension $\gamma$, which is its direct cause.  
- The height is inversely proportional to the tube radius. The smaller the radius $r$, the higher the fluid can be raised since a smaller tube holds less mass.
- The height is also inversely proportional to the fluid density $\rho$, since a larger density means a greater mass in the same volume.




*How does sap get to the tops of tall trees?*

It appears that it is pulled up like a chain held together by cohesive forces.  As each molecule of sap enters a leaf and evaporates, the entire chain is pulled up a notch.  A negative pressure created by water evaporation must be present to pull the sap up through the xylem vessels.  In most situations, *fluids can push but can exert a negligible pull* because the cohesive forces seem to be too small to hold the molecules tightly together.  In this case, the cohesive force of water molecules provides a very strong pull.

### Blood Pressure
Common arterial blood pressure measurements produce values of $120\ {\rm mm\ Hg}$ and $80\ {\rm mm\ Hg}$ for systolic and diastolic pressures, respectively.  Both pressures have health implications.  

When systolic pressure is chronically high, there is an increased risk of stroke and heart attack.  On the other hand, low systolic pressure can increase the risk of fainting.  Systolic pressure increases dramatically during exercise to increase blood flow and returns to normal afterward.  This change in pressure may be beneficial to the tone of the circulatory system rather than causing any ill effects.

Diastolic pressure can be an indicator of fluid balance.  When it is low, this may indicate that a person is hemorrhaging internally and needs a transfusion.  Conversely, high diastolic pressure indicates a ballooning of the blood vessels, which may be due to the transfusion of too much fluid into the circulatory system.  High diastolic pressure is also an indication that blood vessels are not dilating properly to pass blood through.  This can seriously strain the heart in its attempt to pump blood.

Blood leaves the heart at about $120\ {\rm mm\ Hg}$, but its pressure continues to decrease as it goes from the aorta to smaller arteries and veins.  The pressure differences in the circulation system are caused by blood flow through the system as well as position of the person.  For a person standing up, the pressure in the feet will be larger than at the heart due to the weight of the blood ($P = \rho g h$).  If we assume that the distance between the heart and the feet of a person is about $1.4\ {\rm m}$ when standing upright, then the increase in pressure in the feet is given by

\begin{align}
\Delta P &= \rho g \Delta h, \\
&= \left(1050\ {\rm kg/m^3}\right)(9.81\ {\rm m/s^2})(1.4\ {\rm m}) = 1.4\times 10^4\ {\rm Pa} \approx 1.1\times10^2\ {\rm mm\ Hg}.
\end{align}

Standing a long time can lead to an accumulation of blood in the legs and swelling.  This is the reason why soldiers who stand still for a long time have been known to faint.  Elastic bandages around the calf can help prevent this accumulation and can also help provide increased pressure to enable veins to send blood back up to the heart.

Blood pressure may also be measured in major veins, the heart chambers, arteries to the brain, and the lungs.  But these pressures are usually only monitored during surgery or for patients in intensive care since the measurements are invasive.  Health workers thread thin tubes called **catheters** into appropriate locations to transmit pressures to external monitoring devices.

The heart consists of two pumps: (1) the right side forcing blood through the lungs and (2) the left causing blood to flow through the rest of the body.  

- Right-heart failure results in a rise in the pressure in the vena cavae and a drop in pressure in the arteries to the lungs.  
- Left-heart failure results in a rise in the pressure entering the left side of the heart and a drop in aortic pressure.

### Pressure in the Eye
The shape of the eye is maintained by fluid pressure called **intraocular pressure**, which is normally in the range of $12-24\ {\rm mm\ Hg}$.  Impaired fluid drainage can increase this pressure. Elevated pressure is an important risk factor for **glaucoma**, which damages the optic nerve, but glaucoma can also occur at normal pressure. See the [National Eye Institute overview of glaucoma](https://www.nei.nih.gov/eye-health-information/eye-conditions-and-diseases/glaucoma).  The net pressure can become as great as $85\ {\rm mm\ Hg}$, which can permanently damage the optic nerve.

Suppose the back of the eye has an area of $6.0\ {\rm cm^2}$ and the net pressure is $85\ {\rm mm\ Hg}$, then we can calculate force ($F = PA$).  First, we must use the unit conversion where $1\ {\rm m^2} = 10^4\ {\rm cm^2}$.  Then we calculate as follows:

\begin{align}
F &= PA = (\rho g h)A, \\
&= (1.36 \times 10^4\ {\rm kg/m^3})(9.81\ {\rm m/s^2})(8.5\times 10^{-2}\ {\rm m})(6.0 \times 10^{-4}\ {\rm m^2}) = 6.8\ {\rm N}.
\end{align}

This force is the weight of about $0.69\ {\rm kg}$ ($1.5\ {\rm lb}$) resting on the eye.  This is sufficient to cause it damage, where a normal force would be less than $1/4$ of our initial value.

Glaucoma risk depends on age, family history, and other factors; a comprehensive eye examination evaluates more than pressure alone.  Most measurements involve exerting a force on the (anesthetized) eye over some area (i.e., a pressure) and observing the eye's response.  A noncontact approach uses a puff of air and a measurement is made of the force needed to indent the eye.  If the intraocular pressure is high, the eye will deform less and rebound more vigorously than normal.




### Pressure in the Lungs

The pressure inside the lungs increases and decreases with each breath.  The pressure drops to below atmospheric pressure (*negative gauge pressure*) when you inhale, causing air to flow into the lungs.  It increases above atmospheric pressure (*positive gauge pressure*) when you exhale, forcing air out.

Lung pressure is controlled by several mechanisms:

1. Muscle action in the diaphragm and rib cage is necessary for inhalation.  This muscle action increases the volume of the lungs thereby reducing the pressure within them.
2. Surface tension in the alveoli creates a positive pressure opposing inhalation.

You can exhale without muscle action by letting surface tension in the alveoli create its own positive pressure.  Muscle action can add to this positive pressure to produce a forced exhalation (e.g., blowing up a balloon, blowing out a candle, or coughing).

The lungs would collapse due to the surface tension in the alveoli, if they were not attached to the inside of the chest wall by liquid adhesion.  The gauge pressure in the liquid attaching the lungs to the inside of the chest wall is *negative*, ranging from $-4\ {\rm mm\ Hg}$ to $-8\ {\rm mm\ Hg}$, during exhalation and inhalation, respectively.  If air is allowed to enter the chest cavity, it breaks the attachment, and either lung (or both lungs) may collapse.  Suction is applied to the chest cavity of surgery patients and trauma victims to reestablish negative pressure and inflate the lungs.

### Spinal Column and Skull
Normally, there is a $5-12\ {\rm mm\ Hg}$ pressure in the fluid surrounding the brain and filling the spinal column.  This cerebrospinal fluid supplies flotation to the brain, among other purposes.  The buoyant force supplied by the fluid nearly equals the weight of the brain since their densities are nearly equal.  If there is a loss of fluid, the brain rests on the inside of the skull, which can cause severe headaches, constricted blood flow, and serious damage.  Spinal fluid pressure is measured by means of a needle inserted between vertebrae that transmits the pressure to a suitable measuring device.

### Bladder Pressure

This bodily pressure climbs steadily from zero to about $25\ {\rm mm\ Hg}$ as the bladder fills to its normal capacity of $0.5\ {\rm L}$.  This pressure triggers the *micturition reflex*, which stimulates the feeling of needing to urinate.  It also causes muscles around the bladder to contract, raising the pressure to over $100\ {\rm mm\ Hg}$.  Coughing, straining, tensing in cold weather, wearing tight clothes, and experiencing nervous tension all can increase bladder pressure and trigger this reflex.  Bladder pressure can be measured by a catheter or by inserting a needle through the bladder wall and transmitting the pressure to an appropriate measuring device.

### Pressures in the Skeletal System

Skeletal pressures are the largest in the body, due to the high values of internal force and the small areas to which this force is applied (e.g., in the joints).  For example, when a person lifts an object improperly, a force of $5000\ {\rm N}$ can be created between the vertebrae in the spine, which may be applied to an area as small as $10\ {\rm cm^2}$.  The pressure created is

$$ P = \frac{F}{A} = \frac{5000\ {\rm N}}{10^{-3}\ {\rm m^2}} = 5.0 \times 10^6\ {\rm N/m^2}, $$

or about $50\ {\rm atm}$.

This pressure can damage both the spinal discs (i.e., cartilage between the vertebrae) as well as the bony vertebrae themselves.  Even under normal circumstances, forces between vertebrae in the spine are large enough to create pressures of several atmospheres.  Most causes of excessive pressure in the skeletal system can be avoided by lifting properly and avoiding extreme physical activity.

There are many medically significant pressures in the body:

- Pressure caused by various muscle actions drives food and waste through the digestive system.
- Stomach pressure behaves much like bladder pressure and is tied to the sensation of hunger.  Positive pressure in the stomach may force acid into the esophagus, causing "heartburn."
- Pressure in the relaxed esophagus is normally negative because pressure in the chest cavity is normally negative.
- Pressure in the middle ear can result in a significant force on the eardrum if it differs greatly from atmospheric pressure (e.g., while scuba diving).  The decrease in external pressure is noticeable during plane flights.  The Eustachian tubes connect the middle ear to the throat and allow us to equalize pressure in the middle ear to avoid an imbalance of force on the eardrum.


### Example Problem: Pressure Difference Across an Eardrum

`````{admonition} Example Problem: Pressure Difference Across an Eardrum
:class: etamu-exercise

**The Problem**

> In a simplified model, a force of $3.0\ {\rm N}$ can rupture an eardrum of area $1.0\ {\rm cm^2}$.
>
> **(a)** What pressure difference across the eardrum corresponds to this force, in pascals and millimeters of mercury?
> **(b)** At what freshwater depth would that difference occur if the middle-ear pressure remained atmospheric? Use $\rho_{\rm w}=1.000\times10^3\ {\rm kg/m^3}$.

````{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

The net force depends on the pressure difference across the eardrum, rather than on the absolute pressure on either side. We assume the difference is uniform over the given area. For the underwater part, depth $h$ increases downward from the surface; the outside pressure rises hydrostatically while the middle-ear pressure is held at atmospheric pressure by the stated assumption.

---

**The Math**

**(a)** The net force on the eardrum equals the pressure difference across it multiplied by its area. The area is $1.0\ {\rm cm^2}=1.0\times10^{-4}\ {\rm m^2}$. Dividing the assumed rupture force by this area gives the corresponding pressure difference in pascals.

$$
\Delta P=\frac{F}{A}=\frac{3.0\ {\rm N}}{1.0\times10^{-4}\ {\rm m^2}}=3.0\times10^4\ {\rm Pa}.
$$

The pressure difference can also be expressed in millimeters of mercury. Dividing by the number of pascals per millimeter of mercury changes the unit without changing the physical pressure difference.

$$
\Delta P=(3.0\times10^4\ {\rm Pa})\left(\frac{1\ {\rm mm\ Hg}}{133.3\ {\rm Pa}}\right)\approx2.3\times10^2\ {\rm mm\ Hg}.
$$

**(b)** The water surface and the middle ear are both at atmospheric pressure under the stated assumptions. Atmospheric pressure therefore cancels when the inner and outer pressures are subtracted. The remaining difference is $\Delta P=\rho_{\rm w}gh$, and dividing it by $\rho_{\rm w}g$ determines the freshwater depth that produces the assumed force.

$$
h=\frac{\Delta P}{\rho_{\rm w}g}=\frac{3.0\times10^4\ {\rm Pa}}{(1.000\times10^3\ {\rm kg/m^3})(9.81\ {\rm m/s^2})}=3.1\ {\rm m}.
$$

---

**The Conclusion**

The assumed rupture force corresponds to a pressure difference of $3.0\times10^4\ {\rm Pa}$, or about $2.3\times10^2\ {\rm mm\ Hg}$. The model reaches that difference at a depth of $3.1\ {\rm m}$ if middle-ear pressure does not change. This is a consequence of the specified force threshold and pressure assumption, not a universal injury depth.
`````
