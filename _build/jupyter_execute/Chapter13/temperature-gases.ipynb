## Temperature

Human perception of temperature has evolved from the relative sensation of what feels hot or cold.  For example,

- place one hand in hot water and the other in cold water;
- then place both hands in tepid (warm) water.

The tepid water will feel cool to the hand that was in hot water, and warm to the one that was in cold water.  The scientific definition of temperature is less ambiguous than your senses of hot and cold.

**Temperature** is operationally defined as what we measure with a thermometer.  However, we will ultimately define the temperature in terms of the kinetic energies of atoms and molecules.

Many physical properties depend on temperature.  For example, volume increases with temperature for most substances.  This property is the basis for the common alcohol thermometer, the old mercury thermometer, and the bimetallic strip (used in modern thermostats).  Other properties used to measure temperature include electrical resistance and color, as well as the emission of infrared (IR) radiation.

### Temperature Scales

Thermometers are used to measure temperature according to well-defined scales that use pre-defined reference points to compare quantities.  The three most common temperature scales are the Fahrenheit, Celsius, and Kelvin scales.  A temperature scale can be created by identifying two easily reproducible temperatures.  The freezing and boiling temperatures of water at standard atmospheric pressure are commonly used.

- The **Celsius** scale has the freezing point of water at $0^\circ{\rm C}$ and the boiling point at $100^\circ{\rm C}$.  Its unit is the **degree Celsius** $^\circ{\rm C}$.  
- On the **Fahrenheit** scale, the freezing point of water is $32^\circ{\rm F}$ and the boiling point is at $212^\circ{\rm F}$, where its unit is the **degree Fahrenheit** $^\circ{\rm F}$.  
    - *Note that a temperature difference of 1 degree in Celsius is greater than 1 degree difference in Fahrenheit.*
    - One degree on the Celsius scale is $9/5 = 1.8$ times larger than one degree on the Fahrenheit scale.

The **Kelvin** scale is most commonly used in science.  It is an *absolute temperature scale* defined to have $0\ {\rm K}$ at the lowest possible temperature (i.e., **absolute zero**).  The *kelvin* ($\rm K$) is the unit on this scale and the adopted SI unit for temperature.  The freezing and boiling points on this scale are $273.15\ {\rm K}$ and $373.15\ {\rm K}$, respectively.  A temperature difference of one degree is the same on the Celsius and Kelvin scales.

### Thermal Equilibrium and the Zeroth Law of Thermodynamics

Thermometers actually take their own temperature, not the temperature they are measuring.  *Thermal contact* is the heat transfer that occurs between two objects.  

- After some time $t$, both objects will have exactly the same temperature.  Then the objects are in **thermal equilibrium** and no further heat flow (or changes in temperature) will occur.  
- Given enough time, the thermometer registers a temperature of both the thermometer and the system with which it is in thermal equilibrium.  
- A single object is in internal thermal equilibrium when no heat is flowing within the object and therefore its temperature is uniform.


````{admonition} Zeroth Law of Thermodynamics
:class: note

If two systems, $A$ and $B$, are in thermal equilibrium, they have the same temperature. If a third system, $C$, is also in thermal equilibrium with $B$, then $A$ and $C$ must be in thermal equilibrium with each other. All three systems have the same temperature.

```{figure-md} fig-zeroth-law-equilibrium-sequence
<img src="before_contact_and_equilibrium.png" alt="Three-panel illustration of the approach to thermal equilibrium. The left panel, labeled Before, shows three separate blocks A, B, and C at different temperatures: 320 K, 300 K, and 280 K. The middle panel, labeled In contact, shows the blocks touching, with heat flowing from the hotter block A toward the cooler blocks. The right panel, labeled Equilibrium, shows all three blocks at the same temperature, 300 K, with no net heat transfer." style="width: 100%; max-width: 100%; height: auto;">

When systems at different temperatures are brought into thermal contact, heat flows from higher temperature to lower temperature until thermal equilibrium is reached. Once equilibrium is established, all three systems have the same temperature, so there is no net heat transfer between them.
```
````

The zeroth law of thermodynamics was postulated in the 1930s, after the first and second laws of thermodynamics were already developed.  Suppose a cold metal block and a hot metal block are both placed on a metal plate at room temperature.  Eventually the cold block and plate will be in thermal equilibrium.

By the zeroth law, we can conclude that the cold block and hot block are also in thermal equilibrium.  Two thermometers placed in thermal contact (one with each block) will also be in thermal equilibrium and therefore read the same temperature.

### Example Problem: Room Temperature on Different Scales

````{admonition} Example Problem: Room Temperature on Different Scales
:class: etamu-exercise

**The Problem**

> A room is at $25^\circ{\rm C}$. Express this temperature on (a) the Fahrenheit scale and (b) the kelvin scale.

*Adapted from [OpenStax, College Physics 2e, Example 13.1](https://openstax.org/books/college-physics-2e/pages/13-1-temperature).*

```{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

The room has one physical temperature, but each scale assigns it a different numerical value. We use the freezing and boiling reference points to relate Celsius to Fahrenheit, then shift the zero of the Celsius scale to obtain the absolute temperature. These are changes of scale rather than temperature changes, so offsets must be included.

---

**The Math**

**(a)** The Fahrenheit scale has a smaller degree interval and a different zero than the Celsius scale. We define $t_{\rm C}$ and $t_{\rm F}$ as the numerical readings in degrees Celsius and degrees Fahrenheit; their calibration gives the linear conversion

$$
t_{\rm F}=\frac{9}{5}t_{\rm C}+32.
$$

**(b)** A kelvin has the same size as a Celsius degree, so only the origin changes. The absolute temperature is therefore

$$
T=(t_{\rm C}+273.15)\ {\rm K}.
$$

The conversion factors and offsets are exact, so the precision of the given reading sets the precision of the converted temperatures. Substitution into the two established relations gives

\begin{align*}
t_{\rm F}&=\frac{9}{5}(25)+32=77,\\
T&=(25+273.15)\ {\rm K}\approx298\ {\rm K}.
\end{align*}

---

**The Conclusion**

Room temperature is **(a)** $77^\circ{\rm F}$ and **(b)** $298\ {\rm K}$. Both readings lie between the freezing and boiling points of water on their respective scales, as a comfortable room temperature should. The whole-degree reporting follows the precision of the Celsius reading; an offset conversion does not preserve a fixed number of significant figures.

```
````

### Example Problem: Room Temperature on the Réaumur Scale

````{admonition} Example Problem: Room Temperature on the Réaumur Scale
:class: etamu-exercise

**The Problem**

> The Réaumur scale assigns $0^\circ{\rm R}$ to the freezing point of water and $80^\circ{\rm R}$ to its boiling point at standard atmospheric pressure. What reading corresponds to $25^\circ{\rm C}$? Here $^\circ{\rm R}$ denotes Réaumur, not Rankine.

*Adapted from [OpenStax, College Physics 2e, Example 13.2](https://openstax.org/books/college-physics-2e/pages/13-1-temperature).*

```{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

Both scales place zero at the freezing point of water, so no offset is needed. The boiling-to-freezing interval fixes the ratio of the degree sizes. We apply that ratio to the Celsius reading to obtain the Réaumur reading.

---

**The Math**

The fraction of the freezing-to-boiling interval must be the same on either linear scale. We denote the numerical Celsius and Réaumur readings by $t_{\rm C}$ and $t_{\rm R}$ and use their calibration points to write

$$
\frac{t_{\rm R}}{80}=\frac{t_{\rm C}}{100},\qquad t_{\rm R}=\frac{4}{5}t_{\rm C}.
$$

The scale ratio is exact, so the given Celsius reading determines the Réaumur reading through

$$
t_{\rm R}=\frac{4}{5}(25)=20.
$$

---

**The Conclusion**

The room temperature is $20^\circ{\rm R}$. It is one-quarter of the way from freezing to boiling on either scale, which checks the conversion. The smaller numerical reading reflects the larger Réaumur degree, not a colder room.

```
````

## Thermal Expansion of Solids and Liquids

The expansion of alcohol in thermometers is an example of **thermal expansion**, which is the change in size or volume of a given mass with temperature.  Hot air rises because air expands when its temperature increases, which causes the hot air's density to be smaller than the density of the surrounding air.  

- The buoyant (upward) force on a given mass of air increases when the air is heated.
- This results in a net upward force on that air mass.

The same happens in all liquids and gases, driving natural heat transfer upwards in homes, oceans, and weather systems.  Solids also undergo thermal expansion.  Railroad tracks and bridges have expansion joints to allow them to freely expand and contract with temperature changes.

*What are the basic properties of thermal expansion?*

1. Thermal expansion is clearly related to temperature change.  The greater the temperature change, the more a bimetallic strip will bend.
2. It depends on the material.  In a thermometer, the expansion of alcohol is much greater than the expansion of the glass containing it.

*What is the underlying cause of thermal expansion?*

An increase in temperature implies an increase in *kinetic energy* of the individual atoms.  In a solid, the atoms or molecules are closely packed together, but their kinetic energy (through rapid vibrations) pushes neighboring atoms (or molecules) apart from each other.  This pushing of neighboring atoms results in a slightly greater distance (on average) between neighbors, and adds up to a larger size for the whole object.  Under ordinary conditions for most substances, there is no preferred direction and an increase in temperature will increase the solid's size by a certain fraction in each dimension. 

```{admonition} Linear Thermal Expansion in 1D
:class: note

The change in length $\Delta L$ is proportional to length $L$.  The dependence of thermal expansion $\Delta L$ on temperature change $\Delta T$ (in kelvins), substance, and length $L$ is summarized by

\begin{align}
\Delta L = \alpha L \Delta T,
\end{align}

where $\alpha$ (in $1/{\rm K}$) is the **coefficient of linear expansion**.  The coefficient $\alpha$ varies slightly with temperature.
```

### Thermal Expansion in 2D and 3D
Objects expand in all dimensions, where their lengths, areas, and volumes all increase with temperature.  Holes get larger with a larger temperature.  If you cut a hole in a metal plate, the remaining material will expand exactly as it would if the plug (i.e., piece of metal that is cut out) was still in place. The plug would get bigger and so the hole must get bigger too.

```{admonition} Thermal Expansion in 2D and 3D
:class: note

For small temperature changes, the change in area $\Delta A$ is given by

\begin{align}
\Delta A = 2\alpha A\Delta T,
\end{align}

which depends on the change in temperature $\Delta T$ (in kelvins), and $\alpha$ (in $1/{\rm K}$) is the coefficient of linear expansion. 

---

The change in volume $\Delta V$ is very nearly $\Delta V = 3\alpha V\Delta T$, but the formal equation is usually written as

\begin{align}
\Delta V = \beta V\Delta T,
\end{align}

where $\beta$ is the **coefficient of volume expansion** (in $1/{\rm K}$) and $\beta \approx 3\alpha$.
```
In general, objects will expand with increasing temperature.  *Water is the most important exception to this rule.*  Water expands with increasing temperature (its density *decreases*) when it is between $+4^\circ{\rm C}$ and $0^\circ{\rm C}$.  Water is densest at $+4^\circ{\rm C}$.

The most striking effect is the freezing of water in a pond.  When water near the surface cools to $4^\circ{\rm C}$, it is denser than the remaining water and thus will sink to the bottom.  This "turnover" results in a layer of warmer water near the surface, which is then cooled.  Eventually, the pond has a uniform temperature of $4^\circ{\rm C}$.

If the temperature of the surface layer drops below $4^\circ{\rm C}$, the water is less dense than the water below, and thus stays near the top. As a result, the pond surface can completely freeze over.  The ice on top of liquid water provides an insulating layer from winter's harsh exterior air temperatures.  Fish and other aquatic life can survive in $4^\circ{\rm C}$ beneath ice, due to this unusual characteristic of water.  It also produces circulation of water in the pond that is necessary for a healthy ecosystem for the body of water.

### Thermal Stress

**Thermal stress** is created by thermal expansion or contraction.  This can be destructive, such as when expanding gasoline ruptures a tank.  It can also be useful: for example, when two parts are joined together by heating one in manufacturing, then slipping it over the other and allowing the combination to cool.  Thermal stress can explain the weathering of rocks and pavement by expansion of ice when it freezes, as well as other phenomena.

- Railroad tracks and roadways can buckle on hot days if they lack sufficient expansion joints.  
- Power lines sag more in the summer than in the winter, and will snap in cold weather if there is insufficient slack.
- Cracks open and close in plaster walls as a house warms and cools.
- Glass cooking pans will crack if cooled rapidly and unevenly, because of differential contraction and the stresses it creates.
- Nuclear reactor pressure vessels are threatened by overly rapid cooling.
- Biological cells are ruptured when foods are frozen, detracting from their taste.

Metal is regularly used in the human body for hip and knee implants.  Most implants need to be replaced over time because metal does not bond with bone. Researchers are looking for better metal coatings that would allow metal-to-bone bonding.  One challenge is to find a coating that has an expansion coefficient similar to that of metal.  If the expansion coefficients are too different, then the thermal stresses during the manufacturing process lead to cracks at the coating-metal interface.

Dental fillings can expand differently from tooth enamel, causing pain when eating ice cream or having a hot drink.  Cracks might occur in the filling.  Metal fillings (e.g., gold or silver) are generally replaced with composite fillings (i.e., porcelain), which have smaller coefficients of expansion closer to those of teeth.

### Example Problem: Expansion of the Golden Gate Bridge

````{admonition} Example Problem: Expansion of the Golden Gate Bridge
:class: etamu-exercise

**The Problem**

> The main span of the Golden Gate Bridge is $1275\ {\rm m}$ long at $-15^\circ{\rm C}$. What is its increase in length when it reaches $40^\circ{\rm C}$? Model the span as steel with a constant linear expansion coefficient of $12\times10^{-6}\ {\rm K^{-1}}$.

```{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

We treat the span as a uniform steel length that can expand freely as it warms. Linear thermal expansion relates the fractional length change to the temperature change. The given cold length is the initial length, and a positive result represents an increase in the span's length.

---

**The Math**

The temperature change is the final temperature minus the initial temperature. Inserting that symbolic difference into the linear-expansion relation expresses the desired change directly in terms of the two endpoint temperatures:

$$
\Delta L=\alpha L_0\Delta T=\alpha L_0(T_f-T_0).
$$

A temperature interval has the same numerical value in kelvins and Celsius degrees. The specified temperatures differ by $55\ {\rm K}$, so the length change is

$$
\Delta L=(12\times10^{-6}\ {\rm K^{-1}})(1275\ {\rm m})(55\ {\rm K})=0.84\ {\rm m}.
$$

---

**The Conclusion**

The span length increases by $0.84\ {\rm m}$. This is small compared with a span more than a kilometer long, consistent with the small fractional change assumed by the linear model. It is nevertheless large enough that expansion joints must accommodate the movement.
````

````{admonition} Example Problem: Gasoline Overflow from a Warming Tank
:class: etamu-exercise

**The Problem**

> A steel tank holds $60.0\ {\rm L}$ of gasoline and is completely full at $15.0^\circ{\rm C}$. How much gasoline spills when the tank and fuel warm to $35.0^\circ{\rm C}$? Use volume expansion coefficients of $9.5\times10^{-4}\ {\rm K^{-1}}$ for gasoline and $3.5\times10^{-5}\ {\rm K^{-1}}$ for steel.

*Adapted from [OpenStax, College Physics 2e, Example 13.4](https://openstax.org/books/college-physics-2e/pages/13-2-thermal-expansion-of-solids-and-liquids).*

```{dropdown} Show worked solution
:color: secondary
:icon: pencil
:class-container: etamu-dropdown
:class-title: etamu-dropdown-title
:class-body: etamu-dropdown-body

**The Model**

Both the gasoline and the tank's internal capacity increase as they warm. We treat their expansion coefficients as constant and neglect evaporation. The spilled volume is the fuel's free expansion minus the increase in available tank capacity, rather than the fuel's expansion alone.

---

**The Math**

The initially full tank gives the fuel and tank the same initial volume $V_0$. Their separate volume increases follow the same expansion law but use different material coefficients:

\begin{align*}
\Delta V_{\rm gas}&=\beta_{\rm gas}V_0\Delta T,\\
\Delta V_{\rm tank}&=\beta_{\rm steel}V_0\Delta T.
\end{align*}

The excess fuel cannot fit in the enlarged tank. Subtracting the available capacity increase and factoring out the common initial volume gives

$$
V_{\rm spill}=\Delta V_{\rm gas}-\Delta V_{\rm tank}=(\beta_{\rm gas}-\beta_{\rm steel})V_0(T_f-T_0).
$$

The temperature rise is $20.0\ {\rm K}$. Substitution at the end of the symbolic development gives

$$
V_{\rm spill}=[(9.5\times10^{-4}-3.5\times10^{-5})\ {\rm K^{-1}}](60.0\ {\rm L})(20.0\ {\rm K})\approx1.1\ {\rm L}.
$$

---

**The Conclusion**

About $1.1\ {\rm L}$ of gasoline spills. The result is positive because gasoline expands more than steel, and it is only a small fraction of the original tank volume. The two-significant-figure result reflects the approximate expansion coefficients; treating the tank as nonexpanding would slightly overestimate the spill.

```
````

````{admonition} Example Problem: Pressure Rise in a Sealed Gasoline Tank
:class: etamu-exercise

**The Problem**

> The $60.0\ {\rm L}$ steel tank in the preceding example is sealed with no air space, and the gasoline warms from $15.0^\circ{\rm C}$ to $35.0^\circ{\rm C}$. Estimate the pressure increase if the steel tank expands thermally but has negligible additional expansion under pressure. Use $B=1.00\times10^9\ {\rm Pa}$ for gasoline and the same expansion coefficients as before. Assume the tank remains intact.

```{dropdown} Show worked solution

**The Model**

We compare the volume the warm gasoline would occupy freely with the smaller volume available inside the warm tank. Preventing this excess expansion compresses the gasoline relative to its freely expanded state. The bulk modulus connects the fractional volume reduction to the pressure increase, using a small-strain approximation and neglecting pressure-driven tank deformation.

---

**The Math**

The excess free expansion is the same volume that would have spilled from the open tank. Denoting this positive excess by $\Delta V_{\rm excess}$ gives

$$
\Delta V_{\rm excess}=(\beta_{\rm gas}-\beta_{\rm steel})V_0\Delta T.
$$

The bulk-modulus relation uses a negative volume change for compression. The required compression is $\Delta V_{\rm comp}=-\Delta V_{\rm excess}$, so the resulting pressure increase is positive:

$$
\Delta P=-B\frac{\Delta V_{\rm comp}}{V_0}=B\frac{\Delta V_{\rm excess}}{V_0}=B(\beta_{\rm gas}-\beta_{\rm steel})\Delta T.
$$

The initial volume cancels, and evaluating the combined expression avoids reusing the rounded spill volume. The pressure increase is

$$
\Delta P=(1.00\times10^9\ {\rm Pa})[(9.5\times10^{-4}-3.5\times10^{-5})\ {\rm K^{-1}}](20.0\ {\rm K})\approx1.8\times10^7\ {\rm Pa}.
$$

---

**The Conclusion**

The estimated pressure increase is $1.8\times10^7\ {\rm Pa}$. This is much larger than atmospheric pressure, as expected when a nearly incompressible liquid is prevented from expanding. A real fuel tank would deform, leak, or fail before this idealized intact-tank estimate could reliably describe it. The result is a pressure increase, not an absolute pressure.
````

## The Ideal Gas Law

Gases are easily compressed because their molecules are relatively far apart. They also have the largest coefficients of volume expansion (see Table {numref}`{number}<thermal-expansion-table>`). These large coefficients mean that gases undergo much greater changes in volume for a given temperature change than most solids and liquids, assuming constant pressure.

Because atoms and molecules have large separations, forces between them can be ignored, except when they collide with each other.  The motion of atoms and molecules is fast, such that the gas occupies all of the accessible volume and the expansion of gases is rapid.  In liquids and solids, atoms and molecules are closer together, which makes them quite sensitive to the forces between them.

```{figure-md} fig-tire-pressure-temperature
<img src="tire_pressure_and_temperature_triptych.png" alt="Three-panel realistic comparison of tire pressure changes. In panel (c), after the tire is inflated, increasing the temperature raises the pressure. Pressure gauges are connected to the tire valve stems by hoses, and cutaway views in the right-hand tire of each panel show the changing behavior of the gas inside the tire." style="width: 100%; max-width: 100%; height: auto;">

Pressure in a tire depends on both the amount of air in the tire and its temperature. (a) When air is first added to a deflated tire, the tire expands noticeably, so its volume increases much more than its pressure. (b) Once the tire is nearly full and the tire walls resist further expansion, adding more air produces a larger increase in pressure. (c) After the tire is inflated, increasing the temperature raises the pressure of the gas inside the tire. The cutaway views illustrate the gas particles inside the tire, with faster particle motion in the warmer tire. *Adapted from: [OpenStax, College Physics 2e, Figure 13.19](https://openstax.org/books/college-physics-2e/pages/13-3-the-ideal-gas-law).*
```

Consider what happens when you pump air into an initially deflated tire.  The tire's volume first increases in direct proportion to the amount of air injected, without much increase in the tire pressure.  Once the tire has expanded to nearly its full size, the walls limit volume expansion.  If we continue to pump air into it, the pressure increases.  The pressure will further increase when the car is driven and the tires move.

For room temperature air, the gas particles have negligible volume and do not interact with each other where they experience perfectly elastic collisions.  In such cases, the gas is called an **ideal gas**.  There is a mathematical relationship between the pressure, volume, and temperature which is called the **ideal gas law**.  An equation (e.g., the ideal gas law) which relates the behavior of a physical system in terms of its thermodynamic properties is called an *equation of state*.

```{admonition} Ideal Gas Law
:class: note

The **ideal gas law** states that 

\begin{align}
PV = NkT,
\end{align}

where $P$ is the absolute pressure (in $\rm Pa$) of a gas, $V$ (in $\rm m^3$) is the volume it occupies, $N$ is the number of atoms and molecules in the gas, and $T$ (in ${\rm K}$) is its absolute temperature.  The constant $k$ is called the Boltzmann constant, which has the value

$$ k = 1.38\times 10^{-23}\ {\rm J/K}.$$
```

The ideal gas law can be derived from basic principles, but it was originally deduced from experimental measurements of 

- Charles' law (that volume is proportional to temperature at a fixed pressure),
- Boyle's law (that the product $PV$ is constant for a fixed temperature), and
- Gay-Lussac's law (that the ratio $P/T$ is constant for a fixed volume).

The ideal gas law describes the behavior of real gases under most conditions.

The ideal gas law describes a tire when it is pumped slowly and the temperature is constant.  At first, the pressure $P$ is essentially equal to the atmospheric pressure, and the volume $V$ increases in direct proportion to the number of atoms and molecules $N$ put into the tire.  Once the volume of the tire is constant, the equation $PV = NkT$ predicts that the pressure should increase in proportion to the number $N$ of atoms and molecules.

### Moles and Avogadro's Number 
Chemists use another unit instead of number of molecules when measuring the amount of a substance.  A **mole** (abbreviated $\rm mol$) is defined as the amount of substance that contains the same number of atoms as $12\ {\rm g}$ of carbon-$12$.  The actual number of molecules in one mole is **Avogadro's number** ($N_{\rm A}$) in recognition of [Amedeo Avogadro](https://en.wikipedia.org/wiki/Amedeo_Avogadro).  He developed the concept of the mole based on the hypothesis that equal volumes of gas (at the same pressure and temperature) contain equal numbers of molecules.

```{admonition} Avogadro's Number
:class: note

One mole contains $6.02 \times 10^{23}$ particles (atoms or molecules), independent of the element or substance.  A mole of any substance has a mass in *grams* equal to its molecular (*molar*) mass, which can be calculated by multiplying the number of moles of the substance by its atomic mass.  The atomic masses of elements are given in the [periodic table](https://en.wikipedia.org/wiki/Periodic_table).  Mathematically, Avogadro's number is given by

\begin{align}
N_{\rm A} = 6.02 \times 10^{23}\ {\rm mol^{-1}}.
\end{align}
```

### The Ideal Gas Law Restated Using Moles
Another version of the ideal gas law uses the number of moles $n$, rather than the number of atoms and molecules $N$.  To find this alternative relation, we begin with the ideal gas law and multiply the right-side by Avogadro's number on top and bottom ($N_{\rm A}/N_{\rm A} = 1$) to get

\begin{align}
PV &= NkT,\\
PV &= \left(\frac{N}{N_{\rm A}}\right) \left(N_{\rm A}k\right)T.
\end{align}

The number of moles is just the ratio of the number of molecules with Avogadro's number, or $n= N/N_{\rm A}$.  We define the universal gas constant $R = N_{\rm A}k$, and then we obtain the ideal gas law in terms of moles.

```{admonition} Ideal Gas Law (in Terms of Moles)
:class: note

The ideal gas law (in terms of moles) is given as

\begin{align}
PV = nRT.
\end{align}


The **universal gas constant**, $R$, is related to Avogadro's constant $N_{\rm A}$ and Boltzmann's constant $k$ through $R=N_{\rm A}k$. Since the 2019 revision of the International System of Units (SI), both $N_{\rm A}$ and $k$ have exact defined values:

\begin{align}
N_{\rm A} &= 6.02214076 \times10^{23}\ {\rm mol}^{-1}, \\
k &= 1.380649 \times10^{-23}\ {\rm J/K}.
\end{align}

As a result, the universal gas constant has the exact value

\begin{align}
R = N_{\rm A}k = 8.31446261815324\ {\rm J/(mol\cdot K)}.
\end{align}

For most calculations, we use the rounded value $R=8.31\ {\rm J/(mol\cdot K)}$. The gas constant can also be expressed in other commonly used units:

\begin{align*}
R &\approx 1.99\ {\rm cal/(mol\cdot K)},\\
R &\approx 0.0821\ {\rm L\cdot atm/(mol\cdot K)}.
\end{align*}

You can use whichever form of $R$ is most convenient for a particular problem, provided the units are consistent.
```

The ideal gas law can be considered as another manifestation of the law of conservation of energy.  Work done on a gas results in an increase in its energy, by increasing pressure and/or temperature.  This increased energy can also be viewed as increasing the kinetic energy of the atoms and molecules within the gas.


### The Ideal Gas Law and Energy
When you inflate a bike tire by hand, you do work by repeatedly exerting a force through a distance.  This energy goes into increasing the pressure of air inside the tire and increasing the temperature of the pump and the air.

The ideal gas law is closely related to energy: the dimensions on both sides are those of energy (with units of joules).  

- The right-hand side of the ideal gas law is $NkT$.  This term is proportional to the amount of translational energy for $N$ atoms or molecules at an absolute temperature $T$.
- The left-hand side of the ideal gas law is $PV$, which also has units of joules.  Pressure is force per unit area, so we have

$$ PV = \left(\frac{F}{A}\right)V = F\left(\frac{V}{A}\right) = Fd. $$

There is energy in a gas related to both its pressure and its volume.  The energy can be changed when the gas is doing work as it expands.

## Kinetic Theory

Pressure is the force divided by the area on which the force is exerted and temperature is measured with a thermometer.  The kinetic theory of gases provides further insight by assuming that gas particles are in continuous random motion.

Consider an elastic collision of a gas molecule with the wall of a container, where it exerts a force on the wall by Newton's third law.  We observe an average force per unit area through the huge number of molecules that collide with the wall in a short time.  These collisions are the source of pressure in a gas.


As the number of molecules increases, the pressure increases due to the higher number of collisions.  The gas pressure is higher if the average velocity of the molecules is higher.  

Consider the elastic collision of a single molecule with a vertical wall within a box of width $\ell$.  Using conservation of linear momentum and energy, we can show that there is a change in momentum along the $x$-direction such that $-mv_x$ changes to $+mv_x$.  The total change in momentum  is then $\Delta mv = +mv_x - (-mv_x) = 2mv_x$.  Then the force on the molecule is given by

$$ F = \frac{\Delta p}{\Delta t} = \frac{2mv_x}{\Delta t}. $$

There is no force between the wall and the molecule until the molecule is in contact with (i.e., hits) the wall.  During the short time of the collision, the force between the molecule and the wall is relatively large.  Since we are looking for the average force, we take $\Delta t$ to be the average time between collisions of the molecule and this wall.  

It is the time for the molecule to fully go across to the other side of the box (a distance of $2\ell$) at a speed $v_x$.  We can determine that $\Delta t = 2\ell/v_x$ using the definition of velocity.  The expression for the force becomes

$$ F = \left(\frac{ 2mv_x}{1}\right) \left(\frac{v_x}{2\ell}\right) = \frac{mv_x^2}{\ell}. $$

This is the force per molecule, so we must multiply by the number of molecules $N$ and use the average squared velocity $\overline{v_x^2}$.  Thus, we now have the force as

$$ F = \frac{N}{\ell}m\overline{v_x^2}, $$

where the bar over a quantity denotes the average value.  To get the force in terms of just the average velocity squared, we consider that collisions must be occurring with the other sides of the boxes (i.e., in the $y$- and $z$-directions).  We expect the average velocity squared is represented by

$$ \overline{v^2} = \overline{v_x^2} +\overline{v_y^2} + \overline{v_z^2}. $$

Because the velocities are random, their average components in all directions are the same ($\overline{v_x^2}=\overline{v_y^2}=\overline{v_z^2}$).  So we can simplify so that $\overline{v^2} = 3\overline{v_x^2}$, or $\overline{v_x^2} = \overline{v^2}/3$.  Substituting into our expression for $F$, we find that

$$ F = \frac{N}{3\ell}m\overline{v^2}. $$

The pressure is $F/A$, so we can divide both sides by $A$ to get

$$ P = \frac{F}{A} = \frac{N}{3A\ell}m\overline{v^2} = \frac{N}{3V}m\overline{v^2}, $$

where we used the volume $V = A\ell$.  If we multiply the volume $V$ to both sides, we have

\begin{align}
PV = \frac{N}{3}m\overline{v^2},
\end{align}

which is another expression of the ideal gas law.  We can set this equal to our expected result, or

\begin{align}
\frac{N}{3}m\overline{v^2} &= NkT, \\
kT &= \frac{1}{3}m\overline{v^2}.
\end{align}

If we multiply both sides by $3/2$, then we find the average translational kinetic energy, or

\begin{align}
\overline{\rm KE} = \frac{3}{2}kT = \frac{1}{2}m\overline{v^2}.
\end{align}

The average translational kinetic energy of a molecule $\overline{KE}$ is called **thermal energy**.  This is another definition of the expression of temperature based on an expression of the molecular energy.

We can also solve for the *root-mean-square* velocity $v_{\rm rms}$, which describes the average velocity weighted by the velocity squared.  To solve for it, we use 

\begin{align}
v_{\rm rms} = \sqrt{\overline{v^2}} = \sqrt{\frac{3kT}{m}}.
\end{align}

The distribution has a long tail, because few molecules may go several times the $\rm rms$ speed.  The most probable speed $v_{\rm p}$ is less than the $\rm rms$ speed $v_{\rm rms}$.

## Phase Changes

Real gases are like ideal gases at high temperatures.  At lower temperatures, the interactions between the molecules and their volumes *cannot* be ignored.  When the molecules are very close condensation can occur and there is a dramatic decrease in volume.  The substance changes from a gas to a liquid.  When liquid is cooled to even lower temperatures, it becomes a solid.  The volume never reaches zero because of the finite volume of the molecules.

High pressure may also force a phase change for a gas to become a liquid. For example, carbon dioxide $\rm CO_2$ is a gas at room temperature and atmospheric pressure, but becomes a liquid under sufficiently high pressure.  If the pressure is reduced, the temperature drops to $-78^\circ{\rm C}$ and the liquid carbon dioxide solidifies into a snow-like substance.  Solid $\rm CO_2$ is called "dry ice."

Nitrogen can be cooled and compressed to make liquid nitrogen $\rm LN_2$.  It boils at $77\ {\rm K}$ (or $-196^\circ{\rm C}$) at atmospheric pressure.  Liquid nitrogen is useful as a refrigerant and allows for the preservation of blood, sperm, and other biological materials.  It is also used to reduce thermal noise in electronic sensors and equipment.  In dermatology, $\rm LN_2$ is used to freeze and painlessly remove warts and other growths from the skin.

### PV Diagrams

We can plot a graph comparing the pressure versus the volume, called a **PV diagram**.  When the substance behaves like an ideal gas, the ideal gas law describes the relationship between its pressure and volume (i.e., $PV = NkT$).  If we assume that the number of molecules and temperature are fixed,

\begin{align}
PV = \text{constant} \qquad (\text{ideal gas, constant temperature}).
\end{align}

For example, the volume of the gas will decrease as the pressure increases.  If you plot the relationship $PV = \text{constant}$ on a $PV$ diagram, you find a hyperbola.  The hyperbolas (in Figure {numref}`{number}<fig-real-gas-pv-isotherms>`) are called *isotherms* and represent ideal-gas behavior at various fixed temperatures.

### Equilibrium
Liquid and gas phases are in equilibrium at the boiling temperature.  If a substance is in a closed container at the boiling point, then the liquid is boiling and the gas is condensing at the same rate without a net change in their relative amounts.  Molecules in the liquid escape as a gas at the same rate at which gas molecules: (1) stick to the liquid, or (2) form droplets and become part of the liquid phase.


```{figure-md} fig-liquid-vapor-equilibrium
<img src="vaporization_and_condensation_equilibrium.png" alt="Two sealed glass containers partially filled with water, each equipped with a pressure gauge and thermometer. The right container has a higher temperature and vapor pressure. Green upward arrows represent vaporization, while purple downward arrows represent condensation. The arrows are larger in the higher-temperature container." style="width: 95%; max-width: 100%; height: auto;">

**Dynamic equilibrium between liquid and vapor.** (a) At a lower temperature, molecules leave the liquid through vaporization and return through condensation at equal average rates, maintaining equilibrium between the two phases. (b) At a higher temperature, more molecules enter the vapor phase, increasing the equilibrium vapor pressure. Water boils at $100^\circ{\rm C}$, when the pressure is $1\ {\rm atm}$ and means that the water/steam are in equilibrium.  *Why does an open pot of water boil completely away?*
```

The gas surrounding an open pot is not pure water: it is mixed with air.  If pure water and steam are in a closed container at $100^\circ{\rm C}$ and $1\ {\rm atm}$, they would coexist.  However, the air over the pot means that there are fewer water molecules to condense and water boils.  *What about room temperature ($20^\circ{\rm C}$) water at standard pressure?*

This temperature and pressure correspond to the liquid region, yet an open glass at this temperature will completely evaporate.  The gas around it is air and not pure water vapor, which means that the reduced evaporation rate is greater than the condensation rate of water from dry air.  If the glass is sealed, then the liquid phase remains.  We call the gas a vapor when it exists at a temperature below the boiling temperature.

### Vapor Pressure, Partial Pressure, and Dalton's Law

**Vapor pressure** is the pressure at which a gas coexists with its solid or liquid phase.  Vapor pressure is created by faster molecules that break away from the liquid (or solid)  and enter the gas phase (see Figure {numref}`{number}<fig-mb-oxygen-300K>`).  The vapor pressure of a substance depends on both the substance and its temperature.

Partial pressure is the pressure a gas would create if it occupied the total volume available.  In a mixture of gases, *the total pressure is the sum of the partial pressures of the component gases*, assuming ideal gas behavior and no chemical reactions.  This is known as **Dalton's law of partial pressures** after [John Dalton](https://en.wikipedia.org/wiki/John_Dalton).

Dalton's law is based on kinetic theory, where each gas exerts its pressure through molecular collisions, independent of the gases present.  It is consistent with how pressures add according to Pascal's principle.  Thus water evaporates and ice sublimates when their vapor pressures exceed the partial pressure of water vapor in the surrounding mixture of gases.  If their vapor pressures are less than the partial pressures in the surrounding gas, liquid droplets or ice crystals (frost) form.

## Humidity, Evaporation, and Boiling

We keep cool in hot weather by evaporating sweat from our skin and water from our breathing passages (i.e., nose and mouth).  Evaporation is inhibited by high humidity, so we feel hotter at a given temperature when the humidity is high.  Low humidity can cause discomfort from excessive drying of mucous membranes leading to an increased risk of respiratory infections and skin irritation.

When we say humidity, we really mean **relative humidity**, where this tells us how much water vapor is in the air compared with the maximum possible amount.

**Saturation** occurs at maximum humidity and the relative humidity is 100\%.  Evaporation is inhibited because adding more energy to water does not allow it to increase its motion for long before colliding with another water molecule and slowing back down.  The amount of water vapor depends on temperature.

If the air temperature decreases, the humidity can rise (e.g., in the evening) and sometimes the temperature can fall and reach the **dew point** temperature.  At the dew point temperature, the relative humidity is still $100\%$, where water vapor can condense into water droplets if they are small enough to remain in suspension and fog begins to form.  

If you wish to dry something, it is more effective to blow hot air over it rather than cold air because the hot air delivers more energy to the medium.  This means that the molecules can gain more energy and the rate of evaporation increases.

The amount of water vapor in the air depends on the vapor pressure of water.  The liquid and solid phases are continuously giving off vapor because some of the molecules have high enough speeds to enter the gas phase.  If a lid is placed over a container, evaporation increases the pressure between the liquid and the lid until sufficient vapor has built up for condensation to balance evaporation.

Then equilibrium has been achieved, and the vapor pressure is equal to the partial pressure of water in the container.  Vapor pressure increases with temperature because the molecular speeds are higher as the temperature increases.

*Why does water boil at $100^\circ{\rm C}$?*

*Why does it form bubbles when it boils?*

Water ordinarily contains significant amounts of dissolved air and other impurities, which are observed as small bubbles of air in a glass of water.  If a bubble starts out at the bottom of the container at $20^\circ{\rm C}$, it contains water vapor (about $2.3\%$).   The pressure inside the bubble is fixed at $1\ {\rm atm}$.

As the temperature rises, the amount of air in the bubble stays the same, but the water vapor increases.  The bubble expands to keep the pressure at $1\ {\rm atm}$.  This expansion displaces more water around the bubble, thereby increasing the buoyant force.  At $100^\circ{\rm C}$, water vapor enters the bubble continuously since the partial pressure of water is equal to $1\ {\rm atm}$ in equilibrium.  The bubble breaks away and rises rapidly to the surface, which we call boiling.

````{admonition} Example Problem: Saturated Water-Vapor Density
**The Model**

We model the water vapor as an ideal gas and apply the gas law to its partial pressure, not to the total pressure of the air. The gas law first connects pressure to the amount of vapor per volume. Multiplying that amount by the molar mass converts it into mass density.

---

**The Math**

The water-vapor partial pressure $P_{\rm v}$ obeys the ideal gas law. Dividing by volume and the thermal factor gives the number of moles per volume:

$$
P_{\rm v}V=nRT,\qquad \frac{n}{V}=\frac{P_{\rm v}}{RT}.
$$

The mass is the amount in moles times the molar mass $M$. The vapor density $\rho_{\rm v}$ is therefore

$$
\rho_{\rm v}=\frac{m}{V}=M\frac{n}{V}=\frac{MP_{\rm v}}{RT}.
$$

At saturation, the vapor partial pressure equals the tabulated saturation pressure. The temperature is $293.15\ {\rm K}$, so the density evaluates to

$$
\rho_{\rm v}=\frac{(18.0\ {\rm g/mol})(2.34\times10^3\ {\rm Pa})}{(8.31\ {\rm J/(mol\,K)})(293.15\ {\rm K})}=17.3\ {\rm g/m^3}.
$$

---

**The Conclusion**

The estimated saturated vapor density is $17.3\ {\rm g/m^3}$, agreeing with the updated table at the reported precision. This agreement supports the ideal-gas approximation for dilute water vapor under these conditions. Saturation represents equal average rates of evaporation and condensation in equilibrium, rather than a complete halt to molecular exchange.
````

````{admonition} Example Problem: Relative Humidity and Condensation Temperature
**The Model**

Relative humidity compares the actual vapor density with the saturation density at the same temperature. We follow a fixed-volume sample, so its vapor density remains constant during cooling until condensation or ice deposition begins. The table then identifies the saturation temperature. Its below-freezing entries refer to ice, so the final part uses a frost point rather than a liquid-water dew point.

---

**The Math**

**(a)** We denote the actual vapor density by $\rho_{\rm v}$ and the saturation vapor density by $\rho_{\rm sat}(T)$. Their ratio determines the percent relative humidity:

$$
{\rm RH}=\frac{\rho_{\rm v}}{\rho_{\rm sat}(T)}\times100\%.
$$

**(b)** Cooling lowers the saturation density while leaving the actual vapor density unchanged in this model. The condensation temperature $T_{\rm d}$ satisfies

$$
\rho_{\rm sat}(T_{\rm d})=\rho_{\rm v}.
$$

**(c)** At the frost point $T_{\rm f}$, the actual density equals the saturation density over ice. The warm sample's relative humidity with respect to liquid water therefore follows from

$$
{\rm RH}=\frac{\rho_{{\rm sat,ice}}(T_{\rm f})}{\rho_{{\rm sat,water}}(T)}\times100\%.
$$

\begin{align*}
{\rm RH}_{(a)}&=\frac{9.40\ {\rm g/m^3}}{23.1\ {\rm g/m^3}}\times100\%=40.7\%,\\
{\rm RH}_{(c)}&=\frac{2.14\ {\rm g/m^3}}{23.1\ {\rm g/m^3}}\times100\%=9.26\%.
\end{align*}

For part **(b)**, the table's $9.41\ {\rm g/m^3}$ entry at $10^\circ{\rm C}$ is very close to the specified $9.40\ {\rm g/m^3}$. The table therefore supports an estimate of about $10^\circ{\rm C}$, rather than a more precise temperature.

---

**The Conclusion**

The answers are **(a)** $40.7\%$ relative humidity, **(b)** a condensation temperature of about $10^\circ{\rm C}$ in the stated fixed-volume model, and **(c)** $9.26\%$ relative humidity for the sample with the specified frost point. Both warm samples are unsaturated, and the sample with the lower saturation temperature is much drier, as expected. A real atmospheric parcel that cools at nearly constant pressure changes volume, so its dew-point calculation should use vapor pressures rather than treating vapor density as constant.
````
