# Figures and Graphs Primer

Figures are part of physical reasoning. A useful graph, vector diagram, or free-body diagram should make the model easier to understand rather than merely decorate the page.

This primer focuses on how to **read, draw, and explain** common physics figures. It does not require Python or plotting code.

In this course, figures are used to:

- define coordinate directions,
- show vectors and components,
- display measured data,
- compare data with a model,
- interpret slopes and geometric areas,
- and communicate the physical meaning of a result.


## Reading a Graph

Before interpreting the shape of a graph, identify its basic structure.

Ask:

1. What quantity is shown on the horizontal axis?
2. What quantity is shown on the vertical axis?
3. What units are used?
4. Is the scale uniform?
5. Are the marks measured data, a model curve, or both?
6. What physical question is the graph meant to answer?

A graph of position versus time does not directly show the path through space. It shows how one coordinate changes as time passes.

```{admonition} Quick check
:class: tip

On a graph of velocity $v$ versus time $t$, the horizontal axis has units of seconds and the vertical axis has units of meters per second. Always read the axis labels before describing the graph.
```


## Slope and Rate of Change

The slope between two points is

$$
{\rm slope}=\frac{\Delta y}{\Delta x}.
$$

The physical meaning depends on the quantities assigned to the axes.

```{list-table}
:header-rows: 1
:widths: 34 28 38

* - Graph
  - Slope units
  - Physical meaning
* - Position versus time
  - ${\rm m/s}$
  - Velocity
* - Velocity versus time
  - ${\rm m/s^2}$
  - Acceleration
* - Force versus position
  - ${\rm N/m}$
  - How rapidly the force changes with position
```

A steeper line has a larger magnitude of slope. A negative slope means the vertical-axis quantity decreases as the horizontal-axis quantity increases.

For a curved graph, the slope changes from one point to another. In PHYS 1401, you may estimate a local slope by drawing a tangent line or calculate an average slope over a stated interval.

### Mini Exercise

A position–time graph passes through $(t,x)=(1.0\ {\rm s},3.0\ {\rm m})$ and $(4.0\ {\rm s},15.0\ {\rm m})$. Find the average velocity over the interval.

````{admonition} Solution
:class: dropdown

$$
v_{\rm avg}
=\frac{\Delta x}{\Delta t}
=\frac{15.0\ {\rm m}-3.0\ {\rm m}}{4.0\ {\rm s}-1.0\ {\rm s}}
=4.0\ {\rm m/s}.
$$
````


## Geometric Area on a Graph

In several physics graphs, a geometric area represents a physical quantity.

```{list-table}
:header-rows: 1
:widths: 38 28 34

* - Graph
  - Area units
  - Physical meaning
* - Velocity versus time
  - ${\rm m}$
  - Displacement
* - Force versus time
  - ${\rm N\,s}$
  - Impulse
* - Force versus position
  - ${\rm N\,m}={\rm J}$
  - Work
```

For straight-line segments, areas can often be found using rectangles, triangles, and trapezoids. Areas below the horizontal axis count as negative when the vertical-axis quantity has direction or sign.

### Mini Exercise

An object moves at a constant velocity of $6.0\ {\rm m/s}$ for $4.0\ {\rm s}$. What displacement is represented by the area under the velocity–time graph?

````{admonition} Solution
:class: dropdown

The graph forms a rectangle:

$$
\Delta x=(6.0\ {\rm m/s})(4.0\ {\rm s})=24\ {\rm m}.
$$
````


## Data Points and Model Curves

Measured values should normally be shown as individual points. A line or smooth curve may represent:

- a theoretical prediction,
- a best-fit relationship,
- a guide to the eye,
- or a connection between sequential measurements.

Do not assume that a smooth curve is measured more precisely than the points. Ask what the curve represents.

When comparing data with a model, look for:

- the overall trend,
- systematic departures from the model,
- unusually scattered points,
- possible outliers,
- and whether the scale makes the agreement look better or worse than it is.

A strong explanation does more than say, “The graph goes up.” It identifies the quantities and states the relationship:

> As time increases, the position increases more rapidly, so the positive slope becomes steeper and the velocity increases.


## Vectors and Components

A vector diagram should show:

- a labeled coordinate system,
- a clear tail and arrowhead,
- the vector label,
- and components when they are useful.

For a vector $\vec{A}$ with magnitude $A$ at an angle $\theta$ measured counterclockwise from the positive $x$-axis,

$$
A_x=A\cos\theta
$$

and

$$
A_y=A\sin\theta.
$$

Signs come from the direction of the components. Do not rely on the trigonometric formula alone; inspect the diagram.

A vector drawing does not have to be artistic, but it should be large enough to read and consistent with the chosen axes.


## Free-Body Diagrams

A free-body diagram isolates one object or system and shows the external forces acting on it.

A useful free-body diagram should:

1. Identify the object being analyzed.
2. Replace the object with a simple dot or box.
3. Draw one arrow for each external force.
4. Label each force by type or source.
5. Choose coordinate axes that simplify the problem.
6. Avoid including velocity or acceleration arrows as though they were forces.

Common force labels include:

- weight, $\vec{w}=m\vec{g}$,
- normal force, $\vec{N}$,
- tension, $\vec{T}$,
- friction, $\vec{f}$,
- spring force, $\vec{F}_{\rm s}$,
- and an applied force, $\vec{F}_{\rm app}$.

The length of an arrow may suggest relative magnitude, but exact scale is not required unless the problem requests it.


## Figures in Laboratory Work

Laboratory graphs should make the evidence easy to inspect.

Unless your laboratory instructions specify a different format, include:

- axis labels with quantities and units,
- readable numerical scales,
- visible data markers,
- a legend when more than one dataset or model is shown,
- uncertainty bars when they are available and relevant,
- and a caption that explains what the reader should notice.

A caption should identify the plotted quantities and state the main point. It should not merely repeat the axis labels.

Weak caption:

> Position versus time.

Stronger caption:

> Position measured at equal time intervals for a cart released from rest. The upward curvature shows that the cart's speed increased during the run.


## Figure Checklist

Before submitting a figure or diagram, check the following:

- Are the axes or coordinate directions labeled?
- Are units included?
- Is the scale readable and appropriate?
- Are vectors and forces labeled clearly?
- Can the reader distinguish data from a model?
- Does the caption explain the physical point?
- Does the text refer to the figure and interpret it?
- Is the figure large enough to read without excessive zooming?

Figures are evidence. They should make your reasoning easier for another person to follow.
