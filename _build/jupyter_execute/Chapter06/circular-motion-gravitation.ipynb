## Rotation Angle and Angular Velocity

Newton's first law tells us that motion is along a straight line at constant speed unless there is a net external force.  **Uniform circular motion** is the simplest form of curved motion, where the object follows a curved path at constant speed.  Moving in a circular path is also connected to *rotation* or rotational motion.  Pure *translational* motion is motion with no rotation.  Some motion combines both types.

### Rotation Angle

When objects rotate about some axis (e.g., a CD or compact disc), each point in the object follows a circular arc.  Consider a line from the center of the CD to its edge.  CD's are recorded where the pits fall along a curved line that moves through the same angle in the same amount of time.  The **rotation angle** $\Delta \theta$ is the amount of rotation and is analogous to linear distance.  Mathematically, we define the rotation angle as teh ratio of the arclength $\Delta s$ to the radius of curvature $r$:

```{math}
:label: ang_displace_eqn

\Delta \theta = \frac{\Delta s}{r}.
```

```{figure-md} fig-angular-displacement
<img src="rotation_geometry.png" alt="Two-panel figure illustrating angular displacement. Panel a shows a rotating CD with points at different radii sweeping through the same angle delta theta. Panel b shows the geometric relationship between angular displacement, arclength, and radius." width="90%">

**(a)** All points on a rotating CD move through the same angular displacement $\Delta\theta$ during a given time interval, although points farther from the axis travel a greater distance. **(b)** For a point a distance $r$ from the axis of rotation, the angular displacement in radians is related to the arclength $\Delta s$ by $\Delta\theta = \Delta s/r$.
```

We know that for one complete revolution, the arclength is the *circumference* of a circle of radius $r$ and a circle's circumference is $2\pi r$.  For one complete revolution, the rotation angle is

\begin{align}
\Delta \theta = \frac{2\pi r}{r} = 2\pi.
\end{align}

This result is the basis for defining the units used to measure rotation angles $\Delta \theta$ in **radians** (rad).  Then we can relate radians to degrees using 

\begin{align}
2\pi\ {\rm rad} = 360^\circ = 1\ {\rm revolution}.
\end{align}

We can solve the above equation with some algebra to determine the number of degrees in a radian by

\begin{align}
1\ {\rm rad} = \frac{360^\circ}{2\pi} = \frac{180^\circ}{\pi} \approx 57.3^\circ.
\end{align}

From Eqn. {eq}`ang_displace_eqn`, we see that $\Delta s = r\Delta \theta$.  We can substitute this relation into our equation for linear velocity $v$ and substitute for the definition of angular velocity $\omega$ to get

\begin{align}
v = \frac{r\Delta \theta}{\Delta t} = r\omega.
\end{align}

This relationship can be given in two different ways that corresponds to different insights:

1. $v = r\omega$ states that the linear speed is largest at a point on the rim (i.e., largest $r$).  The velocity of a point on the rim is always perpendicular (or tangent) to the radius.  Thus we also call the linear speed as the *tangential speed*.
2. $\omega = v/r$ tells us how fast something rotates given the linear speed and the radius.  Consider the tire of a moving car, where the car moves forward with a velocity $v$ that is equal to the linear velocity of the tire.  The faster the care moves (i.e., large $v$), then the faster the tire spins (i.e., large $\omega$).

Both $\omega$ and $v$ have directions since they both represent velocities.  Angular velocity two directions (e.g., clockwise or counterclockwise), while linear velocity is tangent to the path.  The linear velocity $v$ can be decomposed into vector components $v_x$ and $v_y$ at a given time $t$ given a constant spin rate $\omega$, or

\begin{align}
v_x &= v\cos\left(\omega t\right), \\
v_y &= v\sin\left(\omega t\right).
\end{align}


## Centripetal Acceleration

Recall from kinematics that acceleration is a change in velocity (in its magnitude, direction, or both).  In uniform circular motion, the direction of the velocity constantly changes, so there is an associated acceleration even though the magnitude of the velocity is constant.

```{figure-md} fig-centripetal-acceleration
<img src="https://openstax.org/apps/image-cdn/v1/f=webp/apps/archive/20260604.144757/resources/11e58ada3ac9ddc55f67ebca76ad3f26a96643fb" alt="An object moves along a circular path from point B to point C. Tangential velocity vectors v1 and v2 are shown at the two positions, along with a vector diagram showing that the change in velocity points toward the center of the circle." width="80%">

The velocity of an object moving in a circle changes direction as the object moves along its path. The change in velocity $\Delta \vec{v}$ points toward the center of curvature, so the centripetal acceleration $\vec{a}_c = \Delta \vec{v}/\Delta t$ is also directed toward the center. For a sufficiently small angular displacement $\Delta\theta$, the arc length $\Delta s$ is approximately equal to the chord length $\Delta r$. *Adapted from OpenStax, College Physics 2e, Figure 6.7.*
```

For an object moving in a circular path at constant speed (see Figure {numref}`{number}<fig-centripetal-acceleration>`), the direction of the instantaneous velocity can be determine at two points ($B$ and $C$) along the path.  Acceleration points in the direction of the change in velocity, which is toward the center of rotation in this case.  This pointing is shown with the vector diagram in Fig. {numref}`{number}<fig-centripetal-acceleration>`.  We call the acceleration of an object moving in uniform circular motion the **centripetal acceleration** $a_c$, where *centripetal* means "toward the center" or "center seeking".

The direction of the centripetal acceleration is toward the center, but *what is its magnitude?*

We can use the two triangles(i.e., the triangles $ABC$ and $PQR$), where we find that they are similar *isosceles* triangles with two equal sides.  Uniform speed means that $v_1 = v_2 = v$.  Also with the properties of similar triangles, we have

\begin{align}
\frac{\Delta v}{v} = \frac{\Delta s}{r}.
\end{align}

Recall the definition of acceleration $a = \Delta v/\Delta t$ and our linear velocity $v = \Delta s/\Delta t$, that means we can rearrange the above expression and divide both sides by $\Delta t$ (allowed by our algebra rules) to get

\begin{align*}
\frac{\Delta v}{v} &= \frac{\Delta s}{r}, \qquad &(\text{start here})\\[10pt]
\Delta v &= \frac{v}{r}\Delta s, \qquad &(\text{rearranging first}) \\[10pt]
\frac{\Delta v}{\Delta t} &= \left(\frac{v}{r}\right)\left(\frac{\Delta s}{\Delta t}\right), \qquad &(\text{dividing by } \Delta t) \\[10pt]
a_c &= \frac{v^2}{r}. \qquad & \left(\text{substituting }v = \frac{\Delta s}{\Delta t}\right)
\end{align*}

We see that the magnitude of the centripetal acceleration is given by

```{math}
:label: cent_accel_eqn

a_c = \frac{v^2}{r},
```

which is the acceleration of an object in a circle of radius $r$ at a speed $v$.  This relation tells us the following about centripetal acceleration:

- it is greater at high speeds (i.e., large $v$) and in sharp curves (i.e., small $r$).
- it is proportional to the speed *squared*, which means that it is $4\times$ as hard to take a curve at $2\times$ the speed (e.g., $100\ {\rm kph}$ vs. $50\ {\rm kph}$).

It is also useful to express $a_c$ in terms of angular velocity $\omega$.  We can substitute $v = r\omega$ to find another form as

```{math}
:label: cent_accel_alt_eqn
a_c = \frac{v^2}{r} = \frac{\left(r\omega\right)^2}{r} = r\omega^2.
```

A **centrifuge** is a rotating device used to separate specimens of different densities.  High centripetal acceleration significantly decreases the time it takes for separation to occur and makes separation possible with small samples.  They are used in a variety of applications in science and medicine, including

- the separation of single cell suspensions (e.g., bacteria, viruses, and blood cells) from a liquid medium, and
- the separation of macromolecules (e.g., DNA and protein) from a solution.

Maximum centripetal acceleration of several thousand times the acceleration due to gravity $g$ can be achieved in a vacuum.  Extremely large centrifuges (containing humans) are used to test the tolerance of test pilots and astronauts against the effects of accelerations larger than that of Earth's surface gravity.


## Centripetal Force

Any force or combination of forces can cause a centripetal or radial acceleration (e.g., tension in a rope on a tether ball, Earth's gravity on the Moon, banked roadway's force on a car).  Any net fore causing uniform circular motion is called **centripetal force**.  The direction of a centripetal force is the same as the direction of centripetal acceleration (i.e., toward the center of curvature).

According to Newton's second law, net force is mass times acceleration ($\vec{F}_{\rm net} = m\vec{a}$).  For uniform circular motion, the acceleration is the centripetal acceleration $a_c$.  Therefore, the magnitude of the centripetal force $\vec{F}_c$ is given by

\begin{align}
F_c &= ma_c, \\
&= \frac{mv^2}{r},\\
&= mr\omega^2,
\end{align}

where Eqn. {eq}`cent_accel_eqn` and {eq}`cent_accel_alt_eqn` can both be substituted for $a_c$.  You may use whichever expression for centripetal force is more convenient.

Centripetal force $\vec{F}_c$ is always *perpendicular* to the velocity vector $\vec{v}$ and pointing toward the center of curvature.  If you know the magnitude of the centripetal force $F_c$, the mass $m$, and the speed squared $v^2$, then you can solve for $r$ to get

\begin{align}
r &= \frac{mv^2}{F_c},
\end{align}

using algebra.  This rearrangement of the equation show that for a given mass and velocity, the centripetal force is *inversely* proportional to the radius of curvature $r$.

```{note}
An linear proportion means that as a variable $x$ is increased, there is a corresponding proportional variable $y$ increases too.  This occurs in a similar way for decreases in $x$ that correspond to a decrease in $y$.  

However, an *inversely* proportional quantity shows a different property.  Suppose that $x$ is inversely proportional to $y$. Then as $x$ increases, the magnitude of $y$ *decreases*, and vice versa.
```

In an "ideally banked curve", the banking angle $\theta$ allows you to follow the curve at a certain speed *without* the aid of friction between the tires and the road.  For **ideal banking**, the net external force equals the horizontal centripetal force in the absence of friction (see Figure {numref}`{number}<fig-banked-curve-fbd>`).  The components of the normal force $\vec{N}$ in the horizontal and vertical directions must equal the centripetal force and the weight of the car, respectively.

```{figure-md} fig-banked-curve-fbd
<img src="banked-car-analysis.png" alt="Two-panel diagram of a car traveling around a banked curve. Panel a shows the car on a road banked at angle theta. Panel b shows the free-body diagram, the decomposition of the normal force into vertical and horizontal components, and the corresponding force equations." width="100%">

A car travels around a curve banked at an angle $\theta$. **(a)** The roadway is tilted so that the normal force is no longer vertical. **(b)** The normal force $N$ is resolved into vertical and horizontal components. With no vertical acceleration, $N\cos\theta=w$. The horizontal component points toward the center of the curve and provides the centripetal force, so $N\sin\theta=F_c=mv^2/r$. At the design speed shown here, no friction is required for the car to follow the curve.
```

If the banking angle $\theta$ is ideal for the speed and radius, then the net external force will equal the necessary centripetal force.  the only two external forces acting on teh are are its weight $\vec{w}$ and the normal force of the road $\vec{N}$.  These two forces must add to give a net external force that is horizontal toward the center of curvature and has a magnitude $F_c$.  

We choose a coordinate system, where the normal force is parallel to the banking surface and the weight is straight down (i.e., along the $-y$-axis).  The normal force then has separate components aligned with the $x$- and $y$-axis.  We add the forces along each coordinate direction, where taking the horizontal components we have

\begin{align}
\sum F_x &= N\sin{\theta} = m a_c, \\
&= N\sin{\theta} = \frac{mv^2}{r}.
\end{align}

Because the car does not leave the road's surface, the net vertical force must be zero (i.e., the vertical components must cancel).  We apply the same procedure for adding the vertical forces to get

\begin{align}
\sum F_y &= N\cos{\theta} - mg = 0, \\
&\quad N\cos{\theta} = mg.
\end{align}

We can combine these two results to eliminate $N$ and get an expression for the banking angle $\theta$ in terms of the other physical variables.  We divide the two equations to get

\begin{align}
\frac{N\sin{\theta}}{N\cos{\theta}} &= \frac{mv^2}{mgr}, \\
\tan{\theta} &= \frac{v^2}{rg}.
\end{align}

We can find the angle directly by using the inverse tanget function to get

\begin{align}
\theta &= \tan^{-1}\!{\left(\frac{v^2}{rg}\right)}. \qquad \text{(ideally banked curve, no friction)}
\end{align}

This expression can be understood by considering how the banking angle $\theta$ depends speed $v$ and radius $r$.  A large $\theta$ results for a large $v$ (high speed) and a small $r$ (sharp curve).  Friction helps because it allows you to take the curve at greater of lower speed than if the curve is frictionless.  Note that $\theta$ does not depend on the mass of the vehicle. 

## Newton's Universal Law of Gravitation

## Satellites and Kepler's Laws
