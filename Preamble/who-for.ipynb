# Who Are These Notes For?

This Jupyter Book supports **PHYS 1401: College Physics** at East Texas A&M University.

The course is intended primarily for students whose degree plans require an introductory, algebra- and trigonometry-based physics course rather than the calculus-based sequence for physics and engineering majors. Students enter the course from many programs and with different levels of confidence in mathematics.

These notes assume that **curiosity, preparation, and steady practice matter more than prior experience**. You are not expected to arrive knowing calculus, computer programming, or advanced physics.

---

## How the Course Is Organized

The lecture portion of the course meets three days each week. Class meetings include explanations, demonstrations, worked examples, and selected opportunities for individual or small-group problem solving.

The course is not taught entirely in a studio format. Most class meetings are organized around lecture and guided examples, but you should still be prepared to draw diagrams, estimate quantities, use a calculator, and explain your reasoning.

The laboratory meets as a separate section. Lab activities give you opportunities to make measurements, organize data, interpret graphs, consider uncertainty, and compare experimental results with physical models.

---

## What These Notes Are—and Are Not

This book is not meant to replace lecture, laboratory, the assigned textbook, or your own problem solving.

Instead, it is designed as:

- a **guided companion** to lecture,
- a source of worked examples and practice problems,
- a place to review diagrams, graphs, and physical models,
- and a collection of figures, videos, and study resources that support the course.

The goal is not to memorize a list of formulas. The goal is to learn how to recognize the relevant physical principle, represent the situation clearly, carry out the mathematics, and decide whether the result makes sense.

## How to Read Worked Examples

Most worked examples follow a consistent structure:

- **The Problem** — What information is given, and what must be determined?
- **The Model** — What physical principles, diagrams, and assumptions describe the situation?
- **The Math** — How can algebra, trigonometry, graphs, or proportional reasoning be used to find the result?
- **The Conclusion** — What does the answer mean physically, and is it reasonable?

When studying an example:

1. Read the problem and identify the known and unknown quantities.
2. Study the model before focusing on the equations.
3. Try to predict the form or approximate size of the answer.
4. Follow the algebra and keep units attached to numerical values.
5. Read the conclusion and check whether the result is physically sensible.

A worked example is most useful when you pause and attempt each step before revealing or reading the completed solution.

## Mathematical Background

This course uses algebra, geometry, trigonometry, vectors, scientific notation, and graph interpretation. Calculus is not required.

You should be prepared to practice:

- rearranging equations,
- converting units,
- using sine, cosine, and tangent,
- resolving vectors into components,
- interpreting slopes and areas on graphs,
- and reporting answers with units and appropriate significant figures.

These skills will be reviewed as they appear, but they become easier only through regular use.

---

## Connecting Lecture and Laboratory

Lecture and laboratory emphasize different parts of the same scientific process.

In lecture, you will usually begin with an idealized model and use it to predict what should happen. In laboratory, you will make measurements and decide how well the evidence agrees with the model.

A useful question to carry between the two parts of the course is:

> What assumptions made the lecture model manageable, and how closely does the laboratory situation satisfy those assumptions?
