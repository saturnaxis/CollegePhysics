## What is a Fluid?

Matter most commonly exists as a solid, liquid, gas, or plasma, which are the common *phases of matter*.

- Solids have a definite shape, a specific volume, and strong chemical (molecular) bonds holding the structure together.
- Liquids have a definite volume, but their bonds are weaker so that their shape changes depending on the contained they are held.
- Gases have neither a definite shape nor a specific volume, which means the bonds are either extremely weak or nonexistent.
- Plasmas are similar to gases, but they are ionized which means that electric currents can freely flow within them.

Liquids, gases, and plasmas behave like a fluid because they yield to shearing forces, whereas solids resist them.  The fluids we are accustomed to are characterized by their *viscosity*, which measures the extent to which the fluids yield to shearing (i.e., how easily and quickly they flow).

Atoms in *solids* are in close contact, which allows the atoms to vibrate but not to change positions with neighboring atoms.  As a result, we can think of the molecular bonds like springs that can be stretched or compressed, but not easily broken.  

- A solid resists all types of stress (see [Elasticity: Stress and Strain](https://saturnaxis.github.io/CollegePhysics/Chapter05/friction-drag-elasticity.html#elasticity-stress-and-strain)).  
- A solid cannot be easily deformed because the bonds prevent the atoms from moving freely.  
- Solids resist compression because their atoms form part of a lattice structure in which the atoms remain a relatively fixed distance apart.  Under compression the atoms would be forced into one another

Liquids deform easily when stressed and do not spring back into their original shape once the force is removed because the atoms are free to slide about and change neighbors.  They flow with the molecules held together by their mutual attraction.  When a liquid is placed in a container, it takes the shape of the container where the atoms resist compression (like solids).

Atoms in gases and plasmas are separated by distances that are very large compared to the size of the individual particles.  The forces between the particles are therefore very weak (except when they collide with one another). Gases and plasmas flow easily and are relatively easy to compress because there is so much space (and little force) between the particles.  In an open container, gases and plasmas will escape because they have so much energy.

Due to their general properties of flow, we treat liquids, gases, and plasmas simply as fluids, where the distinction lies in their viscosity.

## Density

*Which weighs more: a tone of feathers or a ton of bricks?*

This old riddle plays with the distinction between mass, weight, and density.  A ton of something weighs the same no matter its composition.  Bricks have a greater density than feathers, so we are tempted to think of them as heavier.

**Density** is crucial in determining whether an object sinks or floats in a fluid.

```{admonition} Density
:class: note

Density $\rho$ is mass $m$ per unit volume $V$, or:

\begin{align}
\rho = \frac{m}{V},
\end{align}

where the volume $V$ is usually determined by an assumed shape of the matter (e.g., cube, sphere, cylinder, etc.).
```

In the riddle regarding the feathers and bricks, the weights are the same but the volume occupied by the feathers is much greater resulting in a lower average density.  The $\rm SI$ unit of density is $\rm kg/m^3$, which was used to derive the first units in the metric system.  The basic mass unit (the kilogram) was defined by the mass of $1\ {\rm L}$ of water, where $1\ L$ is a volume of $1000\ {\rm mL}$, $1000\ {\rm cm^3}$, or $1000\ {\rm cc}$.

The densities in the following table are given in ${\rm g/cm^3}$. Multiply these values by $1000$ to express them in SI units of ${\rm kg/m^3}$.

```{list-table} Densities of Various Substances
:name: tab-densities-various-substances
:header-rows: 1
:widths: 22 12 21 12 21 12
:align: center

* - **Substance**
  - **Density** (${\rm g/cm^3}$)
  - **Substance**
  - **Density** (${\rm g/cm^3}$)
  - **Substance**
  - **Density** (${\rm g/cm^3}$)
* - **Solids**
  -
  - **Liquids**
  -
  - **Gases**
  -
* - Aluminum
  - $2.7$
  - Water ($4^\circ{\rm C}$)
  - $1.000$
  - Air
  - $1.29\times10^{-3}$
* - Brass
  - $8.44$
  - Blood
  - $1.05$
  - Carbon dioxide
  - $1.98\times10^{-3}$
* - Copper (average)
  - $8.8$
  - Sea water
  - $1.025$
  - Carbon monoxide
  - $1.25\times10^{-3}$
* - Gold
  - $19.32$
  - Mercury
  - $13.6$
  - Hydrogen
  - $0.090\times10^{-3}$
* - Iron or steel
  - $7.8$
  - Ethyl alcohol
  - $0.79$
  - Helium
  - $0.18\times10^{-3}$
* - Lead
  - $11.3$
  - Gasoline
  - $0.68$
  - Methane
  - $0.72\times10^{-3}$
* - Polystyrene
  - $0.10$
  - Glycerin
  - $1.26$
  - Nitrogen
  - $1.25\times10^{-3}$
* - Tungsten
  - $19.30$
  - Olive oil
  - $0.92$
  - Nitrous oxide
  - $1.98\times10^{-3}$
* - Uranium
  - $18.70$
  -
  -
  - Oxygen
  - $1.43\times10^{-3}$
* - Concrete
  - $2.30$–$3.0$
  -
  -
  - Steam ($100^\circ{\rm C}$)
  - $0.60\times10^{-3}$
* - Cork
  - $0.24$
  -
  -
  -
  -
* - Glass, common (average)
  - $2.6$
  -
  -
  -
  -
* - Granite
  - $2.7$
  -
  -
  -
  -
* - Earth's crust
  - $3.3$
  -
  -
  -
  -
* - Wood
  - $0.3$–$0.9$
  -
  -
  -
  -
* - Ice ($0^\circ{\rm C}$)
  - $0.917$
  -
  -
  -
  -
* - Bone
  - $1.7$–$2.0$
  -
  -
  -
  -
* - Silver
  - $10.49$
  -
  -
  -
  -
```

## Pressure

```{admonition} Pressure
:class: note

Pressure is defined as the force divided by the area perpendicular to the force over which the force is applied, or

\begin{align}
P = \frac{F_\perp}{A},
\end{align}

where the force $F_\perp$ is perpendicular to the area $A$.
```

A given force can have a significantly different effect depending on the area over which it is exerted.  The $\rm SI$ unit for pressure is the *pascal*, where $1\ {\rm Pa} = 1\ {\rm N/m^2}$.  

In addition to the pascal, there are many other units for pressure that are in common use.

```{list-table} Common Units of Pressure (Referenced to Pascals)
:header-rows: 1
:name: pressure-units
:widths: 40 60

* - Unit
  - Equivalent in Pascals

* - Pascal (Pa)
  - $1\ {\rm Pa} = 1\ {\rm N/m^2}$

* - pound per square inch (psi)
  - $1\ {\rm psi} = 6895\ {\rm Pa}$

* - atmosphere (atm)
  - $1\ {\rm atm} = 1.013\times10^{5}\ {\rm Pa}$

* - bar
  - $1\ {\rm bar} = 10^{5}\ {\rm Pa}$

* - torr (mm Hg)
  - $1\ {\rm torr} = 133.3\ {\rm Pa}$

* - inches of mercury (in Hg)
  - $1\ {\rm in\ Hg} = 3386\ {\rm Pa}$

* - millibar (mbar)
  - $1\ {\rm mbar} = 100\ {\rm Pa}$
```

The force exerted on the end of a tank is perpendicular to its inside surface, which is a property of a static, or stationary, fluid.  Fluids cannot withstand *shearing* (or sideways) forces, which means that they cannot *exert* shearing forces either.

Fluid pressure has no direction, which means it is a scalar quantity.  The forces due to pressure have well-defined directions (i.e., they are always exerted perpendicular to any surface).  Pressure is exerted on all surfaces.  Swimmers and tires feel pressure on all sides.

## Variation of Pressure with Depth in a Fluid

At the Earth's surface, the air pressure exerted on you is a result of the weight of air above you.  This pressure is reduced as you climb in altitude and the weight of the air above you decreases.  Under water, the pressure exerted on you increases as you descend into deeper waters.  In this case, the pressure exerted on you is a result of both the weight of water and that of the atmosphere above you.

You many notice an air pressure change on an elevator ride in a very tall building, but you need only dive a meter (or so) below the surface of a pool to feel a pressure increase.  The difference is that water is much denser than air (about $775\times$ as dense).

Consider a container of water, where its bottom supports the weight of the water in it.  We can calculate the pressure exerted on the bottom by the weight of the water.  The weight of water is determined as the mass of the water $m$ times the acceleration due to gravity $g$, or $W= mg$.  This is the force exerted that is perpendicular to the area $A$ at the base of the container.  Then the pressure is calculated directly as

\begin{align}
P = \frac{F_\perp}{A} = \frac{mg}{A}.
\end{align}

We can find the mass of the water from its volume and density:

\begin{align}
m = \rho V, 
\end{align}

where the volume $V$ is the column of fluid above the surface.  Thus, it is natural to use the volume of a cylinder, or

\begin{align}
V = Ah,
\end{align}

where $A$ is the area of a circle and $h$ is the depth (height below the surface of the liquid).  Combining these expressions, we find that the mass is given by

\begin{align}
m = \rho Ah.
\end{align}

We can then substitute this mass into our expression for the weight of the fluid and thereby into the equation for the pressure, or

\begin{align}
P = \frac{(\rho A h) g}{A} = \rho g h,
\end{align}

where the area $A$ canceled.  This is the pressure due to *the weight of a fluid*.  This equation for pressure is generally applicable to static fluids.  For liquids, this equation holds to great depths because fluids are nearly incompressible.  For gases, one can apply this equation as long as the density changes are small over the depth consider because they are quite compressible.

*Atmospheric pressure* is an example of pressure due to the weight of a fluid (i.e., air) above a given height.  The atmospheric pressure at the Earth's surface varies a little due to the large-scale flow of the atmosphere, which creates regions of high and low pressure for weather.  The average pressure ate sea level is given by the *standard atmospheric pressure* $P_{\rm atm}$, which is given as

\begin{align}
1\ \text{atmosphere (atm)} &= P_{\rm atm}, \\
P_{\rm atm} &= 1.01 \times 10^5\ {\rm N/m^2} = 101\ {\rm kPa}.
\end{align}

On average (at sea level) a column of air above 1-${\rm m^2}$ of Earth's surface has a weight of $101\ {\rm kN}$, or $1.01\times 10^5\ {\rm N}$.

- *What is the total pressure at a depth of $10.3\ {\rm m}$ in a swimming pool?*  
- *Does the atmospheric pressure on the water's surface affect the pressure below?*

Yes.  Both the water's weight and the atmosphere's weight must be supported.  So the *total pressure* at a depth of $10.3\ {\rm m}$ is 

\begin{align*}
P &= \rho_wgh_w + P_{\rm atm}, \\
&= (1000\ {\rm kg/m^3})(9.81\ {\rm m/s^2})(10.3\ {\rm m}) + 1.01 \times 10^5\ {\rm N/m^2} = 2.02 \times 10^5\ {\rm N/m^2} = \boxed{2\ {\rm atm}}.
\end{align*}

## Pascal's Principle

*Can pressure be increased in a fluid by pushing directly on the fluid?*

Yes, but it is much easier if the fluid is enclosed.  For example, the heart increases blood pressure by pushing directly on the blood in an enclosed system (e.g., valves closed in a chamber).  If you try to push on a fluid in an open system (e.g., a river), the fluid flows away. An enclosed fluid cannot flow away, so pressure is more easily increased by an applied force.

*What happens to a pressure in an enclosed fluid?*

Atoms in a fluid are free to move about (and rotate), they transmit the pressure to all parts of the fluid and to the walls of the container.  The pressure is transmitted *undiminished*, or $P_{\rm in} = P_{\rm out}$.  This phenomenon is called **Pascal's principle** because it was first clearly stated by the French philosopher [Blaise Pascal](https://en.wikipedia.org/wiki/Blaise_Pascal).

```{admonition} Pascal's Principle
:class: note

A change in pressure applied to an enclosed fluid is transmitted undiminished to all portions of the fluid and to the walls of the container.  Mathematically, this is represented by 

\begin{align}
P_{\rm in} = P_{\rm out}.
\end{align}
```

Pascal's principle is important in fluids since a change in pressure is transmitted undiminished in an enclosed fluid.  It implies that the *total pressure in a fluid is the sum of the pressures from different sources*, meaning that pressures add.

### Application of Pascal's Principle

Pascal's principle is found in a hydraulic system, which is an enclosed system used to exert forces.  The most common hydraulic system are those that operate car brakes.

Consider two fluid-filled pistons where each piston has an area $A_1$ and $A_2$ at the top.  The two pistons are at the same height so that there will be no difference in pressure due to a difference in depth.  The pressure due to $F_1$ acts on the area $A_1$ of the first piston and is defined as

$$ P_1 = \frac{F_1}{A_1}. $$

According to Pascal's principle, this pressure is transmitted undiminished throughout the fluid and to the walls of the container.  Thus, a pressure $P_2$ is felt at the other pistion that is equal to $P_1$, and defined by

$$ P_2 = \frac{F_2}{A_2}. $$

Since $P_1 = P_2$, we  can see that

$$ \frac{F_1}{A_1} = \frac{F_2}{A_2}. $$

This equation relates the ratios of force to area in any hydraulic system, providing the pistons are at the same vertical height and that friction in the system is negligible.

Hydraulic systems can increase of decrease the force applied to them.  To make the force larger, the pressure is applied to a larger area.  For example, if a $100$-$\rm N$ force is applied to the smaller cylinder and the larger one has an area $5\times$ greater, then the force out is $500\ {\rm N}$.  More explicitly, we can show

$$ F_2 = \frac{A_2}{A_1}F_1 = \left(\frac{5A_1}{A_1}\right)(100\ {\rm N}) = 500\ {\rm N}. $$

Hydraulic systems are analogous to simple levers, but they have the advantage that pressure can be sent through complicated, curved lines to several places at once.

A simple hydraulic system (e.g., a simple machine) can increase the force but cannot do more work than is done on it.  Work is force times distance moved, where the output cylinder usually moves very little despite exerting a greater force.  Many hydraulic systems have a motorized pump that actually does most of the work in the system.  The movement of a spider's leg is achieved partly by hydraulics, where a jumping spider can create a force that launches it to heights that are $25\times$ the leg's length.

## Gauge Pressure, Absolute Pressure, and Pressure Measurement

When you have a flat tire, the tire gauge reads nearly zero when you begin to fill it.  If there were a hole in your tire, the gauge would read zero even though atmospheric pressure exists in the tire.  Tire gauges are designed to read zero at atmospheric pressure and positive when the pressure is greater than atmospheric.

Atmospheric pressure adds to blood pressure in every part of the circulatory system.  But atmospheric pressure has not effect on blood flow since it adds to the pressure coming out of the heart and going back into it too. Blood pressure measurements are made relative to atmospheric pressure like tire pressure.

It is very common for pressure gauges to ignore atmospheric pressure, where they read zero at atmospheric pressure.  We define *gauge pressure* as the pressure relative to atmospheric pressure.

```{admonition} Gauge vs Absolute Pressure
:class: note

**Gauge pressure** $P_{\rm G}$ is the pressure relative to atmospheric pressure.  Gauge pressure is positive for pressures above atmospheric pressures and negative for pressures below it.

**Absolute pressure**, or total pressure, $P_{\rm abs} is the sum of gauge and atmospheric pressure. Mathematically, we can relate absolute pressure to gauge pressure by

\begin{align}
P_{\rm abs} = P_{\rm G} + P_{\rm atm}.
\end{align}

For example, if your tire gauge reads $34\ {\rm psi}$ (pounds per square inch), then the absolute pressure is $34\ {\rm psi} + 14.7\ {\rm psi} = 48.7\ {\rm psi}$ and is equivalent to  $336\ {\rm kPa}$.
```

In most cases, the absolute pressure in fluids cannot be negative.  Fluids push rather than pull, so the smallest absolute pressure is zero. The smallest possible gauge pressure is $P_{\rm G} = -P_{\rm atm}$ so that $P_{\rm abs} = 0$.  There is not theoretical limit to how large a gauge pressure can be.

Pressure can be measured using tire gauges as well as blood pressure cuffs.  Pascal's principle prescribes for the undiminished transmission of pressure through a fluid, which allows for the precise remote sensing of pressures.  Remote sensing is more convenient than putting a measuring device into a person's artery.

Mechanical pressure gauges measure pressure through the force transmitted to a secondary device (e.g., flexible bellows).  A U-shaped tube is called a *manometer*, where it has both sides of the tube are open to the atmosphere.  Atmospheric pressure pushes down on each side so its effect cancels.  If the fluid is deeper on one side, there is a greater pressure on that side.  The fluid flows away from the deeper side until the depths are equal.

Suppose one side of the manometer is connect to some source of absolute pressure (e.g., a toy ballon or a vacuum-packed peanut jar).  Pressure is transmitted undiminished to the manometer, and the fluid levels are no longer equal.  In both cases, the absolute pressure differs from the atmospheric pressure by an amount $h\rho g$, where $\rho$ is the density of the fluid in the manometer.

- In the case of the toy ballon, the absolute pressure is greater than atmospheric pressure ($P_{\rm abs}>P_{\rm atm}$).  The manometer can support a column of fluid with a height $h$ and so it must exert a pressure $h\rho g$ greater than atmospheric pressure ($P_{\rm G}>0$).
- For the peanut jar, the absolute pressure is less than atmospheric pressure ($P_{\rm abs}<P_{\rm atm}$).  The atmospheric pressure can support a column of fluid with a height $h$ and so $P_{\rm abs}<P_{\rm atm}$ by an amount $h\rho g$, which means that the gauge pressure is negative ($P_{\rm G}<0$).

Mercury manometers are often used to measure arterial blood pressure.  An inflatable cuff is placed on the upper arm.  By squeezing the bulb, the person making the measurement exerts pressure, which is transmitted undiminished to both the main artery in the arm and the manometer.  When this applied pressure exceeds blood pressure, blood flow below the cuff is cut off.  The person making the measurement then slowly lowers the applied pressure and listens for blood flow to resume.  Blood pressure pulsates because of the pumping action of the heart reaching a maximum (**systolic pressure**) and a minimum (**diastolic pressure**) with each heartbeat.

Systolic pressure is measured by noting the value of $h$ when blood flow first begins as the cuff pressure is lowered.  Diastolic pressure is measured by noting $h$ when blood flows without interruption.  The typical blood pressure of a young adult raises the mercury to a height of $120\ {\rm mm}$ at systolic and $80\ {\rm mm}$ at diastolic, where this is commonly quoted as $120/80$.  

The first pressure represents the maximum output of the heart, while the second is due to the elasticity of the arteries in maintaining the pressure between beats.  The density of the mercury fluid in the manometer is $13.6\times$ greater than water, so the height of the fluid will be $1/13.6$ of that in a water manometer.  

A *barometer* is a device that measures atmospheric pressure.  This device measures atmospheric pressure instead of gauge pressure because there is a nearly pure vacuum above the mercury in the tube.  The height of the mercury is such that $h\rho g = P_{\rm atm}$.  When atmospheric pressure varies, the mercury rises or falls, which can give important clues to weather forecasters.

## Archimedes' Principle

*Why do some things float while others do not?* 

This is partly answered by the fact that pressure increases with depth in a fluid.  This means that the upward force on the bottom of an object (in a fluid) is greater than the downward force on the top of the object.  There is a net upward, or **buoyant force**, on any object in any fluid.  

- If the buoyant force is grater than the object's weight, then the object will rise to the surface and float.
- If the buoyant force is less than the object's weight, the object will sink.
- If the buoyant force equals the object's weight, the object will remain suspended at that depth. 

The buoyant force is always present whether the object floats, sinks, or is suspended in a fluid.

Consider what happens when a submerge object is removed from a fluid.  The space occupied is filled by fluid having a weight $w_{\rm fl}$.  This weight is supported by the surrounding fluid so that the buoyant force must equal $w_{\rm fl}$, the weight of the fluid displaced by the object.  

The Greek mathematician and inventor [Archimedes](https://en.wikipedia.org/wiki/Archimedes) stated this principle long before concepts of force were established.  The king wanted to know whether a wreath he commissioned was pure gold or simply covered in gold and wanted a process that would not  destroy the wreath in the process.  Archimedes discovered that the fluid displaced was equal to the weight of the wreath.  From there, he could also measure the volume of the fluid displaced and determine the submerged wreath's density.  Since gold is more dense than iron, this was an ingenious way to determine the authenticity of the golden wreath.

```{admonition} Archimedes' Principle
:class: note

According to this principle, the buoyant force on an object equals the weight of the fluid it displaces.  In equation form, Archimedes' principle is 

\begin{align}
F_{\rm B} = w_{\rm fl} = m_{\rm fl}g,
\end{align}

where $F_{\rm B}$ is the buoyant force and $w_{\rm fl}$ is the weight of the fluid displaced by the object.
```

### Density and Archimedes' Principle
Density plays a critical role in Archimedes' principle.  the average density of an object is what ultimately determines whether it floats.  If its average density is less than that of the surrounding fluid, it will float.

This is because the fluid (having higher density) contains more mass and hence more weight in the same volume.  The buoyant force is thus greater than the weight of the object.  Likewise, an object denser than the fluid will sink.

The extent to which a floating object is submerged depends on how the object's density is related to that of the fluid.  For example, an unloaded ship has a lower density and less of it is submerged compared to the same ship fully loaded.  The fraction submerged is the ratio of the volume submerged to the volume of the object, or

\begin{align}
\text{fraction submerged} = \frac{V_{\rm sub}}{V_{\rm obj}} = \frac{V_{\rm fl}}{V_{\rm obj}}.
\end{align}

The volume submerged $V_{\rm sub}$ equals the volume of the fluid displaced $V_{\rm fl}$.  We can obtain the relationship between the densities by substituting $\rho = m/V$ into the the equation, which gives

\begin{align}
\frac{V_{\rm fl}}{V_{\rm obj}} = \left(\frac{m_{\rm fl}}{\rho_{\rm fl}}\right) \left(\frac{\bar{\rho}_{\rm obj}}{m_{\rm obj}}\right),
\end{align}

where $\bar{\rho}_{\rm obj}$ is the average density of the object and $\rho_{\rm fl}$ is the density of the fluid.  Since the object floats, its mass and that of the displaced fluid are equal ($m_{\rm fl}=m_{\rm obj}$). This results in

\begin{align}
\text{fraction submerged} = \frac{\bar{\rho}_{\rm obj}}{\rho_{\rm fl}}.
\end{align}

We use this last relationship to measure densities.  We can measure the fraction of a floating object that is submerged (e.g., with a hydrometer).  If is now useful to define ratio of the average density of an object $\bar{\rho}$ to a fluid (usually water) as **specific gravity**:

\begin{align}
\text{specific gravity} = \frac{\bar{\rho}}{\rho_{\rm w}},
\end{align}

where $\bar{\rho}$ is th average density of the object and $\rho_{\rm w}$ is the density of water at $4.0^\circ{\rm C}$. Specific gravity is dimensionless.

- If an object floats, its specific gravity is less than one.
- If it sinks, its specific gravity is greater than one.

The fraction of a floating object that is submerged equals its specific gravity.  If an object's specific gravity is exactly 1, then it will remain suspended in the fluid.

Scuba divers try to obtain this state so they can hover in the water.  We measure the specific gravity of fluids (e.g., battery acid, radiator fluid, and urine) as an indicator of their condition.

### More Density Measurements

An object is weighted in air and then weighed again while submerged in a liquid.  The density of the object (here a coin) can be calculated if the fluid density is known.  This same technique can also be used to determine the density of the fluid if the density of the object is known.  All of these calculations are based on Archimedes' principle.

Archimedes' principle means that the object appears to weigh less when submerged.  We call this measurement: the object's *apparent weight*.  the object suffers an apparent *weight loss* equal to the weight of the fluid displaced.  Under constant gravity, a balance can measure the mass, where the object suffers an *apparent mass loss* equal to the mass of the fluid displaced:

\begin{align}
\text{apparent weight loss} &= \text{weight of fluid displaced}, \\
\text{apparent mass loss} &= \text{mass of fluid displaced}.
\end{align}

## Surface Tension and Capillary Action

Attractive forces between molecules of the same type are called **cohesive forces**.  Liquids can be held in open containers because cohesive forces hold the molecules together.  Attractive forces between molecules of different types are called **adhesive forces**.

### Surface Tension

Cohesive forces between molecules cause the surface of a liquid to contract to the *smallest possible surface area*.  This general effect is called **surface tension**.  Molecules on the surface are pulled inward by cohesive forces, reducing the surface area.  Molecules inside the liquid experience zero net force, since they have neighbors on all sides.

We can explain surface tension effects using a model of a liquid surface acting like a stretched elastic sheet.  For example, some insects can walk on water as we would walk on a trampoline.  

A needle can rest on a water surface.  The iron needle cannot float because its density is greater than that of water.  Its weight is supported by forces in the stretched surface that try to make the surfaces smaller or flatter.  If the needle were placed point down on the surface, its weight action on a smaller area would break the surface, and it would sink.

Surface tension is proportional to the strength of the cohesive force, which varies with the type of liquid.  Surface tension $\gamma$ is defined as the force $F$ per unit length $L$ exerted by a stretched liquid membrane:

\begin{align}
\gamma = \frac{F}{L}.
\end{align}

```{list-table} Surface Tension of Some Liquids
:name: tab-surface-tension-liquids
:header-rows: 1
:widths: 60 40
:align: center

* - **Liquid**
  - **Surface tension** $\gamma$ (${\rm N/m}$)
* - Water at $0^\circ{\rm C}$
  - $0.0756$
* - Water at $20^\circ{\rm C}$
  - $0.0728$
* - Water at $100^\circ{\rm C}$
  - $0.0589$
* - Soapy water (typical)
  - $0.0370$
* - Ethyl alcohol
  - $0.0223$
* - Glycerin
  - $0.0631$
* - Mercury
  - $0.465$
* - Olive oil
  - $0.032$
* - Tissue fluids (typical)
  - $0.050$
* - Blood, whole at $37^\circ{\rm C}$
  - $0.058$
* - Blood plasma at $37^\circ{\rm C}$
  - $0.073$
* - Gold at $1070^\circ{\rm C}$
  - $1.000$
* - Oxygen at $-193^\circ{\rm C}$
  - $0.0157$
* - Helium at $-269^\circ{\rm C}$
  - $0.00012$
```

For an insect on water, its weight $w$ is supported by the upward components of the surface tension force: $w = \gamma L \sin{\theta}$, where $L$ is the circumference of the insect's foot in contact with the water.

Surface tension is the reason why liquids form bubbles and droplets.  The inward surface tension force causes bubbles to be approximately spherical and raises the pressure of the gas trapped inside relative to atmospheric pressure outside.  The gauge pressure $P$ inside a spherical bubble is given as

\begin{align}
P = \frac{4\gamma}{r},
\end{align}

where $r$ is the radius of the bubble.  The pressure inside the bubble is greatest when the bubble is smallest.

Our lungs contain hundreds of millions of mucus-lined sacs called *alveoli*, which are very similar in size and about $0.1\ {\rm mm}$ in diameter. You can exhale without muscle action by allowing surface tension to contract these sacs.  Medical patients whose breathing is aided by a respirator have air blown into the lungs, but are generally allowed to exhale on their own.

Even if there is paralysis, surface tension in the alveoli will expel air from the lungs.  Since pressure increases as the radii of th alveoli decrease, an occasional deep cleansing breath is needed to fully reinflate the alveoli.

The tension in the walls of the alveoli results from the membrane tissue and a liquid on the walls of the alveoli containing a long lipoprotein that acts as a surfactant (i.e., surface-tension reducing substance).  The need for the surfactant results from the tendency of small alveoli to collapse and the air to fill into the larger alveoli making them even larger.

- During inhalation, the lipoprotein molecules are pulled apart and the wall tension increases as the radius increases (i.e., increased surface tension).  
- During exhalation, the molecules slide back together and the surface tension decreases, helping to prevent a collapse of the alveoli.

The surfactant serves to change the wall tension so that hte small alveoli don't collapse and large alveoli are prevented from expanding too much.  The tension change is a unique property of the surfactants, and is not shared by detergents, which simply lower surface tension.

If water gets into the lungs, the surface tension is too great and you cannot inhale.  This is a problem in resuscitating drowning victims and newborn infants who are bore without this surfactant.  This condition is known as *hyaline membrane disease* and is a leading cause of death for infants.  

Emphysema produces the opposite problem with alveoli.  Alveolar walls of emphysema patients deteriorate, and the sacs combine to form larger sacs.  Because pressure produced by surface tension decreases with increasing radius, these larger sacs produce smaller pressure, reducing their ability to exhale.  A common test for emphysema is to measure the pressure and volume of air that can be exhaled.

### Adhesion and Capillary Action

*Why is it that water beads up on a waxed car but does not on bare paint?*

Adhesive forces between wax and water are much smaller than those between water and paint.  Competition between forces of adhesion and cohesion are important in the macroscopic behavior of liquids.  The roles of these two forces depends on the the angle $\theta$ between the tangent to the liquid surface and the surface.  The **contact angle** $\theta$ is directly related to the relative strength of the cohesive and adhesive forces.  \

- The larger the strength of the cohesive force relative to the adhesive force (i.e., the larger $\theta$ is) and the more the liquid tends to form a droplet.
- The smaller the relative strength (i.e., the smaller $\theta$ is) so that the adhesive force is able to flatten the drop.

```{list-table} Contact Angles of Various Substances
:name: tab-contact-angles
:header-rows: 1
:widths: 60 40
:align: center

* - **Interface**
  - **Contact Angle** $\theta$
* - Mercury-glass
  - $140^\circ$
* - Water-glass
  - $0^\circ$
* - Water-paraffin (wax)
  - $107^\circ$
* - Water-silver
  - $90^\circ$
* - Organic liquids (most)-glass
  - $0^\circ$
* - Ethyl alcohol-glass
  - $0^\circ$
* - Kerosene-glass
  - $26^\circ$
```

*Source: OpenStax, College Physics 2e, Table 11.4.*

The relative strength of cohesive and adhesive forces is **capillary action**, which is the tendency of a fluid to be raised or suppressed in a narrow tube, or *capillary tube*.  This action causes blood to be drawn into a small-diameter tube when the tube touches a drop.

If a capillary tube is placed vertically into a liquid, capillary action will raise or suppress the liquid inside the tube depending on the combination of substances.  The actual effect depends on the relative strength of the cohesive and adhesive forces (i.e., the contact angle).

- If $\theta <90^\circ$, then the fluid will be raised.
- If $\theta > 90^\circ$ then it will be suppressed.

For example, mercury has a very large surface tension and a large contact angle with glass.  When it is placed in a tube, the surface of a column of mercury curves downward, somewhat like a drop.  The curved surface of a fluid in a tube is called a **meniscus**.  The tendency of surface tension is always to reduce the surface area.  Surface tension thus flattens the curved liquid surface in a capillary tube.  This results in a downward force in mercury and an upward force in water.

Capillary action can move liquids horizontally over very large distances, but the height to which it can raise (or suppress) a liquid is limited by its weight.  This is shown mathematically b

\begin{align}
h = \frac{2\gamma}{\rho g r}\cos{\theta}.
\end{align}

There are some insights we can gain from examining this expression.

- The height $h$ is directly proportional to the surface tension $\gamma$, which is its direct cause.  
- The height is inversely proportional to the tube radius. The smaller the radius $r$, the higher the fluid can be raised since a smaller tube holds less mass.
- The height is also inversely proportional to the fluid density $\rho$, since a larger density means a greater mass in the same volume.

*How does sap get to the tops of tall trees?*

It appears that it is pulled up like a chain held together by cohesive forces.  As each molecule of sap enters a leaf and evaporates, the entire chain is pulled up a notch.  A negative pressure created by water evaporation must be present to pull the sap up through the xylem vessels.  In most situations, *fluids can push but can exert a negligible pull* because the cohesive forces seem to be too small to hold the molecules tightly together.  In this case, the cohesive force of water molecules provides a very strong pull.

## Pressures in the Body

Measuring blood pressure is the most common of all medical examinations after taking a person's temperature and weight.  The pressures in various parts of the body can be measured and often provide valuable medical indicators.

```{list-table} Typical Pressures in Humans
:header-rows: 1
:widths: 60 40

* - **Body system**
  - **Gauge pressure in mm Hg**
* - Blood pressures in large arteries (resting)
  - 
* - *Maximum (systolic)*
  - 100–140
* - *Minimum (diastolic)*
  - 60–90
* - Blood pressure in large veins
  - 4–15
* - Eye
  - 12–24
* - Brain and spinal fluid (lying down)
  - 5–12
* - Bladder
  - 
* - *While filling*
  - 0–25
* - *When full*
  - 100–150
* - Chest cavity between lungs and ribs
  - −8 to −4
* - Inside lungs
  - −2 to +3
* - Digestive tract
  - 
* - *Esophagus*
  - −2
* - *Stomach*
  - 0–20
* - *Intestines*
  - 10–20
* - Middle ear
  - <1
```

###  Blood Pressure
Common arterial blood pressure measurements produce values of $120\ {\rm mm\ Hg}$ and $80\ {\rm mm\ Hg}$ for systolic and diastolic pressures, respectively.  Both pressures have health implications.  

When systolic pressure is chronically high, there is a increased risk of stroke and heart attack.  On the other hand, low systolic pressure can increase the risk of fainting.  Systolic pressure increases dramatically during exercise to increase blood flow and returns to normal afterward.  This change in pressure may be beneficial to the tone of the circulatory system rather than causing any ill effects.

Diastolic pressure can be an indicator of fluid balance.  When it is low, this may indicate that a person is hemorrhaging internally and needs a transfusion.  Conversely, high diastolic pressure indicates a ballooning of the blood vessels, which may be due tot he transfusion of too much fluid into the circulatory system.  High diastolic pressure is also an indication that blood vessels are dilating properly to pass blood through.  This can seriously strain the heart in its attempt to pump blood.

Blood leaves the heart a about $120\ {\rm mm\ Hg}$, but its pressure continues to decrease as it goes from the aorta to smaller arteries and veins.  The pressure differences in the circulation system are caused by blood flow through the system as well as position of the person.  For a person standing up, the pressure in the feet will be larger than at the heart due to the weight of the blood ($P = \rho g h$).  If we assume that the distance between the heart and the feet of a person is about $1.4\ {\rm m}$ when standing upright, then the increase in pressure in the feet is given by

\begin{align}
\Delta P &= \rho g \Delta h, \\
&= \left(1050\ {\rm kg/m^3}\right)(9.81\ {\rm m/s^2})(1.4\ {\rm m}) = 1.4\ times 10^4\ {\rm Pa} = 108\ {\rm mm\ Hg}.
\end{align}

Standing a long time can lead to an accumulation of blood in the legs and swelling.  This is the reason why soldiers who stand still for a long time have been known to faint.  Elastic bandages around the calf can help prevent this accumulation anc can also help provide increased pressure to enable veins to send blood back up to the heart.

Blood pressure may also be measured in major veins, the heart chambers, arteries to the brain, and the lungs.  But these pressures are usually only monitored during surgery or for patients in intensive care since the measurements are invasive.  Health workers thread thin tubes called **catheters** into appropriate locations to transmit pressures to external monitoring devices.

The heart consists of two pumps: (1) the right side forcing blood through the lungs and (2) the left causing blood to flow through the rest of the body.  

- Right-heart failure results in a rise in the pressure in the vena cavae and a drop in pressure in the arteries to the lungs.  
- Left-heart failure results in a rise in the pressure entering the left side of the heart and a drop in aortal pressure.

### Pressure in the Eye
The shape of the eye is maintained by fluid pressure called **intraocular pressure**, which is normally in the range of $12-24\ {\rm mm\ Hg}$.  When the circulation of fluid in the eye is blocked, it can lead to a buildup in pressure which is a medical condition called **glaucoma**.  The net pressure can become as great as $85\ {\rm mm\ Hg}$, which can permanently damage the optic nerve.

Suppose the back of the eye has an area of $6.0\ {\rm cm^2}$ and the net pressure is $85\ {\rm mm\ Hg}$, then we can calculate force ($F = PA$).  First, we must use the unit conversion where $1\ {\rm m^2} = 10^4\ {\rm cm^2}$.  The we calculate as follows:

\begin{align}
F &= PA = (\rho g h)A, \\
&= (1.36 \times 10^4\ {\rm kg/m^3})(9.81\ {\rm m/s^2})(8.5\ times 10^{-2}\ {\rm m})(6.0 \times 10^{-4}\ {\rm m^2}) = 6.8\ {\rm N}.
\end{align}

This force is the weight of about $0.68\ {\rm kg}$ ($1.5\ {\rm lb}$) resting on the eye.  This is sufficient to cause it damage, where a normal force would be the less than $1/4$ of our initial value.

People over 40 years old are at greatest risk of developing glaucoma and should have their intraocular pressure tested routinely.  Most measurements involve exerting a force on the (anesthetized) eye over som area (i.e., a pressure) and observing the eye's response.  A noncontact approach uses a puff of air and a measurement is made of the force needed to indent the eye.  If the intraocular pressure is high, the eye will deform less and rebound more vigorously than normal.

### Pressure in the Lungs

The pressure inside the lungs increases and decreases with each breath.  The pressure drops to below atmospheric pressure (*negative gauge pressure*) when you inhale, causing air to flow into the lungs.  It increases above atmospheric pressure (*positive gauge pressure*) when you exhale, forcing air out.

Lung pressure is controlled by several mechanisms:

1. Muscle action int the diaphragm and rib cage is necessary for inhalation.  This muscle action increases the volume of the lungs thereby reducing the pressure within them.
2. Surface tension in the alveoli creates a positive pressure opposing inhalation.

You can exhale without muscle action by letting surface tension in the alveoli create its own positive pressure.  Muscle action can add to this positive pressure to produce a forced exhalation (e.g., blowing up a balloon, blowing out a candle, or coughing).

The lungs would collapse due to the surface tension in the alveoli, if they were not attached to the inside of the chest wall by liquid adhesion.  The gauge pressure in the liquid attaching the lungs to the inside of the chest wall is *negative*, ranging from ${-4}\text{ to }{-8}\ {\rm mm\ Hg}$, during exhalation and inhalation, respectively.  If air is allowed to enter the chest cavity, it breaks the attachment, and either lung (or both lungs) may collapse.  Suction is applied to the chest cavity of surgery patients and trauma victims to reestablish negative pressure and inflate the lungs.

### Spinal Column and Skull
Normally, there is a $5-12\ {\rm mm\ Hg}$ pressure in the fluid surrounding the brain and filling the spinal column.  This cerebrospinal fluid supplies flotation to the brain, among other purposes.  The buoyant force supplied by the fluid nearly equals the weight of the brain since their densities are nearly equal.  If there is a loss of fluid, the brain rests on the inside of the skull, which can cause severe headaches, constricted blood flow, and serious damage.  Spinal fluid pressure is measured by means of a needle inserted between vertebrae that transmits the pressure to a suitable measuring device.

### Bladder Pressure

This bodily pressure climbs steadily from zero to about $25\ {\rm mm\ Hg}$ as the bladder fills to its normal capacity of $0.5\ {\rm L}$.  This pressure triggers the *micturition reflex*, which stimulates the feeling of needing to urinate.  It also causes muscles around the bladder to contract, raising the pressure to over $100\ {\rm mm\ Hg}$.  Coughing, straining, tensing in cold weather, wearing tight clothes, and experiencing nervous tension all can increase bladder pressure and trigger this reflex.  Bladder pressure can be measured by a catheter or by inserting a needle through the bladder wall and transmitting the pressure to an appropriate measuring device.

### Pressures in the Skeletal System

Skeletal pressures are the largest in the body, due to the high values of initial force and the small areas to which this force is applied (e.g., in the joints).  For example, when a person lifts an object improperly, a force of $5000\ {\rm N}$ can be created between the vertebrae in the spine, which may be applied to an area as small as $10\ {\rm cm^2}$.  The pressure created is

$$ P = \frac{F}{A} = \frac{5000\ {\rm N}}{10^{-3}\ {\rm m^2}} = 5.0 \times 10^5\ {\rm N/m^2}, $$

or about $50\ {\rm atm}$.

This pressure can damage both the spinal discs (i.e., cartilage between the vertebrae) as well as the bony vertebrae themselves.  Even under normal circumstances, forces between vertebrae in the spine are large enough to create pressures of several atmospheres.  Most causes of excessive pressure in the skeletal system can be avoided by lifting properly and avoiding extreme physical activity.

There are many medically significant pressures in the body:

- Pressure caused by various muscle actions drives food and waste through the digestive system.
- Stomach pressure behaves much like bladder pressure and is tied to the sensation of hunger.  Positive pressure in the stomach may force acid into the esophagus, causing "heartburn."
- Pressure in the relaxed esophagus is normally negative because pressure in the chest cavity is normally negative.
- Pressure in the middle ear can result in a significant force on the eardrum if it differs greatly from atmospheric pressure (e.g., while scuba diving).  The decrease in external pressure is noticeable during plane flights.  The Eustachian tubes connect the middle ear to the throat and allow us to equalize pressure in the middle ear to avoid an imbalance of force on the ear drum.
